```python
# ============================================================
# DELTA_MAP "publication" sampler (ATTN+MLP, early/mid/late) — FIXED for paper use
#
# What changed vs your cell (all fixes implemented):
#   ✅ FP32 gradients (no AMP) to avoid tail distortion by bf16/fp16 quantization.
#   ✅ Rolling differences: d_t = |g_t - g_{t-1}| (uses ALL batches, not only pairs).
#   ✅ True reservoir sampling (Vitter-style) AFTER initial fill:
#        - unbiased sample from the whole stream of deltas
#        - replaces your approximate "random overwrite" refresh
#   ✅ Multiple coordinate seeds (N_COORD_SEEDS) to reduce coordinate-selection bias:
#        - each group stores a mixture of seeds (still one final buffer per group)
#   ✅ Memory/throughput sane defaults (still "publication-grade"):
#        - SAMPLES_PER_GROUP reduced to 5M (recommended); change if you need.
#
# Progress:
#   - tqdm postfix: filled/min/avg/max/sec + reservoir replace rate
#   - periodic detailed prints
#
# Saves:
#   RUN_DIR/figures_tempered/delta_map_iterXXXXXXX.npz
#   RUN_DIR/figures_tempered/delta_map_iterXXXXXXX.meta.json
# ============================================================

import os, sys, json, time
from pathlib import Path
from contextlib import nullcontext

import numpy as np
import torch

try:
    from tqdm import tqdm
except Exception:
    tqdm = None

# -----------------------------
# CONFIG
# -----------------------------
RUN_DIR = "out/E4_clean_2k_adamw"
ITER = 800

K_BATCHES = 1024
SEED = 123

# coordinate sampling
COORD_BUDGET_PER_GROUP = 600_000
N_COORD_SEEDS = 3              # <- NEW: mix multiple coordinate selections (2..4 typical)

# how many deltas to keep per group (CPU float32)
SAMPLES_PER_GROUP = 5_000_000  # <- NEW: was 20M; 5M is usually enough for CCDF tails

FIG_DIRNAME = "figures_tempered"
OVERWRITE = True
COMPRESS = True

UPDATE_EVERY = 8
PRINT_EVERY  = 64
# -----------------------------

sys.path.insert(0, str(Path.cwd()))

if not torch.cuda.is_available():
    raise RuntimeError(
        "CUDA не доступна в этом окружении (torch.cuda.is_available() = False).\n"
        "Нужно запустить в CUDA-окружении PyTorch."
    )

DEVICE = "cuda"
DTYPE  = torch.float32     # <- FIX: FP32 gradients for tail correctness

# speed flags
torch.backends.cuda.matmul.allow_tf32 = True
torch.set_float32_matmul_precision("high")
torch.backends.cudnn.benchmark = True

torch.manual_seed(SEED)
np.random.seed(SEED)
rng = np.random.default_rng(SEED)

run_dir = Path(RUN_DIR).resolve()
assert run_dir.exists(), f"RUN_DIR not found: {run_dir}"

fig_dir = run_dir / FIG_DIRNAME
fig_dir.mkdir(parents=True, exist_ok=True)

npz_path  = fig_dir / f"delta_map_iter{ITER:07d}.npz"
meta_path = fig_dir / f"delta_map_iter{ITER:07d}.meta.json"

print("[info] RUN_DIR =", run_dir)
print("[info] ITER    =", ITER)
print("[info] DEVICE  =", DEVICE, "| DTYPE =", DTYPE)
print("[info] saving  =", npz_path)

if npz_path.exists() and not OVERWRITE:
    print("[ok] exists and OVERWRITE=False -> loading")
    z = np.load(npz_path, allow_pickle=False)
    delta_map = {k: np.asarray(z[k]) for k in z.files}
    print("[ok] loaded groups:", len(delta_map))
    delta_map

else:
    from model import GPT, GPTConfig

    cfg = json.load(open(run_dir / "config_resolved.json", "r", encoding="utf-8"))
    n_layer = int(cfg.get("n_layer", 12))
    n_head  = int(cfg.get("n_head", 12))
    n_embd  = int(cfg.get("n_embd", 768))
    block_size = int(cfg.get("block_size", 1024))
    bias = bool(cfg.get("bias", False))
    dropout = float(cfg.get("dropout", 0.0))
    batch_size = int(cfg.get("batch_size", 12))

    data_dir = Path(cfg.get("data_dir", "data/openwebtext"))
    data_dir = data_dir if data_dir.is_absolute() else (Path.cwd() / data_dir).resolve()

    model = GPT(GPTConfig(
        block_size=block_size,
        vocab_size=50304,
        n_layer=n_layer,
        n_head=n_head,
        n_embd=n_embd,
        dropout=dropout,
        bias=bias,
    )).to(DEVICE)

    ckpt_path = run_dir / "checkpoints" / f"ckpt_iter{ITER:07d}.pt"
    assert ckpt_path.exists(), f"Checkpoint not found: {ckpt_path}"
    ckpt = torch.load(str(ckpt_path), map_location=DEVICE)
    model.load_state_dict(ckpt["model"], strict=True)

    # <- FIX: force fp32 weights and fp32 backward
    model = model.to(dtype=torch.float32)
    model.train()

    name2param = dict(model.named_parameters())
    print(f"[ok] model loaded. n_layer={n_layer} batch={batch_size} block={block_size} bias={bias}")

    # -----------------------------
    # Keep memmap open once
    # -----------------------------
    train_bin = data_dir / "train.bin"
    assert train_bin.exists(), f"train.bin not found: {train_bin}"
    train_data = np.memmap(train_bin, dtype=np.uint16, mode="r")
    train_len = int(train_data.shape[0])
    assert train_len > block_size + 2

    def get_batch_fast(bs: int, T: int, device: str):
        ix = torch.randint(train_len - T - 1, (bs,))
        x = torch.stack([torch.from_numpy(train_data[i:i+T].astype(np.int64, copy=False)) for i in ix])
        y = torch.stack([torch.from_numpy(train_data[i+1:i+1+T].astype(np.int64, copy=False)) for i in ix])
        x = x.pin_memory().to(device, non_blocking=True)
        y = y.pin_memory().to(device, non_blocking=True)
        return x, y

    # -----------------------------
    # Phases
    # -----------------------------
    b0 = n_layer // 3
    b1 = 2 * n_layer // 3
    PHASE_LAYERS = {
        "early": list(range(0, b0)),
        "mid":   list(range(b0, b1)),
        "late":  list(range(b1, n_layer)),
    }
    print("[info] phase layers:", PHASE_LAYERS)

    # -----------------------------
    # Samplers (GPU indices) — now per-seed
    # -----------------------------
    def sample_indices(numel: int, m: int, rng_local: np.random.Generator) -> np.ndarray:
        m = min(m, numel)
        return rng_local.choice(numel, size=m, replace=False).astype(np.int64)

    def to_idx_t(idx_np: np.ndarray) -> torch.Tensor:
        return torch.from_numpy(idx_np).to(device=DEVICE, dtype=torch.long)

    def build_attn_qkv_samplers(layers, block: str, total_samples: int, rng_local):
        assert block in ("q","k","v")
        samplers = []
        per_layer = max(1, total_samples // max(1, len(layers)))
        for li in layers:
            pname = f"transformer.h.{li}.attn.c_attn.weight"
            p = name2param[pname]
            numel = int(p.numel())
            block_numel = numel // 3
            offset = {"q":0,"k":1,"v":2}[block] * block_numel
            idx = sample_indices(block_numel, per_layer, rng_local) + offset
            samplers.append((pname, to_idx_t(idx)))
        return samplers

    def build_attn_proj_samplers(layers, total_samples: int, rng_local):
        samplers = []
        per_layer = max(1, total_samples // max(1, len(layers)))
        for li in layers:
            pname = f"transformer.h.{li}.attn.c_proj.weight"
            p = name2param[pname]
            idx = sample_indices(int(p.numel()), per_layer, rng_local)
            samplers.append((pname, to_idx_t(idx)))
        return samplers

    def build_mlp_fc_samplers(layers, total_samples: int, rng_local):
        samplers = []
        per_layer = max(1, total_samples // max(1, len(layers)))
        for li in layers:
            pname = f"transformer.h.{li}.mlp.c_fc.weight"
            p = name2param[pname]
            idx = sample_indices(int(p.numel()), per_layer, rng_local)
            samplers.append((pname, to_idx_t(idx)))
        return samplers

    def build_mlp_proj_samplers(layers, total_samples: int, rng_local):
        samplers = []
        per_layer = max(1, total_samples // max(1, len(layers)))
        for li in layers:
            pname = f"transformer.h.{li}.mlp.c_proj.weight"
            p = name2param[pname]
            idx = sample_indices(int(p.numel()), per_layer, rng_local)
            samplers.append((pname, to_idx_t(idx)))
        return samplers

    # group_samplers[gname] = list_of_seeds; each seed is list[(pname, idx_t)]
    group_samplers = {}

    for phase, layers in PHASE_LAYERS.items():
        group_samplers[f"attn_{phase}_q"]    = []
        group_samplers[f"attn_{phase}_k"]    = []
        group_samplers[f"attn_{phase}_v"]    = []
        group_samplers[f"attn_{phase}_proj"] = []
        group_samplers[f"mlp_{phase}_fc"]    = []
        group_samplers[f"mlp_{phase}_proj"]  = []

        for s in range(N_COORD_SEEDS):
            rng_s = np.random.default_rng(SEED + 10_000*s + hash(phase) % 10_000)
            group_samplers[f"attn_{phase}_q"].append(build_attn_qkv_samplers(layers, "q", COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"attn_{phase}_k"].append(build_attn_qkv_samplers(layers, "k", COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"attn_{phase}_v"].append(build_attn_qkv_samplers(layers, "v", COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"attn_{phase}_proj"].append(build_attn_proj_samplers(layers, COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"mlp_{phase}_fc"].append(build_mlp_fc_samplers(layers, COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"mlp_{phase}_proj"].append(build_mlp_proj_samplers(layers, COORD_BUDGET_PER_GROUP, rng_s))

    def ordered_groups():
        out = []
        for phase in ["early","mid","late"]:
            for comp in ["q","k","v","proj"]:
                out.append(f"attn_{phase}_{comp}")
            for comp in ["fc","proj"]:
                out.append(f"mlp_{phase}_{comp}")
        return out

    GROUPS = ordered_groups()
    print("[info] groups:", GROUPS)
    print("[info] N_COORD_SEEDS:", N_COORD_SEEDS)

    used_param_names = sorted({pname for seedlist in group_samplers.values()
                              for samps in seedlist
                              for (pname, _) in samps})

    # -----------------------------
    # Buffers (CPU) + true reservoir state
    # -----------------------------
    out_buf = {g: np.empty((SAMPLES_PER_GROUP,), dtype=np.float32) for g in GROUPS}
    out_pos = {g: 0 for g in GROUPS}

    # reservoir counters: how many deltas have been seen *after* first "prev" exists
    seen = {g: 0 for g in GROUPS}
    replaced = {g: 0 for g in GROUPS}

    # rolling prev per (group, seed): store vec from previous batch (GPU tensor)
    prev = {g: [None]*N_COORD_SEEDS for g in GROUPS}

    # no AMP
    amp_ctx = lambda: nullcontext()

    def _fill_stats():
        fills = np.array([out_pos[g] / SAMPLES_PER_GROUP for g in GROUPS], dtype=np.float64)
        filled_cnt = int(np.sum(fills >= 0.999))
        return fills, filled_cnt

    if tqdm is None:
        it = range(K_BATCHES)
    else:
        it = tqdm(range(K_BATCHES), total=K_BATCHES, desc="batches", mininterval=0.5, dynamic_ncols=True)

    t0 = time.time()
    for k in it:
        model.zero_grad(set_to_none=True)

        X, Y = get_batch_fast(batch_size, block_size, DEVICE)
        with amp_ctx():
            _, loss = model(X, Y)
        loss.backward()

        # flat grads cache once (GPU fp32)
        flat_cache = {}
        for pname in used_param_names:
            gg = name2param[pname].grad
            flat_cache[pname] = None if gg is None else gg.detach().flatten()  # fp32

        # iterate all groups and all coord-seeds
        for gname in GROUPS:
            seed_samplers_list = group_samplers[gname]

            for sidx, samplers in enumerate(seed_samplers_list):
                # build vec on GPU fp32
                chunks = []
                for pname, idx_t in samplers:
                    flat = flat_cache[pname]
                    if flat is None:
                        chunks.append(torch.zeros((idx_t.numel(),), device=DEVICE, dtype=torch.float32))
                    else:
                        chunks.append(flat.index_select(0, idx_t))
                vec = torch.cat(chunks, dim=0)

                if prev[gname][sidx] is None:
                    prev[gname][sidx] = vec
                    continue

                # FIX: rolling difference uses every batch
                d = (prev[gname][sidx] - vec).abs_()
                prev[gname][sidx] = vec

                # stream length update
                seen[gname] += int(d.numel())

                # move to CPU float32 once
                d_cpu = d.detach().to(dtype=torch.float32).cpu().numpy()

                # Fill then unbiased reservoir
                pos = out_pos[gname]
                if pos < SAMPLES_PER_GROUP:
                    take = min(SAMPLES_PER_GROUP - pos, d_cpu.size)
                    if take > 0:
                        # If d is larger than remaining space, sample uniformly from d
                        if take < d_cpu.size:
                            jj = rng.integers(0, d_cpu.size, size=take, endpoint=False)
                            out_buf[gname][pos:pos+take] = d_cpu[jj]
                        else:
                            out_buf[gname][pos:pos+take] = d_cpu[:take]
                        out_pos[gname] = pos + take

                    # If still leftover and buffer just became full, fallthrough to reservoir for the rest
                    if out_pos[gname] < SAMPLES_PER_GROUP:
                        continue
                    d_cpu = d_cpu[take:] if take < d_cpu.size else np.empty((0,), dtype=np.float32)

                # Reservoir sampling for remaining points (unbiased)
                # For each new item i (1-indexed), keep with prob S/i; if kept, replace random slot
                if d_cpu.size:
                    # current stream index starts at (seen - len(d_cpu)) + 1
                    base = seen[gname] - d_cpu.size
                    # iterate in vectorized-ish blocks
                    # We'll do per-element Bernoulli keep with prob S/(base+i)
                    i = np.arange(1, d_cpu.size + 1, dtype=np.int64)
                    prob = SAMPLES_PER_GROUP / (base + i)
                    prob = np.clip(prob, 0.0, 1.0)

                    # decide keeps
                    u = rng.random(d_cpu.size)
                    keep_mask = (u < prob)
                    nk = int(np.sum(keep_mask))
                    if nk > 0:
                        # choose replacement indices uniformly in buffer
                        repl_idx = rng.integers(0, SAMPLES_PER_GROUP, size=nk, endpoint=False)
                        out_buf[gname][repl_idx] = d_cpu[keep_mask]
                        replaced[gname] += nk

        # tqdm postfix
        if tqdm is not None and (k % UPDATE_EVERY == 0 or k == K_BATCHES-1):
            fills, filled_cnt = _fill_stats()
            minfill = float(fills.min()*100)
            avgfill = float(fills.mean()*100)
            maxfill = float(fills.max()*100)
            avg_it = (time.time()-t0)/max(1,(k+1))
            avg_repl = float(np.mean(list(replaced.values())))
            it.set_postfix(
                filled=f"{filled_cnt}/{len(fills)}",
                min=f"{minfill:4.1f}%",
                avg=f"{avgfill:4.1f}%",
                max=f"{maxfill:4.1f}%",
                sec=f"{avg_it:4.2f}",
                repl=f"{avg_repl/1e6:4.2f}M"
            )

        if (k + 1) % PRINT_EVERY == 0:
            fills, filled_cnt = _fill_stats()
            avg_repl = float(np.mean(list(replaced.values())))
            avg_seen = float(np.mean(list(seen.values())))
            print("\n[detail]")
            print(f"  batch {k+1}/{K_BATCHES}")
            print(f"  groups filled: {filled_cnt}/{len(fills)}")
            print(f"  fill % (min/avg/max): {fills.min()*100:.1f} / {fills.mean()*100:.1f} / {fills.max()*100:.1f}")
            print(f"  avg reservoir replacements per group: {avg_repl/1e6:.2f}M")
            print(f"  avg deltas seen per group: {avg_seen/1e6:.2f}M")

    # finalize delta_map (full buffers preferred)
    delta_map = {g: (out_buf[g].copy() if out_pos[g] >= SAMPLES_PER_GROUP else out_buf[g][:out_pos[g]].copy())
                 for g in GROUPS}

    print("\n[done] collected delta_map.")
    for g in GROUPS[:8]:
        print(f"  {g:16s} N={delta_map[g].size:,} replaced={replaced[g]:,} seen={seen[g]:,}")

    # save
    print("\n[saving] writing NPZ...")
    tS = time.time()
    if COMPRESS:
        np.savez_compressed(str(npz_path), **delta_map)
    else:
        np.savez(str(npz_path), **delta_map)
    print(f"[saving] NPZ done in {time.time()-tS:.2f}s -> {npz_path}")

    meta = dict(
        run_dir=str(run_dir),
        iter=int(ITER),
        device=str(DEVICE),
        dtype=str(DTYPE),
        k_batches=int(K_BATCHES),
        seed=int(SEED),
        n_coord_seeds=int(N_COORD_SEEDS),
        coord_budget_per_group=int(COORD_BUDGET_PER_GROUP),
        samples_per_group=int(SAMPLES_PER_GROUP),
        replaced=replaced,
        seen=seen,
        groups=GROUPS,
        npz=str(npz_path),
        note="FP32 grads, rolling deltas, unbiased reservoir sampling, multiple coord seeds"
    )
    with open(meta_path, "w", encoding="utf-8") as f:
        json.dump(meta, f, ensure_ascii=False, indent=2)

    print("[saved] npz :", npz_path)
    print("[saved] meta:", meta_path)

    delta_map
```

```python
# ============================================================
# UPGRADED: stable CCDF + robust *beta-tempered* fit
# with ENDPOINT/BALANCE improvements (better on both ends),
# while keeping the SAME visualization (3 figures) as your original cell.
#
# Changes vs your last beta-tempered cell:
#   ✅ Tail-aware weights: combine (tail_count^p) + (ccdf^-cc_power) + explicit right/left boosts
#   ✅ Endpoint penalties in the score (log-domain mismatch at BOTH ends of fit window)
#   ✅ Extra "far-right" penalty using last K CCDF points (very important when few points)
#   ✅ Beta grid + powerlaw fallback (powerlaw = lambda=inf)
#   ✅ Still anchored at left edge of fit window (keeps body alignment), but now
#      the scoring forces the right edge to match too.
#
# Output visuals remain:
#   1) CCDF (log-log) + fit
#   2) check plot (linear in u=(x/lambda)^beta)
#   3) B_eff on fit window
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

# --- knobs (feel free to tweak) ---
Q_LO, Q_HI = 0.001, 0.99998
NBINS_CCDF = 800
NBINS_FIT  = 180
MIN_FIT_BINS = 60
BEFF_WINDOW = 21

# Window search space
CCDF_MAX_CANDIDATES = (0.2, 0.1, 0.05, 0.02, 0.01, 0.005, 0.002, 0.001)
MIN_TAIL_COUNT_CANDIDATES = (300, 500, 1000, 2000, 5000, 10000, 20000)

# Beta grid for stretched cutoff (beta-tempered)
BETA_GRID = np.unique(np.concatenate([np.linspace(0.2, 1.0, 17), np.array([0.15, 0.12, 0.10])]))
CUTOFF_EPS_EFFECT = 0.02  # if (xmax/lambda)^beta < this -> cutoff invisible, treat as powerlaw

# Weights: tail_count^p (keeps mid-tail informative) + CCDF emphasis (fix right end)
WEIGHT_TC_POWER   = 0.65      # similar to your old WEIGHT_POWER, but a bit smaller
WEIGHT_CC_POWER   = 0.30      # emphasize small CCDF (right tail)
WEIGHT_CLIP_MAX   = 200_000.0

# Explicit boosts for ends
LEFT_K,  LEFT_BOOST  = 6,  2.0
RIGHT_K, RIGHT_BOOST = 18, 80.0    # increase RIGHT_BOOST if right end still drifts

# Endpoint penalties (log-domain abs error)
ENDPOINT_PENALTY = 0.12            # 0.08..0.20 typical
FARRIGHT_PENALTY = 0.18            # extra emphasis for the very last points
FARRIGHT_K       = 8               # last K points in the whole CCDF curve
EPS = 1e-300

# -----------------------------
# CCDF utilities
# -----------------------------
def ccdf_logbins(x, q_lo=Q_LO, q_hi=Q_HI, nbins=NBINS_CCDF):
    x = np.asarray(x, dtype=np.float64)
    x = x[np.isfinite(x)]
    x = x[x > 0]
    x.sort()
    n = x.size
    if n < 10:
        return np.array([]), np.array([]), np.array([]), np.array([])

    xmin = float(np.quantile(x, q_lo))
    xmax = float(np.quantile(x, q_hi))
    if not np.isfinite(xmin) or not np.isfinite(xmax) or xmax <= xmin:
        return np.array([]), np.array([]), np.array([]), np.array([])

    edges = np.logspace(np.log10(xmin), np.log10(xmax), nbins + 1)
    g = edges[1:]
    idx = np.searchsorted(x, g, side="right")
    cc = (n - idx) / n
    cc = np.clip(cc, 1.0/(n+1.0), 1.0)
    tc = n * cc

    keep = np.r_[True, np.diff(idx) != 0]
    return g[keep], cc[keep], tc[keep], idx[keep]

def aggregate_by_logx_bins(x, y, tc, cc=None, nbins=NBINS_FIT):
    """
    Aggregate points into uniform bins in log(x):
      - x_bin: geometric mean
      - y_bin: weighted mean
      - w_bin: sum of weights
    weights use both tail_count and ccdf emphasis.
    """
    x = np.asarray(x, dtype=np.float64)
    y = np.asarray(y, dtype=np.float64)
    tc = np.asarray(tc, dtype=np.float64)
    if cc is None:
        cc = np.clip(tc / max(1.0, float(np.max(tc))), EPS, 1.0)
    cc = np.asarray(cc, dtype=np.float64)

    lx = np.log(x)
    lo, hi = float(lx.min()), float(lx.max())
    if not np.isfinite(lo) or not np.isfinite(hi) or hi <= lo:
        return x, y, np.ones_like(y)

    edges = np.linspace(lo, hi, nbins + 1)
    b = np.searchsorted(edges, lx, side="right") - 1
    b = np.clip(b, 0, nbins - 1)

    xs, ys, ws = [], [], []
    for bi in range(nbins):
        m = (b == bi)
        if not np.any(m):
            continue
        x_m = x[m]
        y_m = y[m]
        tc_m = tc[m]
        cc_m = cc[m]

        w = (np.clip(tc_m, 1.0, WEIGHT_CLIP_MAX) ** WEIGHT_TC_POWER) * (np.clip(cc_m, EPS, 1.0) ** (-WEIGHT_CC_POWER))
        wsum = float(np.sum(w))
        if wsum <= 0:
            continue

        x_bin = float(np.exp(np.sum(w * np.log(x_m)) / wsum))
        y_bin = float(np.sum(w * y_m) / wsum)

        xs.append(x_bin); ys.append(y_bin); ws.append(wsum)

    return np.asarray(xs), np.asarray(ys), np.asarray(ws)

# -----------------------------
# Regression helpers
# -----------------------------
def weighted_lstsq(A, y, w):
    w = np.asarray(w, dtype=np.float64)
    w = np.clip(w, 1e-12, None)
    sw = np.sqrt(w)
    beta, *_ = np.linalg.lstsq(A * sw[:, None], y * sw, rcond=None)
    return beta

def weighted_r2(y, yhat, w):
    w = np.asarray(w, dtype=np.float64)
    w = np.clip(w, 1e-12, None)
    y = np.asarray(y, dtype=np.float64)
    yhat = np.asarray(yhat, dtype=np.float64)
    ybar = np.sum(w * y) / np.sum(w)
    ss_res = np.sum(w * (y - yhat) ** 2)
    ss_tot = np.sum(w * (y - ybar) ** 2)
    return 1 - ss_res / ss_tot if ss_tot > 0 else np.nan

# -----------------------------
# Diagnostics
# -----------------------------
def beff_from_ccdf(g, cc, window=BEFF_WINDOW):
    logx = np.log(g)
    logy = np.log(np.clip(cc, EPS, 1.0))
    w = window // 2
    slopes = np.full_like(logx, np.nan, dtype=np.float64)
    for i in range(len(logx)):
        j0 = max(0, i - w)
        j1 = min(len(logx), i + w + 1)
        Xw = logx[j0:j1]; Yw = logy[j0:j1]
        if len(Xw) < 2:
            continue
        Xc = Xw - Xw.mean()
        denom = np.sum(Xc * Xc)
        if denom <= 0:
            continue
        slopes[i] = np.sum(Xc * (Yw - Yw.mean())) / denom
    return -slopes

# -----------------------------
# Model
# CCDF(x) = exp(c0) * x^{-alpha} * exp(-(x/lambda)^beta)
# powerlaw is lambda=inf
# -----------------------------
def model_ccdf(x, alpha, lam, beta, c0):
    x = np.asarray(x, dtype=np.float64)
    if np.isfinite(lam) and lam > 0:
        y = np.exp(c0) * (x ** (-alpha)) * np.exp(- (x / lam) ** beta)
    else:
        y = np.exp(c0) * (x ** (-alpha))
    return np.minimum(y, 1.0)

def beta_check_fit(x, y_check, w, lam, beta):
    # y_check = log(CCDF)+alpha log x = c0 - (x/lambda)^beta
    if not (np.isfinite(lam) and lam > 0):
        A = np.stack([np.ones_like(x)], axis=1)
        coef = weighted_lstsq(A, y_check, w)
        yhat = A @ coef
        r2 = weighted_r2(y_check, yhat, w)
        return float(r2), float(np.nan), yhat

    u = (x / lam) ** beta
    A = np.stack([np.ones_like(u), u], axis=1)
    coef = weighted_lstsq(A, y_check, w)
    yhat = A @ coef
    r2 = weighted_r2(y_check, yhat, w)
    slope = float(coef[1])
    return float(r2), slope, yhat

def endpoint_logerr(g_fit, cc_fit, alpha, lam, beta, c0):
    # log-domain abs error at both ends of fit window
    iL = int(np.argmin(g_fit))
    iR = int(np.argmax(g_fit))
    xL, yL = float(g_fit[iL]), float(cc_fit[iL])
    xR, yR = float(g_fit[iR]), float(cc_fit[iR])
    mL = float(model_ccdf(np.array([xL]), alpha, lam, beta, c0)[0])
    mR = float(model_ccdf(np.array([xR]), alpha, lam, beta, c0)[0])
    yL = max(yL, EPS); yR = max(yR, EPS); mL = max(mL, EPS); mR = max(mR, EPS)
    eL = abs(np.log(mL / yL))
    eR = abs(np.log(mR / yR))
    return float(eL), float(eR)

def far_right_err(g, cc, alpha, lam, beta, c0, k=FARRIGHT_K):
    # extra penalty for the last K points of the whole CCDF curve
    k = int(min(k, len(g)))
    if k <= 2:
        return 0.0
    xR = g[-k:]
    yR = np.clip(cc[-k:], EPS, 1.0)
    mR = np.clip(model_ccdf(xR, alpha, lam, beta, c0), EPS, 1.0)
    return float(np.mean(np.abs(np.log(mR) - np.log(yR))))

def make_tail_weights_for_fit(cc_fit, tc_fit):
    # weights for aggregated fit points (works via aggregate_by_logx_bins too)
    w = (np.clip(tc_fit, 1.0, WEIGHT_CLIP_MAX) ** WEIGHT_TC_POWER) * (np.clip(cc_fit, EPS, 1.0) ** (-WEIGHT_CC_POWER))
    # explicit boosts: left and right edge inside the fit window
    n = w.size
    if n > 0:
        w[:min(LEFT_K, n)] *= LEFT_BOOST
        w[max(0, n-RIGHT_K):] *= RIGHT_BOOST
    med = np.median(w)
    if np.isfinite(med) and med > 0:
        w /= med
    return np.clip(w, 1e-12, 1e9)

# -----------------------------
# Fit: beta-tempered (anchored) + endpoint penalties
# -----------------------------
def fit_beta_tempered_auto_anchored(g, cc, tc):
    best = None

    for ccdf_max in CCDF_MAX_CANDIDATES:
        for min_tail in MIN_TAIL_COUNT_CANDIDATES:
            fitmask = (cc <= ccdf_max) & (tc >= min_tail)
            if int(fitmask.sum()) < max(80, NBINS_FIT // 2):
                continue

            g_fit = g[fitmask]
            cc_fit = cc[fitmask]
            tc_fit = tc[fitmask]

            # left anchor point (first x in window)
            x0 = float(g_fit.min())
            cc0 = float(cc_fit[np.argmin(g_fit)])
            if not (np.isfinite(x0) and np.isfinite(cc0) and x0 > 0 and cc0 > 0):
                continue

            Y = np.log(np.clip(cc_fit, EPS, 1.0))

            # Fit weights on raw fit points (for aggregation)
            w_raw = make_tail_weights_for_fit(cc_fit, tc_fit)

            # aggregate (pass cc explicitly so aggregation respects cc weights)
            xA, YA, wA = aggregate_by_logx_bins(g_fit, Y, tc_fit, cc=cc_fit, nbins=NBINS_FIT)
            if xA.size < MIN_FIT_BINS:
                continue

            # -------- powerlaw candidate --------
            X1 = np.log(xA)
            Apl = np.stack([np.ones_like(X1), X1], axis=1)
            c0_hat, b1 = weighted_lstsq(Apl, YA, wA)
            alpha_pl = float(-b1)
            if np.isfinite(alpha_pl) and alpha_pl > 0:
                lam_pl = np.inf
                beta_pl = 1.0
                c0_pl = float(np.log(cc0) + alpha_pl * np.log(x0))  # anchored
                Yhat_pl = c0_pl - alpha_pl * np.log(xA)
                r2_fit_pl = float(weighted_r2(YA, Yhat_pl, wA))
                y_check_pl = YA + alpha_pl * np.log(xA)
                r2_check_pl, slope_pl, _ = beta_check_fit(xA, y_check_pl, wA, lam_pl, beta_pl)

                eL, eR = endpoint_logerr(g_fit, cc_fit, alpha_pl, lam_pl, beta_pl, c0_pl)
                fr = far_right_err(g, cc, alpha_pl, lam_pl, beta_pl, c0_pl)

                score_pl = float(r2_check_pl) + 0.12 * float(r2_fit_pl) \
                           - ENDPOINT_PENALTY * (eL + eR) \
                           - FARRIGHT_PENALTY * fr \
                           - 0.02 * float(np.log10(ccdf_max / 0.01)) \
                           + 0.01 * float(np.log10(min_tail / 1000))

                cand = (score_pl, alpha_pl, lam_pl, beta_pl, c0_pl, r2_fit_pl, r2_check_pl, slope_pl,
                        fitmask, ccdf_max, min_tail, (xA, YA, wA), (eL, eR, fr))
                if best is None or cand[0] > best[0]:
                    best = cand

            # -------- beta-tempered candidates --------
            for beta in BETA_GRID:
                beta = float(beta)
                x_ref = float(np.exp(np.mean(np.log(xA))))
                u = (xA / x_ref) ** beta

                A = np.stack([np.ones_like(X1), X1, u], axis=1)
                c0_hat, b1, b2 = weighted_lstsq(A, YA, wA)

                if not np.isfinite(b2) or b2 >= 0:
                    continue
                alpha = float(-b1)
                if not (np.isfinite(alpha) and alpha > 0):
                    continue

                Acoef = float(-b2)
                lam = float(x_ref / (Acoef ** (1.0 / beta)))
                if not (np.isfinite(lam) and lam > 0):
                    continue

                # cutoff visibility
                tmax = float((xA.max() / lam) ** beta)
                if tmax < CUTOFF_EPS_EFFECT:
                    continue

                # anchored c0 at x0
                c0 = float(np.log(cc0) + alpha * np.log(x0) + (x0 / lam) ** beta)

                Yhat = c0 - alpha * np.log(xA) - (xA / lam) ** beta
                r2_fit = float(weighted_r2(YA, Yhat, wA))

                y_check = YA + alpha * np.log(xA)
                r2_check, slope, _ = beta_check_fit(xA, y_check, wA, lam, beta)

                # slope near -1 is good
                slope_pen = 0.12 * float(abs(slope + 1.0)) if np.isfinite(slope) else 0.25

                # endpoint penalties
                eL, eR = endpoint_logerr(g_fit, cc_fit, alpha, lam, beta, c0)
                fr = far_right_err(g, cc, alpha, lam, beta, c0)

                score = float(r2_check) + 0.12 * float(r2_fit) \
                        - slope_pen \
                        - ENDPOINT_PENALTY * (eL + eR) \
                        - FARRIGHT_PENALTY * fr \
                        - 0.02 * float(np.log10(ccdf_max / 0.01)) \
                        + 0.01 * float(np.log10(min_tail / 1000))

                cand = (score, alpha, lam, beta, c0, r2_fit, r2_check, float(slope),
                        fitmask, ccdf_max, min_tail, (xA, YA, wA), (eL, eR, fr))
                if best is None or cand[0] > best[0]:
                    best = cand

    if best is None:
        fitmask = (cc <= 0.02) & (tc >= 5000)
        return np.nan, np.nan, np.nan, np.nan, np.nan, fitmask, 0.02, 5000, np.nan, np.nan, (np.array([]), np.array([]), np.array([])), (np.nan, np.nan, np.nan)

    (_, alpha, lam, beta, c0, r2_fit, r2_check, slope, fitmask, ccdf_max, min_tail, agg, ends) = best
    return alpha, lam, beta, c0, r2_fit, fitmask, ccdf_max, min_tail, r2_check, slope, agg, ends

# -----------------------------
# Plot suite (same 3 figures, better ends)
# -----------------------------
def plot_tempered_suite(delta_abs, title):
    delta_abs = np.asarray(delta_abs, dtype=np.float64)
    delta_abs = delta_abs[np.isfinite(delta_abs)]
    delta_abs = delta_abs[delta_abs > 0]

    if delta_abs.size < 200_000:
        print("[skip small]", title, "N=", delta_abs.size)
        return

    g, cc, tc, _ = ccdf_logbins(delta_abs)
    if g.size < 200:
        print("[skip ccdf too small]", title, "N=", delta_abs.size)
        return

    alpha, lam, beta, c0, r2_fit, fitmask, ccdf_max_used, min_tail_used, r2_check, slope_check, (xA, YA, wA), (eL, eR, fr) = \
        fit_beta_tempered_auto_anchored(g, cc, tc)

    x_min_fit = float(g[fitmask].min()) if np.any(fitmask) else float(g.min())
    cc_model = model_ccdf(g, alpha, lam, beta, c0)

    left = g < x_min_fit
    right = ~left

    # 1) CCDF + fit (same visual style)
    plt.figure(figsize=(8.2, 5.4))
    plt.plot(g, cc, lw=2, label="CCDF (data)")
    plt.plot(g[left],  cc_model[left],  lw=2, alpha=0.25, label="model (left of fit window)")
    lam_str = "inf" if (not np.isfinite(lam) or lam <= 0) else f"{lam:.2g}"
    plt.plot(g[right], cc_model[right], lw=2, alpha=1.00,
             label=f"fit: alpha={alpha:.2f}, beta={beta:.2f}, lambda={lam_str}")
    plt.xscale("log"); plt.yscale("log")
    plt.xlabel("x (log)"); plt.ylabel("CCDF (log)")
    slope_str = "n/a" if not np.isfinite(slope_check) else f"{slope_check:.3f}"
    plt.title(
        f"{title}\n"
        f"fit: CCDF<= {ccdf_max_used}, tail_count>= {min_tail_used} | "
        f"R2_check={r2_check:.4f} | slope={slope_str} | end=({eL:.3f},{eR:.3f}) fr={fr:.3f}"
    )
    plt.grid(True, which="both", ls="--", alpha=0.5)
    plt.legend()
    plt.show()

    # 2) beta-tempered check (same slot as before, now correct variable u)
    if xA.size >= 20 and np.isfinite(alpha):
        y_check = YA + alpha * np.log(xA)

        plt.figure(figsize=(8.2, 5.0))
        if np.isfinite(lam) and lam > 0:
            u = (xA / lam) ** beta
            o = np.argsort(u)
            u = u[o]; y_check_s = y_check[o]; w_s = wA[o]
            A = np.stack([np.ones_like(u), u], axis=1)
            coef = weighted_lstsq(A, y_check_s, w_s)
            yhat = A @ coef
            r2_lin = weighted_r2(y_check_s, yhat, w_s)
            plt.plot(u, y_check_s, lw=2, label="log(CCDF)+alpha log x (aggregated)")
            plt.plot(u, yhat,  lw=2, label=f"linear-in-u (weighted R2={r2_lin:.4f}, slope={coef[1]:.3f})")
            plt.xlabel(r"u = (x/lambda)^beta  (linear)")
        else:
            yhat = np.full_like(y_check, np.average(y_check, weights=wA))
            r2_lin = weighted_r2(y_check, yhat, wA)
            plt.plot(xA, y_check, lw=2, label="log(CCDF)+alpha log x (aggregated)")
            plt.plot(xA, yhat,  lw=2, label=f"constant (weighted R2={r2_lin:.4f})")
            plt.xscale("log")
            plt.xlabel("x (log)")

        plt.ylabel("value")
        plt.title("beta-tempered check (aggregated fit points)")
        plt.grid(True, ls="--", alpha=0.5)
        plt.legend()
        plt.show()

    # 3) B_eff (fit window only, correct model)
    beff = beff_from_ccdf(g, cc, window=BEFF_WINDOW)
    bm = np.isfinite(beff) & fitmask

    plt.figure(figsize=(8.2, 5.0))
    plt.plot(g[bm], beff[bm], lw=2, label="B_eff(x) on fit window")
    if np.isfinite(lam) and lam > 0:
        plt.plot(g[bm], alpha + beta * (g[bm] / lam) ** beta, lw=2, label="model: alpha + beta*(x/lambda)^beta")
    else:
        plt.plot(g[bm], np.full_like(g[bm], alpha), lw=2, label="model: alpha (power law)")
    plt.xscale("log")
    plt.xlabel("x (log)")
    plt.ylabel("B_eff(x)")
    plt.title(f"B_eff (fit window only), window={BEFF_WINDOW}")
    plt.grid(True, which="both", ls="--", alpha=0.5)
    plt.legend()
    plt.show()

    # Print quick end ratios (helps debugging ends)
    if np.any(fitmask):
        g_fit = g[fitmask]; cc_fit = cc[fitmask]
        xL = float(g_fit.min()); yL = float(cc_fit[np.argmin(g_fit)])
        xR = float(g_fit.max()); yR = float(cc_fit[np.argmax(g_fit)])
        mL = float(model_ccdf(np.array([xL]), alpha, lam, beta, c0)[0])
        mR = float(model_ccdf(np.array([xR]), alpha, lam, beta, c0)[0])
        yL = max(yL, EPS); yR = max(yR, EPS); mL = max(mL, EPS); mR = max(mR, EPS)
        print(f"\n[{title}] N={delta_abs.size:,}")
        print(f"  window: CCDF<= {ccdf_max_used}, tail_count>= {min_tail_used}, bins_fit={int(fitmask.sum())}")
        print(f"  alpha={alpha:.4f}, beta={beta:.4f}, lambda={'inf' if not np.isfinite(lam) else f'{lam:.6g}'}")
        print(f"  R2_check={r2_check:.4f}, R2_fit={r2_fit:.4f}, slope={slope_str}")
        print(f"  end-left : x={xL:.3g} data={yL:.3g} model={mL:.3g} ratio={mL/yL:.3g}")
        print(f"  end-right: x={xR:.3g} data={yR:.3g} model={mR:.3g} ratio={mR/yR:.3g}")

# --- run all groups (same as your original order) ---
def ordered_groups():
    out = []
    for phase in ["early", "mid", "late"]:
        for comp in ["q","k","v","proj"]:
            out.append(f"attn_{phase}_{comp}")
        for comp in ["fc","proj"]:
            out.append(f"mlp_{phase}_{comp}")
    return out

for gname in ordered_groups():
    plot_tempered_suite(delta_map[gname], f"{gname}: delta (iter={ITER})")

# ============================================================
# SAVE ALL CURRENT beta-tempered FIGURES TO A SINGLE PDF
#
# Assumes you already ran the fit/plot cell that defines:
#   - plot_tempered_suite(delta_abs, title)
#   - ordered_groups()
# And you have:
#   - delta_map (dict)
#   - ITER (int)
#   - RUN_DIR (optional)
#
# Output:
#   RUN_DIR/figures_tempered/beta_tempered_report_iterXXXXXXX__ALL.pdf
# ============================================================

import os
from pathlib import Path
from contextlib import contextmanager

import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages

# resolve output directory
_run_dir = Path(RUN_DIR).resolve() if "RUN_DIR" in globals() else Path.cwd().resolve()
_fig_dir = _run_dir / "figures_tempered"
_fig_dir.mkdir(parents=True, exist_ok=True)

assert "ITER" in globals(), "ITER is not defined."
assert "delta_map" in globals() and isinstance(delta_map, dict) and len(delta_map) > 0, "delta_map not found/empty."
assert "plot_tempered_suite" in globals(), "plot_tempered_suite not found. Run the fit/plot cell first."
assert "ordered_groups" in globals(), "ordered_groups not found. Run the fit/plot cell first."

PDF_PATH = _fig_dir / f"beta_tempered_report_iter{int(ITER):07d}__ALL.pdf"
print("[saving PDF to]", PDF_PATH)

@contextmanager
def _no_show():
    old = plt.show
    plt.show = lambda *a, **k: None
    try:
        yield
    finally:
        plt.show = old

pages = 0
with PdfPages(str(PDF_PATH)) as pdf, _no_show():
    plt.close("all")

    groups = ordered_groups()
    for i, gname in enumerate(groups, 1):
        if gname not in delta_map:
            print(f"[{i:02d}/{len(groups):02d}] [skip missing] {gname}")
            continue

        title = f"{gname}: delta (iter={ITER})"

        # track figures created by this call
        before = set(plt.get_fignums())
        plot_tempered_suite(delta_map[gname], title)
        after = set(plt.get_fignums())
        new_figs = sorted(list(after - before))

        for fn in new_figs:
            pdf.savefig(plt.figure(fn), bbox_inches="tight")
            pages += 1
        for fn in new_figs:
            plt.close(plt.figure(fn))

        print(f"[{i:02d}/{len(groups):02d}] saved {len(new_figs)} figs for {gname}")

    # final page
    fig = plt.figure(figsize=(10.5, 2.6))
    plt.axis("off")
    plt.text(0.02, 0.65,
             f"beta-tempered report\niter={ITER}\nfigures={pages}\nPDF={PDF_PATH.name}",
             fontsize=12, family="monospace")
    pdf.savefig(fig, bbox_inches="tight")
    plt.close(fig)
    pages += 1

size = PDF_PATH.stat().st_size if PDF_PATH.exists() else 0
print(f"[done] pages={pages} bytes={size} file={PDF_PATH}")
assert PDF_PATH.exists() and size > 0, f"PDF not created or empty: {PDF_PATH}"
```

```python
# ============================================================
# DELTA_MAP "publication" sampler + beta-tempered tail fitting
# (single-cell, end-to-end; load-or-generate; deterministic)
#
# What this cell does:
#   1) Loads precomputed delta_map NPZ if present (unless OVERWRITE=True).
#   2) Otherwise:
#        - loads nanoGPT-style GPT checkpoint from RUN_DIR/checkpoints/ckpt_iterXXXXXXX.pt
#        - samples FP32 gradients (NO AMP, NO TF32 by default) on random mini-batches
#        - collects rolling absolute deltas |g_t - g_{t-1}| on random coordinates
#        - stores an UNBIASED reservoir sample per parameter-group on CPU (float32)
#        - saves delta_map to NPZ + metadata JSON
#   3) Fits a robust "beta-tempered power law" tail:
#        CCDF(x) ≈ C * x^{-alpha} * exp(-(x/lambda)^beta)
#      with a power-law fallback (lambda=inf)
#   4) Produces a single multi-page PDF report with 3 plots per group:
#        - CCDF + fit
#        - beta-check linearity
#        - B_eff curve + model
#
# Notes:
#   - Deterministic across runs: no use of Python's hash(); fixed seeds.
#   - Unbiased reservoir sampling: correct Vitter-style (Algorithm R).
#   - True FP32 math: TF32 is disabled by default (ENABLE_TF32=False).
# ============================================================

import os, sys, json, time
from pathlib import Path
from contextlib import contextmanager, nullcontext

import numpy as np
import torch
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages

try:
    from tqdm import tqdm
except Exception:
    tqdm = None

# -----------------------------
# CONFIG (edit these)
# -----------------------------
RUN_DIR = "out/E4_clean_2k_adamw"
ITER = 800

# how many random mini-batches to sample (more -> better tail resolution)
K_BATCHES = 1024

# reproducibility
SEED = 123

# coordinate sampling
COORD_BUDGET_PER_GROUP = 600_000   # coordinates per group (split across layers)
N_COORD_SEEDS = 3                  # mix several independent coordinate selections

# reservoir size (CPU float32) per group
SAMPLES_PER_GROUP = 5_000_000      # 5M -> ~20MB per group; 18 groups -> ~360MB

# output
FIG_DIRNAME = "figures_tempered"
OVERWRITE = False                  # if False and NPZ exists -> load instead of recompute
COMPRESS = True                    # np.savez_compressed

# performance / correctness flags
ENABLE_TF32 = False                # IMPORTANT: keep False if you care about tail correctness
USE_CUDNN_BENCHMARK = True         # perf; doesn't change FP32 value semantics much

# progress display
UPDATE_EVERY = 8
PRINT_EVERY  = 64

# plotting/report
SHOW_PLOTS = True                  # in notebook; PDF is always saved
# -----------------------------

# -----------------------------
# Sanity checks
# -----------------------------
run_dir = Path(RUN_DIR).resolve()
assert run_dir.exists(), f"RUN_DIR not found: {run_dir}"

fig_dir = run_dir / FIG_DIRNAME
fig_dir.mkdir(parents=True, exist_ok=True)

npz_path  = fig_dir / f"delta_map_iter{ITER:07d}.npz"
meta_path = fig_dir / f"delta_map_iter{ITER:07d}.meta.json"
pdf_path  = fig_dir / f"beta_tempered_report_iter{ITER:07d}__ALL.pdf"

print("[info] RUN_DIR =", run_dir)
print("[info] ITER    =", ITER)
print("[info] saving  =", npz_path)
print("[info] report  =", pdf_path)

torch.manual_seed(SEED)
np.random.seed(SEED)
rng = np.random.default_rng(SEED)

# -----------------------------
# Helper: stable phase seeds (no Python hash)
# -----------------------------
PHASE_SEED = {"early": 111, "mid": 222, "late": 333}

# -----------------------------
# Part 1: Load or build delta_map
# -----------------------------
if npz_path.exists() and not OVERWRITE:
    print("[ok] NPZ exists and OVERWRITE=False -> loading")
    z = np.load(npz_path, allow_pickle=False)
    delta_map = {k: np.asarray(z[k]) for k in z.files}
    print("[ok] loaded groups:", len(delta_map))
else:
    # ---- CUDA required for sampling gradients ----
    if not torch.cuda.is_available():
        raise RuntimeError(
            "CUDA не доступна (torch.cuda.is_available()=False).\n"
            "Этот sampler рассчитан на CUDA (GPU) из-за объёма вычислений."
        )

    DEVICE = "cuda"
    DTYPE  = torch.float32

    # precision flags
    torch.backends.cuda.matmul.allow_tf32 = bool(ENABLE_TF32)
    torch.backends.cudnn.allow_tf32 = bool(ENABLE_TF32)
    torch.set_float32_matmul_precision("highest" if not ENABLE_TF32 else "high")
    torch.backends.cudnn.benchmark = bool(USE_CUDNN_BENCHMARK)

    # ---- load model (nanoGPT-style) ----
    sys.path.insert(0, str(Path.cwd()))
    from model import GPT, GPTConfig  # expects nanoGPT-compatible model.py

    cfg = json.load(open(run_dir / "config_resolved.json", "r", encoding="utf-8"))
    n_layer = int(cfg.get("n_layer", 12))
    n_head  = int(cfg.get("n_head", 12))
    n_embd  = int(cfg.get("n_embd", 768))
    block_size = int(cfg.get("block_size", 1024))
    bias = bool(cfg.get("bias", False))
    dropout = float(cfg.get("dropout", 0.0))
    batch_size = int(cfg.get("batch_size", 12))
    vocab_size = int(cfg.get("vocab_size", 50304))

    data_dir = Path(cfg.get("data_dir", "data/openwebtext"))
    data_dir = data_dir if data_dir.is_absolute() else (Path.cwd() / data_dir).resolve()

    model = GPT(GPTConfig(
        block_size=block_size,
        vocab_size=vocab_size,
        n_layer=n_layer,
        n_head=n_head,
        n_embd=n_embd,
        dropout=dropout,
        bias=bias,
    )).to(DEVICE)

    ckpt_path = run_dir / "checkpoints" / f"ckpt_iter{ITER:07d}.pt"
    assert ckpt_path.exists(), f"Checkpoint not found: {ckpt_path}"
    ckpt = torch.load(str(ckpt_path), map_location=DEVICE)
    model.load_state_dict(ckpt["model"], strict=True)

    # force fp32 weights and fp32 backward
    model = model.to(dtype=torch.float32)
    model.train()

    name2param = dict(model.named_parameters())
    print(f"[ok] model loaded. n_layer={n_layer} batch={batch_size} block={block_size} bias={bias} dropout={dropout}")

    # ---- dataset memmap (nanoGPT format) ----
    train_bin = data_dir / "train.bin"
    assert train_bin.exists(), f"train.bin not found: {train_bin}"
    train_data = np.memmap(train_bin, dtype=np.uint16, mode="r")
    train_len = int(train_data.shape[0])
    assert train_len > block_size + 2

    def get_batch_fast(bs: int, T: int, device: str):
        # random positions; pinned -> GPU
        ix = torch.randint(train_len - T - 1, (bs,), device="cpu")
        x = torch.stack([torch.from_numpy(train_data[i:i+T].astype(np.int64, copy=False)) for i in ix])
        y = torch.stack([torch.from_numpy(train_data[i+1:i+1+T].astype(np.int64, copy=False)) for i in ix])
        x = x.pin_memory().to(device, non_blocking=True)
        y = y.pin_memory().to(device, non_blocking=True)
        return x, y

    # ---- phases (early/mid/late) ----
    b0 = n_layer // 3
    b1 = 2 * n_layer // 3
    PHASE_LAYERS = {
        "early": list(range(0, b0)),
        "mid":   list(range(b0, b1)),
        "late":  list(range(b1, n_layer)),
    }
    print("[info] phase layers:", PHASE_LAYERS)

    # ---- sampler builders ----
    def sample_indices(numel: int, m: int, rng_local: np.random.Generator) -> np.ndarray:
        m = min(m, numel)
        # choice(replace=False) is fine here; m << numel typically
        return rng_local.choice(numel, size=m, replace=False).astype(np.int64)

    def to_idx_t(idx_np: np.ndarray) -> torch.Tensor:
        return torch.from_numpy(idx_np).to(device=DEVICE, dtype=torch.long)

    def build_attn_qkv_samplers(layers, block: str, total_samples: int, rng_local):
        assert block in ("q","k","v")
        samplers = []
        per_layer = max(1, total_samples // max(1, len(layers)))
        for li in layers:
            pname = f"transformer.h.{li}.attn.c_attn.weight"
            p = name2param[pname]
            numel = int(p.numel())
            block_numel = numel // 3
            offset = {"q":0,"k":1,"v":2}[block] * block_numel
            idx = sample_indices(block_numel, per_layer, rng_local) + offset
            samplers.append((pname, to_idx_t(idx)))
        return samplers

    def build_attn_proj_samplers(layers, total_samples: int, rng_local):
        samplers = []
        per_layer = max(1, total_samples // max(1, len(layers)))
        for li in layers:
            pname = f"transformer.h.{li}.attn.c_proj.weight"
            p = name2param[pname]
            idx = sample_indices(int(p.numel()), per_layer, rng_local)
            samplers.append((pname, to_idx_t(idx)))
        return samplers

    def build_mlp_fc_samplers(layers, total_samples: int, rng_local):
        samplers = []
        per_layer = max(1, total_samples // max(1, len(layers)))
        for li in layers:
            pname = f"transformer.h.{li}.mlp.c_fc.weight"
            p = name2param[pname]
            idx = sample_indices(int(p.numel()), per_layer, rng_local)
            samplers.append((pname, to_idx_t(idx)))
        return samplers

    def build_mlp_proj_samplers(layers, total_samples: int, rng_local):
        samplers = []
        per_layer = max(1, total_samples // max(1, len(layers)))
        for li in layers:
            pname = f"transformer.h.{li}.mlp.c_proj.weight"
            p = name2param[pname]
            idx = sample_indices(int(p.numel()), per_layer, rng_local)
            samplers.append((pname, to_idx_t(idx)))
        return samplers

    def ordered_groups():
        out = []
        for phase in ["early","mid","late"]:
            for comp in ["q","k","v","proj"]:
                out.append(f"attn_{phase}_{comp}")
            for comp in ["fc","proj"]:
                out.append(f"mlp_{phase}_{comp}")
        return out

    GROUPS = ordered_groups()
    print("[info] groups:", GROUPS)
    print("[info] N_COORD_SEEDS:", N_COORD_SEEDS)

    # group_samplers[gname] = list_of_seeds; each seed is list[(pname, idx_t)]
    group_samplers = {}

    for phase, layers in PHASE_LAYERS.items():
        for comp in ["q","k","v","proj"]:
            group_samplers[f"attn_{phase}_{comp}"] = []
        for comp in ["fc","proj"]:
            group_samplers[f"mlp_{phase}_{comp}"] = []

        for s in range(N_COORD_SEEDS):
            # FIX: stable seed, no Python hash
            seed_s = SEED + 10_000 * s + PHASE_SEED[phase]
            rng_s = np.random.default_rng(seed_s)

            group_samplers[f"attn_{phase}_q"].append(build_attn_qkv_samplers(layers, "q", COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"attn_{phase}_k"].append(build_attn_qkv_samplers(layers, "k", COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"attn_{phase}_v"].append(build_attn_qkv_samplers(layers, "v", COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"attn_{phase}_proj"].append(build_attn_proj_samplers(layers, COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"mlp_{phase}_fc"].append(build_mlp_fc_samplers(layers, COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"mlp_{phase}_proj"].append(build_mlp_proj_samplers(layers, COORD_BUDGET_PER_GROUP, rng_s))

    used_param_names = sorted({
        pname
        for seedlist in group_samplers.values()
        for samps in seedlist
        for (pname, _) in samps
    })

    # ---- UNBIASED reservoir sampler ----
    class Reservoir:
        """
        Unbiased reservoir sampling (Vitter, Algorithm R), vectorized.
        Keeps a uniform sample of fixed size over a stream.
        """
        def __init__(self, size: int, rng: np.random.Generator):
            self.size = int(size)
            self.rng = rng
            self.buf = np.empty((self.size,), dtype=np.float32)
            self.filled = 0
            self.seen = 0
            self.replaced = 0

        def update(self, x: np.ndarray):
            x = np.asarray(x, dtype=np.float32).reshape(-1)
            m = int(x.size)
            if m <= 0:
                return

            # Fill buffer sequentially with the first items of the stream
            if self.filled < self.size:
                take = min(self.size - self.filled, m)
                self.buf[self.filled:self.filled+take] = x[:take]
                self.filled += take
                self.seen += take
                x = x[take:]
                m = int(x.size)
                if m <= 0:
                    return

            # Now buffer is full: apply Algorithm R for the remaining items
            base = int(self.seen)
            idx = base + np.arange(1, m + 1, dtype=np.int64)  # 1-indexed stream positions
            prob = self.size / idx.astype(np.float64)         # <= 1 because idx >= size+1 here
            u = self.rng.random(m)
            keep = (u < prob)
            nk = int(np.sum(keep))
            if nk > 0:
                repl_idx = self.rng.integers(0, self.size, size=nk, endpoint=False)
                self.buf[repl_idx] = x[keep]
                self.replaced += nk
            self.seen += m

        def to_array(self) -> np.ndarray:
            return self.buf.copy() if self.filled >= self.size else self.buf[:self.filled].copy()

    reservoirs = {g: Reservoir(SAMPLES_PER_GROUP, rng=np.random.default_rng(SEED + 999 + i))
                  for i, g in enumerate(GROUPS)}

    # prev per (group, seed): previous gradient vector on GPU
    prev = {g: [None]*N_COORD_SEEDS for g in GROUPS}

    # no AMP
    amp_ctx = lambda: nullcontext()

    def _fill_stats():
        fills = np.array([reservoirs[g].filled / SAMPLES_PER_GROUP for g in GROUPS], dtype=np.float64)
        filled_cnt = int(np.sum(fills >= 0.999))
        return fills, filled_cnt

    # progress iterator
    if tqdm is None:
        it = range(K_BATCHES)
    else:
        it = tqdm(range(K_BATCHES), total=K_BATCHES, desc="batches", mininterval=0.5, dynamic_ncols=True)

    t0 = time.time()
    for k in it:
        model.zero_grad(set_to_none=True)

        X, Y = get_batch_fast(batch_size, block_size, DEVICE)
        with amp_ctx():
            _, loss = model(X, Y)
        loss.backward()

        # flatten grads once
        flat_cache = {}
        for pname in used_param_names:
            gg = name2param[pname].grad
            flat_cache[pname] = None if gg is None else gg.detach().flatten()  # fp32

        # update all reservoirs
        for gname in GROUPS:
            seed_samplers_list = group_samplers[gname]

            for sidx, samplers in enumerate(seed_samplers_list):
                chunks = []
                for pname, idx_t in samplers:
                    flat = flat_cache[pname]
                    if flat is None:
                        chunks.append(torch.zeros((idx_t.numel(),), device=DEVICE, dtype=torch.float32))
                    else:
                        chunks.append(flat.index_select(0, idx_t))
                vec = torch.cat(chunks, dim=0)

                if prev[gname][sidx] is None:
                    prev[gname][sidx] = vec
                    continue

                # rolling absolute delta uses EVERY batch (FIX)
                d = (prev[gname][sidx] - vec).abs_()
                prev[gname][sidx] = vec

                d_cpu = d.detach().cpu().numpy().astype(np.float32, copy=False)
                reservoirs[gname].update(d_cpu)

        # progress bar
        if tqdm is not None and (k % UPDATE_EVERY == 0 or k == K_BATCHES - 1):
            fills, filled_cnt = _fill_stats()
            avg_it = (time.time() - t0) / max(1, (k + 1))
            avg_repl = float(np.mean([reservoirs[g].replaced for g in GROUPS]))
            it.set_postfix(
                filled=f"{filled_cnt}/{len(fills)}",
                min=f"{fills.min()*100:4.1f}%",
                avg=f"{fills.mean()*100:4.1f}%",
                max=f"{fills.max()*100:4.1f}%",
                sec=f"{avg_it:4.2f}",
                repl=f"{avg_repl/1e6:4.2f}M"
            )

        if (k + 1) % PRINT_EVERY == 0:
            fills, filled_cnt = _fill_stats()
            avg_repl = float(np.mean([reservoirs[g].replaced for g in GROUPS]))
            avg_seen = float(np.mean([reservoirs[g].seen for g in GROUPS]))
            print("\n[detail]")
            print(f"  batch {k+1}/{K_BATCHES}")
            print(f"  groups filled: {filled_cnt}/{len(fills)}")
            print(f"  fill % (min/avg/max): {fills.min()*100:.1f} / {fills.mean()*100:.1f} / {fills.max()*100:.1f}")
            print(f"  avg reservoir replacements per group: {avg_repl/1e6:.2f}M")
            print(f"  avg deltas seen per group: {avg_seen/1e6:.2f}M")

    # finalize delta_map
    delta_map = {g: reservoirs[g].to_array() for g in GROUPS}

    print("\n[done] collected delta_map.")
    for g in GROUPS[:8]:
        r = reservoirs[g]
        print(f"  {g:16s} N={delta_map[g].size:,} replaced={r.replaced:,} seen={r.seen:,}")

    # save NPZ
    print("\n[saving] writing NPZ...")
    tS = time.time()
    if COMPRESS:
        np.savez_compressed(str(npz_path), **delta_map)
    else:
        np.savez(str(npz_path), **delta_map)
    print(f"[saving] NPZ done in {time.time()-tS:.2f}s -> {npz_path}")

    meta = dict(
        run_dir=str(run_dir),
        iter=int(ITER),
        device=str(DEVICE),
        dtype=str(DTYPE),
        enable_tf32=bool(ENABLE_TF32),
        k_batches=int(K_BATCHES),
        seed=int(SEED),
        n_coord_seeds=int(N_COORD_SEEDS),
        coord_budget_per_group=int(COORD_BUDGET_PER_GROUP),
        samples_per_group=int(SAMPLES_PER_GROUP),
        groups=GROUPS,
        reservoir=dict({g: dict(seen=int(reservoirs[g].seen), replaced=int(reservoirs[g].replaced), filled=int(reservoirs[g].filled))
                        for g in GROUPS}),
        npz=str(npz_path),
        note="FP32 weights+backward (no AMP), TF32 optional, rolling deltas, unbiased reservoir sampling, stable coord seeds"
    )
    with open(meta_path, "w", encoding="utf-8") as f:
        json.dump(meta, f, ensure_ascii=False, indent=2)

    print("[saved] npz :", npz_path)
    print("[saved] meta:", meta_path)

# ============================================================
# Part 2: beta-tempered fit + plots (robust, endpoint-aware)
# ============================================================

# --- knobs ---
Q_LO, Q_HI = 0.001, 0.99998
NBINS_CCDF = 800
NBINS_FIT  = 180
MIN_FIT_BINS = 60
BEFF_WINDOW = 21

# fit window search
CCDF_MAX_CANDIDATES = (0.2, 0.1, 0.05, 0.02, 0.01, 0.005, 0.002, 0.001)
MIN_TAIL_COUNT_CANDIDATES = (300, 500, 1000, 2000, 5000, 10000, 20000)

# beta grid for tempering
BETA_GRID = np.unique(np.concatenate([np.linspace(0.2, 1.0, 17), np.array([0.15, 0.12, 0.10])]))
CUTOFF_EPS_EFFECT = 0.02

# weights
WEIGHT_TC_POWER   = 0.65
WEIGHT_CC_POWER   = 0.30
WEIGHT_CLIP_MAX   = 200_000.0

LEFT_K,  LEFT_BOOST  = 6,  2.0
RIGHT_K, RIGHT_BOOST = 18, 80.0

ENDPOINT_PENALTY = 0.12
FARRIGHT_PENALTY = 0.18
FARRIGHT_K       = 8
EPS = 1e-300

def ccdf_logbins(x, q_lo=Q_LO, q_hi=Q_HI, nbins=NBINS_CCDF):
    x = np.asarray(x, dtype=np.float64)
    x = x[np.isfinite(x)]
    x = x[x > 0]
    x.sort()
    n = x.size
    if n < 10:
        return np.array([]), np.array([]), np.array([])

    xmin = float(np.quantile(x, q_lo))
    xmax = float(np.quantile(x, q_hi))
    if not np.isfinite(xmin) or not np.isfinite(xmax) or xmax <= xmin:
        return np.array([]), np.array([]), np.array([])

    edges = np.logspace(np.log10(xmin), np.log10(xmax), nbins + 1)
    g = edges[1:]
    idx = np.searchsorted(x, g, side="right")
    cc = (n - idx) / n
    cc = np.clip(cc, 1.0/(n+1.0), 1.0)
    tc = n * cc

    keep = np.r_[True, np.diff(idx) != 0]
    return g[keep], cc[keep], tc[keep]

def aggregate_by_logx_bins(x, y, tc, cc, nbins=NBINS_FIT):
    x = np.asarray(x, dtype=np.float64)
    y = np.asarray(y, dtype=np.float64)
    tc = np.asarray(tc, dtype=np.float64)
    cc = np.asarray(cc, dtype=np.float64)

    lx = np.log(x)
    lo, hi = float(lx.min()), float(lx.max())
    if not np.isfinite(lo) or not np.isfinite(hi) or hi <= lo:
        return x, y, np.ones_like(y)

    edges = np.linspace(lo, hi, nbins + 1)
    b = np.searchsorted(edges, lx, side="right") - 1
    b = np.clip(b, 0, nbins - 1)

    xs, ys, ws = [], [], []
    for bi in range(nbins):
        m = (b == bi)
        if not np.any(m):
            continue
        x_m = x[m]
        y_m = y[m]
        tc_m = tc[m]
        cc_m = cc[m]

        w = (np.clip(tc_m, 1.0, WEIGHT_CLIP_MAX) ** WEIGHT_TC_POWER) * (np.clip(cc_m, EPS, 1.0) ** (-WEIGHT_CC_POWER))
        wsum = float(np.sum(w))
        if wsum <= 0:
            continue

        x_bin = float(np.exp(np.sum(w * np.log(x_m)) / wsum))
        y_bin = float(np.sum(w * y_m) / wsum)

        xs.append(x_bin); ys.append(y_bin); ws.append(wsum)

    xs = np.asarray(xs); ys = np.asarray(ys); ws = np.asarray(ws)
    if xs.size:
        # explicit end-boosts on aggregated weights (FIX: not computed-but-unused)
        ws = ws.copy()
        ws[:min(LEFT_K, xs.size)] *= LEFT_BOOST
        ws[max(0, xs.size - RIGHT_K):] *= RIGHT_BOOST
        med = np.median(ws)
        if np.isfinite(med) and med > 0:
            ws /= med
        ws = np.clip(ws, 1e-12, 1e12)

    return xs, ys, ws

def weighted_lstsq(A, y, w):
    w = np.asarray(w, dtype=np.float64)
    w = np.clip(w, 1e-12, None)
    sw = np.sqrt(w)
    beta, *_ = np.linalg.lstsq(A * sw[:, None], y * sw, rcond=None)
    return beta

def weighted_r2(y, yhat, w):
    w = np.asarray(w, dtype=np.float64)
    w = np.clip(w, 1e-12, None)
    y = np.asarray(y, dtype=np.float64)
    yhat = np.asarray(yhat, dtype=np.float64)
    ybar = np.sum(w * y) / np.sum(w)
    ss_res = np.sum(w * (y - yhat) ** 2)
    ss_tot = np.sum(w * (y - ybar) ** 2)
    return 1 - ss_res / ss_tot if ss_tot > 0 else np.nan

def beff_from_ccdf(g, cc, window=BEFF_WINDOW):
    logx = np.log(g)
    logy = np.log(np.clip(cc, EPS, 1.0))
    w = window // 2
    slopes = np.full_like(logx, np.nan, dtype=np.float64)
    for i in range(len(logx)):
        j0 = max(0, i - w)
        j1 = min(len(logx), i + w + 1)
        Xw = logx[j0:j1]; Yw = logy[j0:j1]
        if len(Xw) < 2:
            continue
        Xc = Xw - Xw.mean()
        denom = np.sum(Xc * Xc)
        if denom <= 0:
            continue
        slopes[i] = np.sum(Xc * (Yw - Yw.mean())) / denom
    return -slopes

def model_ccdf(x, alpha, lam, beta, c0):
    """Numerically stable CCDF model in log-space."""
    x = np.asarray(x, dtype=np.float64)
    x = np.clip(x, 1e-300, np.inf)
    logx = np.log(x)
    if np.isfinite(lam) and lam > 0:
        # log CCDF = c0 - alpha*log x - (x/lambda)^beta
        logy = c0 - alpha * logx - (x / lam) ** beta
    else:
        logy = c0 - alpha * logx
    # CCDF must be in (0, 1]
    logy = np.minimum(logy, 0.0)
    logy = np.maximum(logy, -745.0)  # avoid float64 underflow issues
    return np.exp(logy)

def beta_check_fit(x, y_check, w, lam, beta):
    # y_check = log(CCDF)+alpha log x = c0 - (x/lambda)^beta
    if not (np.isfinite(lam) and lam > 0):
        A = np.stack([np.ones_like(x)], axis=1)
        coef = weighted_lstsq(A, y_check, w)
        yhat = A @ coef
        r2 = weighted_r2(y_check, yhat, w)
        return float(r2), float(np.nan), yhat

    u = (x / lam) ** beta
    A = np.stack([np.ones_like(u), u], axis=1)
    coef = weighted_lstsq(A, y_check, w)
    yhat = A @ coef
    r2 = weighted_r2(y_check, yhat, w)
    slope = float(coef[1])
    return float(r2), slope, yhat

def endpoint_logerr(g_fit, cc_fit, alpha, lam, beta, c0):
    iL = int(np.argmin(g_fit))
    iR = int(np.argmax(g_fit))
    xL, yL = float(g_fit[iL]), float(cc_fit[iL])
    xR, yR = float(g_fit[iR]), float(cc_fit[iR])
    mL = float(model_ccdf(np.array([xL]), alpha, lam, beta, c0)[0])
    mR = float(model_ccdf(np.array([xR]), alpha, lam, beta, c0)[0])
    yL = max(yL, EPS); yR = max(yR, EPS); mL = max(mL, EPS); mR = max(mR, EPS)
    eL = abs(np.log(mL / yL))
    eR = abs(np.log(mR / yR))
    return float(eL), float(eR)

def far_right_err(g, cc, alpha, lam, beta, c0, k=FARRIGHT_K):
    k = int(min(k, len(g)))
    if k <= 2:
        return 0.0
    xR = g[-k:]
    yR = np.clip(cc[-k:], EPS, 1.0)
    mR = np.clip(model_ccdf(xR, alpha, lam, beta, c0), EPS, 1.0)
    return float(np.mean(np.abs(np.log(mR) - np.log(yR))))

def fit_beta_tempered_auto_anchored(g, cc, tc):
    best = None
    for ccdf_max in CCDF_MAX_CANDIDATES:
        for min_tail in MIN_TAIL_COUNT_CANDIDATES:
            fitmask = (cc <= ccdf_max) & (tc >= min_tail)
            if int(fitmask.sum()) < max(80, NBINS_FIT // 2):
                continue

            g_fit = g[fitmask]
            cc_fit = cc[fitmask]
            tc_fit = tc[fitmask]

            x0 = float(g_fit.min())
            cc0 = float(cc_fit[np.argmin(g_fit)])
            if not (np.isfinite(x0) and np.isfinite(cc0) and x0 > 0 and cc0 > 0):
                continue

            Y = np.log(np.clip(cc_fit, EPS, 1.0))

            # aggregate
            xA, YA, wA = aggregate_by_logx_bins(g_fit, Y, tc_fit, cc_fit, nbins=NBINS_FIT)
            if xA.size < MIN_FIT_BINS:
                continue

            X1 = np.log(xA)

            # ---- power-law fallback ----
            Apl = np.stack([np.ones_like(X1), X1], axis=1)
            c0_hat, b1 = weighted_lstsq(Apl, YA, wA)
            alpha_pl = float(-b1)
            if np.isfinite(alpha_pl) and alpha_pl > 0:
                lam_pl = np.inf
                beta_pl = 1.0
                c0_pl = float(np.log(cc0) + alpha_pl * np.log(x0))  # anchored
                Yhat_pl = c0_pl - alpha_pl * np.log(xA)
                r2_fit_pl = float(weighted_r2(YA, Yhat_pl, wA))
                y_check_pl = YA + alpha_pl * np.log(xA)
                r2_check_pl, slope_pl, _ = beta_check_fit(xA, y_check_pl, wA, lam_pl, beta_pl)

                eL, eR = endpoint_logerr(g_fit, cc_fit, alpha_pl, lam_pl, beta_pl, c0_pl)
                fr = far_right_err(g, cc, alpha_pl, lam_pl, beta_pl, c0_pl)

                score_pl = float(r2_check_pl) + 0.12 * float(r2_fit_pl) \
                           - ENDPOINT_PENALTY * (eL + eR) \
                           - FARRIGHT_PENALTY * fr \
                           - 0.02 * float(np.log10(ccdf_max / 0.01)) \
                           + 0.01 * float(np.log10(min_tail / 1000))

                cand = (score_pl, alpha_pl, lam_pl, beta_pl, c0_pl, r2_fit_pl, r2_check_pl, float(slope_pl),
                        fitmask, ccdf_max, min_tail, (xA, YA, wA), (eL, eR, fr))
                if best is None or cand[0] > best[0]:
                    best = cand

            # ---- beta-tempered candidates ----
            for beta in BETA_GRID:
                beta = float(beta)
                x_ref = float(np.exp(np.mean(np.log(xA))))
                u = (xA / x_ref) ** beta

                A = np.stack([np.ones_like(X1), X1, u], axis=1)
                c0_hat, b1, b2 = weighted_lstsq(A, YA, wA)

                if not np.isfinite(b2) or b2 >= 0:
                    continue
                alpha = float(-b1)
                if not (np.isfinite(alpha) and alpha > 0):
                    continue

                Acoef = float(-b2)
                lam = float(x_ref / (Acoef ** (1.0 / beta)))
                if not (np.isfinite(lam) and lam > 0):
                    continue

                # cutoff visibility on fit window
                tmax = float((xA.max() / lam) ** beta)
                if tmax < CUTOFF_EPS_EFFECT:
                    continue

                # anchored constant at x0
                c0 = float(np.log(cc0) + alpha * np.log(x0) + (x0 / lam) ** beta)

                Yhat = c0 - alpha * np.log(xA) - (xA / lam) ** beta
                r2_fit = float(weighted_r2(YA, Yhat, wA))

                y_check = YA + alpha * np.log(xA)
                r2_check, slope, _ = beta_check_fit(xA, y_check, wA, lam, beta)

                slope_pen = 0.12 * float(abs(slope + 1.0)) if np.isfinite(slope) else 0.25

                eL, eR = endpoint_logerr(g_fit, cc_fit, alpha, lam, beta, c0)
                fr = far_right_err(g, cc, alpha, lam, beta, c0)

                score = float(r2_check) + 0.12 * float(r2_fit) \
                        - slope_pen \
                        - ENDPOINT_PENALTY * (eL + eR) \
                        - FARRIGHT_PENALTY * fr \
                        - 0.02 * float(np.log10(ccdf_max / 0.01)) \
                        + 0.01 * float(np.log10(min_tail / 1000))

                cand = (score, alpha, lam, beta, c0, r2_fit, r2_check, float(slope),
                        fitmask, ccdf_max, min_tail, (xA, YA, wA), (eL, eR, fr))
                if best is None or cand[0] > best[0]:
                    best = cand

    if best is None:
        fitmask = (cc <= 0.02) & (tc >= 5000)
        return np.nan, np.nan, np.nan, np.nan, np.nan, fitmask, 0.02, 5000, np.nan, np.nan, (np.array([]), np.array([]), np.array([])), (np.nan, np.nan, np.nan)

    (_, alpha, lam, beta, c0, r2_fit, r2_check, slope, fitmask, ccdf_max, min_tail, agg, ends) = best
    return alpha, lam, beta, c0, r2_fit, fitmask, ccdf_max, min_tail, r2_check, slope, agg, ends

def plot_tempered_suite(delta_abs, title):
    delta_abs = np.asarray(delta_abs, dtype=np.float64)
    delta_abs = delta_abs[np.isfinite(delta_abs)]
    delta_abs = delta_abs[delta_abs > 0]
    if delta_abs.size < 200_000:
        print("[skip small]", title, "N=", delta_abs.size)
        return

    g, cc, tc = ccdf_logbins(delta_abs)
    if g.size < 200:
        print("[skip ccdf too small]", title, "N=", delta_abs.size)
        return

    alpha, lam, beta, c0, r2_fit, fitmask, ccdf_max_used, min_tail_used, r2_check, slope_check, (xA, YA, wA), (eL, eR, fr) = \
        fit_beta_tempered_auto_anchored(g, cc, tc)

    x_min_fit = float(g[fitmask].min()) if np.any(fitmask) else float(g.min())
    cc_model = model_ccdf(g, alpha, lam, beta, c0)

    left = g < x_min_fit
    right = ~left

    # 1) CCDF + fit
    plt.figure(figsize=(8.2, 5.4))
    plt.plot(g, cc, lw=2, label="CCDF (data)")
    plt.plot(g[left],  cc_model[left],  lw=2, alpha=0.25, label="model (left of fit window)")
    lam_str = "inf" if (not np.isfinite(lam) or lam <= 0) else f"{lam:.2g}"
    plt.plot(g[right], cc_model[right], lw=2, alpha=1.00,
             label=f"fit: alpha={alpha:.2f}, beta={beta:.2f}, lambda={lam_str}")
    plt.xscale("log"); plt.yscale("log")
    plt.xlabel("x (log)"); plt.ylabel("CCDF (log)")
    slope_str = "n/a" if not np.isfinite(slope_check) else f"{slope_check:.3f}"
    plt.title(
        f"{title}\n"
        f"fit: CCDF<= {ccdf_max_used}, tail_count>= {min_tail_used} | "
        f"R2_check={r2_check:.4f} | slope={slope_str} | end=({eL:.3f},{eR:.3f}) fr={fr:.3f}"
    )
    plt.grid(True, which="both", ls="--", alpha=0.5)
    plt.legend()
    if SHOW_PLOTS:
        plt.show()

    # 2) beta-tempered check
    if xA.size >= 20 and np.isfinite(alpha):
        y_check = YA + alpha * np.log(xA)

        plt.figure(figsize=(8.2, 5.0))
        if np.isfinite(lam) and lam > 0:
            u = (xA / lam) ** beta
            o = np.argsort(u)
            u = u[o]; y_check_s = y_check[o]; w_s = wA[o]
            A = np.stack([np.ones_like(u), u], axis=1)
            coef = weighted_lstsq(A, y_check_s, w_s)
            yhat = A @ coef
            r2_lin = weighted_r2(y_check_s, yhat, w_s)
            plt.plot(u, y_check_s, lw=2, label="log(CCDF)+alpha log x (aggregated)")
            plt.plot(u, yhat,  lw=2, label=f"linear-in-u (weighted R2={r2_lin:.4f}, slope={coef[1]:.3f})")
            plt.xlabel(r"u = (x/lambda)^beta  (linear)")
        else:
            yhat = np.full_like(y_check, np.average(y_check, weights=wA))
            r2_lin = weighted_r2(y_check, yhat, wA)
            plt.plot(xA, y_check, lw=2, label="log(CCDF)+alpha log x (aggregated)")
            plt.plot(xA, yhat,  lw=2, label=f"constant (weighted R2={r2_lin:.4f})")
            plt.xscale("log")
            plt.xlabel("x (log)")

        plt.ylabel("value")
        plt.title("beta-tempered check (aggregated fit points)")
        plt.grid(True, ls="--", alpha=0.5)
        plt.legend()
        if SHOW_PLOTS:
            plt.show()

    # 3) B_eff on fit window
    beff = beff_from_ccdf(g, cc, window=BEFF_WINDOW)
    bm = np.isfinite(beff) & fitmask

    plt.figure(figsize=(8.2, 5.0))
    plt.plot(g[bm], beff[bm], lw=2, label="B_eff(x) on fit window")
    if np.isfinite(lam) and lam > 0:
        plt.plot(g[bm], alpha + beta * (g[bm] / lam) ** beta, lw=2, label="model: alpha + beta*(x/lambda)^beta")
    else:
        plt.plot(g[bm], np.full_like(g[bm], alpha), lw=2, label="model: alpha (power law)")
    plt.xscale("log")
    plt.xlabel("x (log)")
    plt.ylabel("B_eff(x)")
    plt.title(f"B_eff (fit window only), window={BEFF_WINDOW}")
    plt.grid(True, which="both", ls="--", alpha=0.5)
    plt.legend()
    if SHOW_PLOTS:
        plt.show()

# ============================================================
# Part 3: Save ALL figures into a single PDF
# ============================================================

@contextmanager
def _no_show():
    old = plt.show
    plt.show = lambda *a, **k: None
    try:
        yield
    finally:
        plt.show = old

def ordered_groups():
    out = []
    for phase in ["early", "mid", "late"]:
        for comp in ["q","k","v","proj"]:
            out.append(f"attn_{phase}_{comp}")
        for comp in ["fc","proj"]:
            out.append(f"mlp_{phase}_{comp}")
    return out

print("[info] writing PDF report:", pdf_path)
pages = 0
with PdfPages(str(pdf_path)) as pdf, _no_show():
    plt.close("all")

    for i, gname in enumerate(ordered_groups(), 1):
        if gname not in delta_map:
            print(f"[{i:02d}] [skip missing] {gname}")
            continue

        title = f"{gname}: delta (iter={ITER})"
        before = set(plt.get_fignums())
        plot_tempered_suite(delta_map[gname], title)
        after = set(plt.get_fignums())
        new_figs = sorted(list(after - before))

        for fn in new_figs:
            pdf.savefig(plt.figure(fn), bbox_inches="tight")
            pages += 1
        for fn in new_figs:
            plt.close(plt.figure(fn))

        print(f"[{i:02d}/{len(ordered_groups()):02d}] saved {len(new_figs)} figs for {gname}")

    # final summary page
    fig = plt.figure(figsize=(10.5, 2.6))
    plt.axis("off")
    plt.text(0.02, 0.65,
             f"beta-tempered report\niter={ITER}\nfigures={pages}\nPDF={pdf_path.name}\nNPZ={npz_path.name}",
             fontsize=12, family="monospace")
    pdf.savefig(fig, bbox_inches="tight")
    plt.close(fig)
    pages += 1

size = pdf_path.stat().st_size if pdf_path.exists() else 0
print(f"[done] pages={pages} bytes={size} file={pdf_path}")
assert pdf_path.exists() and size > 0, f"PDF not created or empty: {pdf_path}"
```

```python
# ============================================================
# NOISE_MAP "publication" sampler + beta-tempered tail fitting
# (single-cell, end-to-end; load-or-generate; deterministic)
#
# WHAT CHANGED vs your DELTA_MAP cell (critical fix):
#   ✅ We now measure *noise-only* proxy at fixed w_t:
#        Δ_noise := g(w; z_a) - g(w; z_b)   with independent mini-batches a,b
#      and we store |Δ_noise| on random coordinates (per group).
#      This matches the paper language "gradient noise" much better than rolling |g_t-g_{t-1}|.
#
# Everything else stays:
#   - FP32, no AMP, TF32 off by default
#   - unbiased reservoir sampling
#   - beta-tempered fit + 3 plots per group
#   - single multi-page PDF
# ============================================================

import os, sys, json, time
from pathlib import Path
from contextlib import contextmanager, nullcontext

import numpy as np
import torch
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages

try:
    from tqdm import tqdm
except Exception:
    tqdm = None

# -----------------------------
# CONFIG (edit these)
# -----------------------------
RUN_DIR = "out/E4_clean_2k_adamw"
ITER = 800

# how many *pairs* of independent mini-batches to sample
K_PAIRS = 512   # NOTE: 1 pair = 2 backward passes. Start with 256..512.

# reproducibility
SEED = 123

# coordinate sampling
COORD_BUDGET_PER_GROUP = 600_000
N_COORD_SEEDS = 3

# reservoir size (CPU float32) per group
SAMPLES_PER_GROUP = 5_000_000

# output
FIG_DIRNAME = "figures_tempered"
OVERWRITE = False
COMPRESS = True

# performance / correctness flags
ENABLE_TF32 = False
USE_CUDNN_BENCHMARK = True

# progress display
UPDATE_EVERY = 4
PRINT_EVERY  = 32

# plotting/report
SHOW_PLOTS = True
# -----------------------------

# -----------------------------
# Sanity checks
# -----------------------------
run_dir = Path(RUN_DIR).resolve()
assert run_dir.exists(), f"RUN_DIR not found: {run_dir}"

fig_dir = run_dir / FIG_DIRNAME
fig_dir.mkdir(parents=True, exist_ok=True)

# renamed outputs to avoid confusion with rolling deltas
npz_path  = fig_dir / f"noise_map_iter{ITER:07d}.npz"
meta_path = fig_dir / f"noise_map_iter{ITER:07d}.meta.json"
pdf_path  = fig_dir / f"beta_tempered_noise_report_iter{ITER:07d}__ALL.pdf"

print("[info] RUN_DIR =", run_dir)
print("[info] ITER    =", ITER)
print("[info] saving  =", npz_path)
print("[info] report  =", pdf_path)

torch.manual_seed(SEED)
np.random.seed(SEED)
rng = np.random.default_rng(SEED)

PHASE_SEED = {"early": 111, "mid": 222, "late": 333}

# -----------------------------
# Part 1: Load or build noise_map (|g_a - g_b|)
# -----------------------------
if npz_path.exists() and not OVERWRITE:
    print("[ok] NPZ exists and OVERWRITE=False -> loading")
    z = np.load(npz_path, allow_pickle=False)
    noise_map = {k: np.asarray(z[k]) for k in z.files}
    print("[ok] loaded groups:", len(noise_map))
else:
    if not torch.cuda.is_available():
        raise RuntimeError("CUDA не доступна. Нужен GPU для этой ячейки.")

    DEVICE = "cuda"
    DTYPE  = torch.float32

    torch.backends.cuda.matmul.allow_tf32 = bool(ENABLE_TF32)
    torch.backends.cudnn.allow_tf32 = bool(ENABLE_TF32)
    torch.set_float32_matmul_precision("highest" if not ENABLE_TF32 else "high")
    torch.backends.cudnn.benchmark = bool(USE_CUDNN_BENCHMARK)

    sys.path.insert(0, str(Path.cwd()))
    from model import GPT, GPTConfig  # nanoGPT-compatible model.py

    cfg = json.load(open(run_dir / "config_resolved.json", "r", encoding="utf-8"))
    n_layer = int(cfg.get("n_layer", 12))
    n_head  = int(cfg.get("n_head", 12))
    n_embd  = int(cfg.get("n_embd", 768))
    block_size = int(cfg.get("block_size", 1024))
    bias = bool(cfg.get("bias", False))
    dropout = float(cfg.get("dropout", 0.0))
    batch_size = int(cfg.get("batch_size", 12))
    vocab_size = int(cfg.get("vocab_size", 50304))

    data_dir = Path(cfg.get("data_dir", "data/openwebtext"))
    data_dir = data_dir if data_dir.is_absolute() else (Path.cwd() / data_dir).resolve()

    model = GPT(GPTConfig(
        block_size=block_size,
        vocab_size=vocab_size,
        n_layer=n_layer,
        n_head=n_head,
        n_embd=n_embd,
        dropout=dropout,
        bias=bias,
    )).to(DEVICE)

    ckpt_path = run_dir / "checkpoints" / f"ckpt_iter{ITER:07d}.pt"
    assert ckpt_path.exists(), f"Checkpoint not found: {ckpt_path}"
    ckpt = torch.load(str(ckpt_path), map_location=DEVICE)
    model.load_state_dict(ckpt["model"], strict=True)

    model = model.to(dtype=torch.float32)
    model.train()

    name2param = dict(model.named_parameters())
    print(f"[ok] model loaded. n_layer={n_layer} batch={batch_size} block={block_size} bias={bias} dropout={dropout}")

    # dataset memmap
    train_bin = data_dir / "train.bin"
    assert train_bin.exists(), f"train.bin not found: {train_bin}"
    train_data = np.memmap(train_bin, dtype=np.uint16, mode="r")
    train_len = int(train_data.shape[0])
    assert train_len > block_size + 2

    def get_batch_fast(bs: int, T: int, device: str):
        ix = torch.randint(train_len - T - 1, (bs,), device="cpu")
        x = torch.stack([torch.from_numpy(train_data[i:i+T].astype(np.int64, copy=False)) for i in ix])
        y = torch.stack([torch.from_numpy(train_data[i+1:i+1+T].astype(np.int64, copy=False)) for i in ix])
        x = x.pin_memory().to(device, non_blocking=True)
        y = y.pin_memory().to(device, non_blocking=True)
        return x, y

    # phases (early/mid/late)
    b0 = n_layer // 3
    b1 = 2 * n_layer // 3
    PHASE_LAYERS = {
        "early": list(range(0, b0)),
        "mid":   list(range(b0, b1)),
        "late":  list(range(b1, n_layer)),
    }
    print("[info] phase layers:", PHASE_LAYERS)

    # sampler builders
    def sample_indices(numel: int, m: int, rng_local: np.random.Generator) -> np.ndarray:
        m = min(m, numel)
        return rng_local.choice(numel, size=m, replace=False).astype(np.int64)

    def to_idx_t(idx_np: np.ndarray) -> torch.Tensor:
        return torch.from_numpy(idx_np).to(device=DEVICE, dtype=torch.long)

    def build_attn_qkv_samplers(layers, block: str, total_samples: int, rng_local):
        assert block in ("q","k","v")
        samplers = []
        per_layer = max(1, total_samples // max(1, len(layers)))
        for li in layers:
            pname = f"transformer.h.{li}.attn.c_attn.weight"
            p = name2param[pname]
            numel = int(p.numel())
            block_numel = numel // 3
            offset = {"q":0,"k":1,"v":2}[block] * block_numel
            idx = sample_indices(block_numel, per_layer, rng_local) + offset
            samplers.append((pname, to_idx_t(idx)))
        return samplers

    def build_attn_proj_samplers(layers, total_samples: int, rng_local):
        samplers = []
        per_layer = max(1, total_samples // max(1, len(layers)))
        for li in layers:
            pname = f"transformer.h.{li}.attn.c_proj.weight"
            p = name2param[pname]
            idx = sample_indices(int(p.numel()), per_layer, rng_local)
            samplers.append((pname, to_idx_t(idx)))
        return samplers

    def build_mlp_fc_samplers(layers, total_samples: int, rng_local):
        samplers = []
        per_layer = max(1, total_samples // max(1, len(layers)))
        for li in layers:
            pname = f"transformer.h.{li}.mlp.c_fc.weight"
            p = name2param[pname]
            idx = sample_indices(int(p.numel()), per_layer, rng_local)
            samplers.append((pname, to_idx_t(idx)))
        return samplers

    def build_mlp_proj_samplers(layers, total_samples: int, rng_local):
        samplers = []
        per_layer = max(1, total_samples // max(1, len(layers)))
        for li in layers:
            pname = f"transformer.h.{li}.mlp.c_proj.weight"
            p = name2param[pname]
            idx = sample_indices(int(p.numel()), per_layer, rng_local)
            samplers.append((pname, to_idx_t(idx)))
        return samplers

    def ordered_groups():
        out = []
        for phase in ["early","mid","late"]:
            for comp in ["q","k","v","proj"]:
                out.append(f"attn_{phase}_{comp}")
            for comp in ["fc","proj"]:
                out.append(f"mlp_{phase}_{comp}")
        return out

    GROUPS = ordered_groups()
    print("[info] groups:", GROUPS)
    print("[info] N_COORD_SEEDS:", N_COORD_SEEDS)

    # group_samplers[gname] = list_of_seeds; each seed is list[(pname, idx_t)]
    group_samplers = {}
    for phase, layers in PHASE_LAYERS.items():
        for comp in ["q","k","v","proj"]:
            group_samplers[f"attn_{phase}_{comp}"] = []
        for comp in ["fc","proj"]:
            group_samplers[f"mlp_{phase}_{comp}"] = []

        for s in range(N_COORD_SEEDS):
            seed_s = SEED + 10_000 * s + PHASE_SEED[phase]
            rng_s = np.random.default_rng(seed_s)

            group_samplers[f"attn_{phase}_q"].append(build_attn_qkv_samplers(layers, "q", COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"attn_{phase}_k"].append(build_attn_qkv_samplers(layers, "k", COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"attn_{phase}_v"].append(build_attn_qkv_samplers(layers, "v", COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"attn_{phase}_proj"].append(build_attn_proj_samplers(layers, COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"mlp_{phase}_fc"].append(build_mlp_fc_samplers(layers, COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"mlp_{phase}_proj"].append(build_mlp_proj_samplers(layers, COORD_BUDGET_PER_GROUP, rng_s))

    used_param_names = sorted({
        pname
        for seedlist in group_samplers.values()
        for samps in seedlist
        for (pname, _) in samps
    })

    # unbiased reservoir sampler
    class Reservoir:
        def __init__(self, size: int, rng: np.random.Generator):
            self.size = int(size)
            self.rng = rng
            self.buf = np.empty((self.size,), dtype=np.float32)
            self.filled = 0
            self.seen = 0
            self.replaced = 0

        def update(self, x: np.ndarray):
            x = np.asarray(x, dtype=np.float32).reshape(-1)
            m = int(x.size)
            if m <= 0:
                return

            if self.filled < self.size:
                take = min(self.size - self.filled, m)
                self.buf[self.filled:self.filled+take] = x[:take]
                self.filled += take
                self.seen += take
                x = x[take:]
                m = int(x.size)
                if m <= 0:
                    return

            base = int(self.seen)
            idx = base + np.arange(1, m + 1, dtype=np.int64)
            prob = self.size / idx.astype(np.float64)
            u = self.rng.random(m)
            keep = (u < prob)
            nk = int(np.sum(keep))
            if nk > 0:
                repl_idx = self.rng.integers(0, self.size, size=nk, endpoint=False)
                self.buf[repl_idx] = x[keep]
                self.replaced += nk
            self.seen += m

        def to_array(self) -> np.ndarray:
            return self.buf.copy() if self.filled >= self.size else self.buf[:self.filled].copy()

    reservoirs = {g: Reservoir(SAMPLES_PER_GROUP, rng=np.random.default_rng(SEED + 999 + i))
                  for i, g in enumerate(GROUPS)}

    amp_ctx = lambda: nullcontext()

    def _fill_stats():
        fills = np.array([reservoirs[g].filled / SAMPLES_PER_GROUP for g in GROUPS], dtype=np.float64)
        filled_cnt = int(np.sum(fills >= 0.999))
        return fills, filled_cnt

    # iterator
    if tqdm is None:
        it = range(K_PAIRS)
    else:
        it = tqdm(range(K_PAIRS), total=K_PAIRS, desc="pairs (2x backward each)", mininterval=0.5, dynamic_ncols=True)

    # Key trick to avoid huge GPU memory:
    #  - after backward on batch A, we extract all group vectors and store them on CPU
    #  - then backward on batch B, extract vectors again and compute |A-B| on CPU
    # This costs CPU RAM (~#groups * #seeds * coords * 4 bytes) but is stable.
    t0 = time.time()
    for k in it:
        # -------------------------
        # Pass A
        # -------------------------
        model.zero_grad(set_to_none=True)
        Xa, Ya = get_batch_fast(batch_size, block_size, DEVICE)
        with amp_ctx():
            _, loss_a = model(Xa, Ya)
        loss_a.backward()

        flat_cache_a = {}
        for pname in used_param_names:
            gg = name2param[pname].grad
            flat_cache_a[pname] = None if gg is None else gg.detach().flatten()  # GPU fp32

        # extract vectors for all (group, seed) and keep on CPU
        vecA_cpu = {}
        for gname in GROUPS:
            for sidx, samplers in enumerate(group_samplers[gname]):
                chunks = []
                for pname, idx_t in samplers:
                    flat = flat_cache_a[pname]
                    if flat is None:
                        chunks.append(torch.zeros((idx_t.numel(),), device=DEVICE, dtype=torch.float32))
                    else:
                        chunks.append(flat.index_select(0, idx_t))
                vec = torch.cat(chunks, dim=0)
                vecA_cpu[(gname, sidx)] = vec.detach().cpu().numpy().astype(np.float32, copy=False)

        # -------------------------
        # Pass B
        # -------------------------
        model.zero_grad(set_to_none=True)
        Xb, Yb = get_batch_fast(batch_size, block_size, DEVICE)
        with amp_ctx():
            _, loss_b = model(Xb, Yb)
        loss_b.backward()

        flat_cache_b = {}
        for pname in used_param_names:
            gg = name2param[pname].grad
            flat_cache_b[pname] = None if gg is None else gg.detach().flatten()

        # compute |g_a - g_b| for selected coords and stream into reservoirs
        for gname in GROUPS:
            for sidx, samplers in enumerate(group_samplers[gname]):
                chunks = []
                for pname, idx_t in samplers:
                    flat = flat_cache_b[pname]
                    if flat is None:
                        chunks.append(torch.zeros((idx_t.numel(),), device=DEVICE, dtype=torch.float32))
                    else:
                        chunks.append(flat.index_select(0, idx_t))
                vecB = torch.cat(chunks, dim=0).detach().cpu().numpy().astype(np.float32, copy=False)

                d = np.abs(vecA_cpu[(gname, sidx)] - vecB)  # noise-only proxy at fixed w
                reservoirs[gname].update(d)

        # progress
        if tqdm is not None and (k % UPDATE_EVERY == 0 or k == K_PAIRS - 1):
            fills, filled_cnt = _fill_stats()
            avg_it = (time.time() - t0) / max(1, (k + 1))
            avg_repl = float(np.mean([reservoirs[g].replaced for g in GROUPS]))
            it.set_postfix(
                filled=f"{filled_cnt}/{len(fills)}",
                min=f"{fills.min()*100:4.1f}%",
                avg=f"{fills.mean()*100:4.1f}%",
                max=f"{fills.max()*100:4.1f}%",
                sec=f"{avg_it:4.2f}",
                repl=f"{avg_repl/1e6:4.2f}M"
            )

        if (k + 1) % PRINT_EVERY == 0:
            fills, filled_cnt = _fill_stats()
            avg_repl = float(np.mean([reservoirs[g].replaced for g in GROUPS]))
            avg_seen = float(np.mean([reservoirs[g].seen for g in GROUPS]))
            print("\n[detail]")
            print(f"  pair {k+1}/{K_PAIRS}")
            print(f"  groups filled: {filled_cnt}/{len(fills)}")
            print(f"  fill % (min/avg/max): {fills.min()*100:.1f} / {fills.mean()*100:.1f} / {fills.max()*100:.1f}")
            print(f"  avg reservoir replacements per group: {avg_repl/1e6:.2f}M")
            print(f"  avg deltas seen per group: {avg_seen/1e6:.2f}M")

    # finalize
    noise_map = {g: reservoirs[g].to_array() for g in GROUPS}

    print("\n[done] collected noise_map.")
    for g in GROUPS[:8]:
        r = reservoirs[g]
        print(f"  {g:16s} N={noise_map[g].size:,} replaced={r.replaced:,} seen={r.seen:,}")

    print("\n[saving] writing NPZ...")
    tS = time.time()
    if COMPRESS:
        np.savez_compressed(str(npz_path), **noise_map)
    else:
        np.savez(str(npz_path), **noise_map)
    print(f"[saving] NPZ done in {time.time()-tS:.2f}s -> {npz_path}")

    meta = dict(
        run_dir=str(run_dir),
        iter=int(ITER),
        device=str(DEVICE),
        dtype=str(DTYPE),
        enable_tf32=bool(ENABLE_TF32),
        k_pairs=int(K_PAIRS),
        seed=int(SEED),
        n_coord_seeds=int(N_COORD_SEEDS),
        coord_budget_per_group=int(COORD_BUDGET_PER_GROUP),
        samples_per_group=int(SAMPLES_PER_GROUP),
        groups=GROUPS,
        reservoir={g: dict(seen=int(reservoirs[g].seen), replaced=int(reservoirs[g].replaced), filled=int(reservoirs[g].filled))
                   for g in GROUPS},
        npz=str(npz_path),
        note="noise-only proxy: |g(w;za) - g(w;zb)| with independent mini-batches; FP32; no AMP; TF32 optional"
    )
    with open(meta_path, "w", encoding="utf-8") as f:
        json.dump(meta, f, ensure_ascii=False, indent=2)

    print("[saved] npz :", npz_path)
    print("[saved] meta:", meta_path)

# ============================================================
# Part 2: beta-tempered fit + plots (unchanged; just rename variables)
# ============================================================

Q_LO, Q_HI = 0.001, 0.99998
NBINS_CCDF = 800
NBINS_FIT  = 180
MIN_FIT_BINS = 60
BEFF_WINDOW = 21

CCDF_MAX_CANDIDATES = (0.2, 0.1, 0.05, 0.02, 0.01, 0.005, 0.002, 0.001)
MIN_TAIL_COUNT_CANDIDATES = (300, 500, 1000, 2000, 5000, 10000, 20000)

BETA_GRID = np.unique(np.concatenate([np.linspace(0.2, 1.0, 17), np.array([0.15, 0.12, 0.10])]))
CUTOFF_EPS_EFFECT = 0.02

WEIGHT_TC_POWER   = 0.65
WEIGHT_CC_POWER   = 0.30
WEIGHT_CLIP_MAX   = 200_000.0

LEFT_K,  LEFT_BOOST  = 6,  2.0
RIGHT_K, RIGHT_BOOST = 18, 80.0

ENDPOINT_PENALTY = 0.12
FARRIGHT_PENALTY = 0.18
FARRIGHT_K       = 8
EPS = 1e-300

def ccdf_logbins(x, q_lo=Q_LO, q_hi=Q_HI, nbins=NBINS_CCDF):
    x = np.asarray(x, dtype=np.float64)
    x = x[np.isfinite(x)]
    x = x[x > 0]
    x.sort()
    n = x.size
    if n < 10:
        return np.array([]), np.array([]), np.array([])

    xmin = float(np.quantile(x, q_lo))
    xmax = float(np.quantile(x, q_hi))
    if not np.isfinite(xmin) or not np.isfinite(xmax) or xmax <= xmin:
        return np.array([]), np.array([]), np.array([])

    edges = np.logspace(np.log10(xmin), np.log10(xmax), nbins + 1)
    g = edges[1:]
    idx = np.searchsorted(x, g, side="right")
    cc = (n - idx) / n
    cc = np.clip(cc, 1.0/(n+1.0), 1.0)
    tc = n * cc

    keep = np.r_[True, np.diff(idx) != 0]
    return g[keep], cc[keep], tc[keep]

def aggregate_by_logx_bins(x, y, tc, cc, nbins=NBINS_FIT):
    x = np.asarray(x, dtype=np.float64)
    y = np.asarray(y, dtype=np.float64)
    tc = np.asarray(tc, dtype=np.float64)
    cc = np.asarray(cc, dtype=np.float64)

    lx = np.log(x)
    lo, hi = float(lx.min()), float(lx.max())
    if not np.isfinite(lo) or not np.isfinite(hi) or hi <= lo:
        return x, y, np.ones_like(y)

    edges = np.linspace(lo, hi, nbins + 1)
    b = np.searchsorted(edges, lx, side="right") - 1
    b = np.clip(b, 0, nbins - 1)

    xs, ys, ws = [], [], []
    for bi in range(nbins):
        m = (b == bi)
        if not np.any(m):
            continue
        x_m = x[m]
        y_m = y[m]
        tc_m = tc[m]
        cc_m = cc[m]

        w = (np.clip(tc_m, 1.0, WEIGHT_CLIP_MAX) ** WEIGHT_TC_POWER) * (np.clip(cc_m, EPS, 1.0) ** (-WEIGHT_CC_POWER))
        wsum = float(np.sum(w))
        if wsum <= 0:
            continue

        x_bin = float(np.exp(np.sum(w * np.log(x_m)) / wsum))
        y_bin = float(np.sum(w * y_m) / wsum)

        xs.append(x_bin); ys.append(y_bin); ws.append(wsum)

    xs = np.asarray(xs); ys = np.asarray(ys); ws = np.asarray(ws)
    if xs.size:
        ws = ws.copy()
        ws[:min(LEFT_K, xs.size)] *= LEFT_BOOST
        ws[max(0, xs.size - RIGHT_K):] *= RIGHT_BOOST
        med = np.median(ws)
        if np.isfinite(med) and med > 0:
            ws /= med
        ws = np.clip(ws, 1e-12, 1e12)

    return xs, ys, ws

def weighted_lstsq(A, y, w):
    w = np.asarray(w, dtype=np.float64)
    w = np.clip(w, 1e-12, None)
    sw = np.sqrt(w)
    beta, *_ = np.linalg.lstsq(A * sw[:, None], y * sw, rcond=None)
    return beta

def weighted_r2(y, yhat, w):
    w = np.asarray(w, dtype=np.float64)
    w = np.clip(w, 1e-12, None)
    y = np.asarray(y, dtype=np.float64)
    yhat = np.asarray(yhat, dtype=np.float64)
    ybar = np.sum(w * y) / np.sum(w)
    ss_res = np.sum(w * (y - yhat) ** 2)
    ss_tot = np.sum(w * (y - ybar) ** 2)
    return 1 - ss_res / ss_tot if ss_tot > 0 else np.nan

def beff_from_ccdf(g, cc, window=BEFF_WINDOW):
    logx = np.log(g)
    logy = np.log(np.clip(cc, EPS, 1.0))
    w = window // 2
    slopes = np.full_like(logx, np.nan, dtype=np.float64)
    for i in range(len(logx)):
        j0 = max(0, i - w)
        j1 = min(len(logx), i + w + 1)
        Xw = logx[j0:j1]; Yw = logy[j0:j1]
        if len(Xw) < 2:
            continue
        Xc = Xw - Xw.mean()
        denom = np.sum(Xc * Xc)
        if denom <= 0:
            continue
        slopes[i] = np.sum(Xc * (Yw - Yw.mean())) / denom
    return -slopes

def model_ccdf(x, alpha, lam, beta, c0):
    x = np.asarray(x, dtype=np.float64)
    x = np.clip(x, 1e-300, np.inf)
    logx = np.log(x)
    if np.isfinite(lam) and lam > 0:
        logy = c0 - alpha * logx - (x / lam) ** beta
    else:
        logy = c0 - alpha * logx
    logy = np.minimum(logy, 0.0)
    logy = np.maximum(logy, -745.0)
    return np.exp(logy)

def beta_check_fit(x, y_check, w, lam, beta):
    if not (np.isfinite(lam) and lam > 0):
        A = np.stack([np.ones_like(x)], axis=1)
        coef = weighted_lstsq(A, y_check, w)
        yhat = A @ coef
        r2 = weighted_r2(y_check, yhat, w)
        return float(r2), float(np.nan), yhat

    u = (x / lam) ** beta
    A = np.stack([np.ones_like(u), u], axis=1)
    coef = weighted_lstsq(A, y_check, w)
    yhat = A @ coef
    r2 = weighted_r2(y_check, yhat, w)
    slope = float(coef[1])
    return float(r2), slope, yhat

def endpoint_logerr(g_fit, cc_fit, alpha, lam, beta, c0):
    iL = int(np.argmin(g_fit))
    iR = int(np.argmax(g_fit))
    xL, yL = float(g_fit[iL]), float(cc_fit[iL])
    xR, yR = float(g_fit[iR]), float(cc_fit[iR])
    mL = float(model_ccdf(np.array([xL]), alpha, lam, beta, c0)[0])
    mR = float(model_ccdf(np.array([xR]), alpha, lam, beta, c0)[0])
    yL = max(yL, EPS); yR = max(yR, EPS); mL = max(mL, EPS); mR = max(mR, EPS)
    eL = abs(np.log(mL / yL))
    eR = abs(np.log(mR / yR))
    return float(eL), float(eR)

def far_right_err(g, cc, alpha, lam, beta, c0, k=FARRIGHT_K):
    k = int(min(k, len(g)))
    if k <= 2:
        return 0.0
    xR = g[-k:]
    yR = np.clip(cc[-k:], EPS, 1.0)
    mR = np.clip(model_ccdf(xR, alpha, lam, beta, c0), EPS, 1.0)
    return float(np.mean(np.abs(np.log(mR) - np.log(yR))))

def fit_beta_tempered_auto_anchored(g, cc, tc):
    best = None
    for ccdf_max in CCDF_MAX_CANDIDATES:
        for min_tail in MIN_TAIL_COUNT_CANDIDATES:
            fitmask = (cc <= ccdf_max) & (tc >= min_tail)
            if int(fitmask.sum()) < max(80, NBINS_FIT // 2):
                continue

            g_fit = g[fitmask]
            cc_fit = cc[fitmask]
            tc_fit = tc[fitmask]

            x0 = float(g_fit.min())
            cc0 = float(cc_fit[np.argmin(g_fit)])
            if not (np.isfinite(x0) and np.isfinite(cc0) and x0 > 0 and cc0 > 0):
                continue

            Y = np.log(np.clip(cc_fit, EPS, 1.0))
            xA, YA, wA = aggregate_by_logx_bins(g_fit, Y, tc_fit, cc_fit, nbins=NBINS_FIT)
            if xA.size < MIN_FIT_BINS:
                continue

            X1 = np.log(xA)

            # power-law fallback
            Apl = np.stack([np.ones_like(X1), X1], axis=1)
            c0_hat, b1 = weighted_lstsq(Apl, YA, wA)
            alpha_pl = float(-b1)
            if np.isfinite(alpha_pl) and alpha_pl > 0:
                lam_pl = np.inf
                beta_pl = 1.0
                c0_pl = float(np.log(cc0) + alpha_pl * np.log(x0))
                Yhat_pl = c0_pl - alpha_pl * np.log(xA)
                r2_fit_pl = float(weighted_r2(YA, Yhat_pl, wA))
                y_check_pl = YA + alpha_pl * np.log(xA)
                r2_check_pl, slope_pl, _ = beta_check_fit(xA, y_check_pl, wA, lam_pl, beta_pl)

                eL, eR = endpoint_logerr(g_fit, cc_fit, alpha_pl, lam_pl, beta_pl, c0_pl)
                fr = far_right_err(g, cc, alpha_pl, lam_pl, beta_pl, c0_pl)

                score_pl = float(r2_check_pl) + 0.12 * float(r2_fit_pl) \
                           - ENDPOINT_PENALTY * (eL + eR) \
                           - FARRIGHT_PENALTY * fr \
                           - 0.02 * float(np.log10(ccdf_max / 0.01)) \
                           + 0.01 * float(np.log10(min_tail / 1000))

                cand = (score_pl, alpha_pl, lam_pl, beta_pl, c0_pl, r2_fit_pl, r2_check_pl, float(slope_pl),
                        fitmask, ccdf_max, min_tail, (xA, YA, wA), (eL, eR, fr))
                if best is None or cand[0] > best[0]:
                    best = cand

            # beta-tempered candidates
            for beta in BETA_GRID:
                beta = float(beta)
                x_ref = float(np.exp(np.mean(np.log(xA))))
                u = (xA / x_ref) ** beta

                A = np.stack([np.ones_like(X1), X1, u], axis=1)
                c0_hat, b1, b2 = weighted_lstsq(A, YA, wA)

                if not np.isfinite(b2) or b2 >= 0:
                    continue
                alpha = float(-b1)
                if not (np.isfinite(alpha) and alpha > 0):
                    continue

                Acoef = float(-b2)
                lam = float(x_ref / (Acoef ** (1.0 / beta)))
                if not (np.isfinite(lam) and lam > 0):
                    continue

                tmax = float((xA.max() / lam) ** beta)
                if tmax < CUTOFF_EPS_EFFECT:
                    continue

                c0 = float(np.log(cc0) + alpha * np.log(x0) + (x0 / lam) ** beta)

                Yhat = c0 - alpha * np.log(xA) - (xA / lam) ** beta
                r2_fit = float(weighted_r2(YA, Yhat, wA))

                y_check = YA + alpha * np.log(xA)
                r2_check, slope, _ = beta_check_fit(xA, y_check, wA, lam, beta)

                slope_pen = 0.12 * float(abs(slope + 1.0)) if np.isfinite(slope) else 0.25
                eL, eR = endpoint_logerr(g_fit, cc_fit, alpha, lam, beta, c0)
                fr = far_right_err(g, cc, alpha, lam, beta, c0)

                score = float(r2_check) + 0.12 * float(r2_fit) \
                        - slope_pen \
                        - ENDPOINT_PENALTY * (eL + eR) \
                        - FARRIGHT_PENALTY * fr \
                        - 0.02 * float(np.log10(ccdf_max / 0.01)) \
                        + 0.01 * float(np.log10(min_tail / 1000))

                cand = (score, alpha, lam, beta, c0, r2_fit, r2_check, float(slope),
                        fitmask, ccdf_max, min_tail, (xA, YA, wA), (eL, eR, fr))
                if best is None or cand[0] > best[0]:
                    best = cand

    if best is None:
        fitmask = (cc <= 0.02) & (tc >= 5000)
        return np.nan, np.nan, np.nan, np.nan, np.nan, fitmask, 0.02, 5000, np.nan, np.nan, (np.array([]), np.array([]), np.array([])), (np.nan, np.nan, np.nan)

    (_, alpha, lam, beta, c0, r2_fit, r2_check, slope, fitmask, ccdf_max, min_tail, agg, ends) = best
    return alpha, lam, beta, c0, r2_fit, fitmask, ccdf_max, min_tail, r2_check, slope, agg, ends

def plot_tempered_suite(delta_abs, title):
    delta_abs = np.asarray(delta_abs, dtype=np.float64)
    delta_abs = delta_abs[np.isfinite(delta_abs)]
    delta_abs = delta_abs[delta_abs > 0]
    if delta_abs.size < 200_000:
        print("[skip small]", title, "N=", delta_abs.size)
        return

    g, cc, tc = ccdf_logbins(delta_abs)
    if g.size < 200:
        print("[skip ccdf too small]", title, "N=", delta_abs.size)
        return

    alpha, lam, beta, c0, r2_fit, fitmask, ccdf_max_used, min_tail_used, r2_check, slope_check, (xA, YA, wA), (eL, eR, fr) = \
        fit_beta_tempered_auto_anchored(g, cc, tc)

    x_min_fit = float(g[fitmask].min()) if np.any(fitmask) else float(g.min())
    cc_model = model_ccdf(g, alpha, lam, beta, c0)

    left = g < x_min_fit
    right = ~left

    # 1) CCDF + fit
    plt.figure(figsize=(8.2, 5.4))
    plt.plot(g, cc, lw=2, label="CCDF (data)")
    plt.plot(g[left],  cc_model[left],  lw=2, alpha=0.25, label="model (left of fit window)")
    lam_str = "inf" if (not np.isfinite(lam) or lam <= 0) else f"{lam:.2g}"
    plt.plot(g[right], cc_model[right], lw=2, alpha=1.00,
             label=f"fit: alpha={alpha:.2f}, beta={beta:.2f}, lambda={lam_str}")
    plt.xscale("log"); plt.yscale("log")
    plt.xlabel("x (log)"); plt.ylabel("CCDF (log)")
    slope_str = "n/a" if not np.isfinite(slope_check) else f"{slope_check:.3f}"
    plt.title(
        f"{title}\n"
        f"fit: CCDF<= {ccdf_max_used}, tail_count>= {min_tail_used} | "
        f"R2_check={r2_check:.4f} | slope={slope_str} | end=({eL:.3f},{eR:.3f}) fr={fr:.3f}"
    )
    plt.grid(True, which="both", ls="--", alpha=0.5)
    plt.legend()
    if SHOW_PLOTS:
        plt.show()

    # 2) beta-check
    if xA.size >= 20 and np.isfinite(alpha):
        y_check = YA + alpha * np.log(xA)
        plt.figure(figsize=(8.2, 5.0))
        if np.isfinite(lam) and lam > 0:
            u = (xA / lam) ** beta
            o = np.argsort(u)
            u = u[o]; y_check_s = y_check[o]; w_s = wA[o]
            A = np.stack([np.ones_like(u), u], axis=1)
            coef = weighted_lstsq(A, y_check_s, w_s)
            yhat = A @ coef
            r2_lin = weighted_r2(y_check_s, yhat, w_s)
            plt.plot(u, y_check_s, lw=2, label="log(CCDF)+alpha log x (aggregated)")
            plt.plot(u, yhat,  lw=2, label=f"linear-in-u (weighted R2={r2_lin:.4f}, slope={coef[1]:.3f})")
            plt.xlabel(r"u = (x/lambda)^beta")
        else:
            yhat = np.full_like(y_check, np.average(y_check, weights=wA))
            r2_lin = weighted_r2(y_check, yhat, wA)
            plt.plot(xA, y_check, lw=2, label="log(CCDF)+alpha log x (aggregated)")
            plt.plot(xA, yhat,  lw=2, label=f"constant (weighted R2={r2_lin:.4f})")
            plt.xscale("log")
            plt.xlabel("x (log)")
        plt.ylabel("value")
        plt.title("beta-tempered check (aggregated fit points)")
        plt.grid(True, ls="--", alpha=0.5)
        plt.legend()
        if SHOW_PLOTS:
            plt.show()

    # 3) B_eff
    beff = beff_from_ccdf(g, cc, window=BEFF_WINDOW)
    bm = np.isfinite(beff) & fitmask
    plt.figure(figsize=(8.2, 5.0))
    plt.plot(g[bm], beff[bm], lw=2, label="B_eff(x) on fit window")
    if np.isfinite(lam) and lam > 0:
        plt.plot(g[bm], alpha + beta * (g[bm] / lam) ** beta, lw=2, label="model: alpha + beta*(x/lambda)^beta")
    else:
        plt.plot(g[bm], np.full_like(g[bm], alpha), lw=2, label="model: alpha (power law)")
    plt.xscale("log")
    plt.xlabel("x (log)")
    plt.ylabel("B_eff(x)")
    plt.title(f"B_eff (fit window only), window={BEFF_WINDOW}")
    plt.grid(True, which="both", ls="--", alpha=0.5)
    plt.legend()
    if SHOW_PLOTS:
        plt.show()

# ============================================================
# Part 3: Save ALL figures into a single PDF
# ============================================================

@contextmanager
def _no_show():
    old = plt.show
    plt.show = lambda *a, **k: None
    try:
        yield
    finally:
        plt.show = old

def ordered_groups():
    out = []
    for phase in ["early", "mid", "late"]:
        for comp in ["q","k","v","proj"]:
            out.append(f"attn_{phase}_{comp}")
        for comp in ["fc","proj"]:
            out.append(f"mlp_{phase}_{comp}")
    return out

print("[info] writing PDF report:", pdf_path)
pages = 0
with PdfPages(str(pdf_path)) as pdf, _no_show():
    plt.close("all")
    groups = ordered_groups()

    for i, gname in enumerate(groups, 1):
        if gname not in noise_map:
            print(f"[{i:02d}] [skip missing] {gname}")
            continue

        title = f"{gname}: |g(za)-g(zb)| (iter={ITER})"
        before = set(plt.get_fignums())
        plot_tempered_suite(noise_map[gname], title)
        after = set(plt.get_fignums())
        new_figs = sorted(list(after - before))

        for fn in new_figs:
            pdf.savefig(plt.figure(fn), bbox_inches="tight")
            pages += 1
        for fn in new_figs:
            plt.close(plt.figure(fn))

        print(f"[{i:02d}/{len(groups):02d}] saved {len(new_figs)} figs for {gname}")

    fig = plt.figure(figsize=(10.5, 2.6))
    plt.axis("off")
    plt.text(0.02, 0.65,
             f"beta-tempered NOISE report\niter={ITER}\nfigures={pages}\nPDF={pdf_path.name}\nNPZ={npz_path.name}",
             fontsize=12, family="monospace")
    pdf.savefig(fig, bbox_inches="tight")
    plt.close(fig)
    pages += 1

size = pdf_path.stat().st_size if pdf_path.exists() else 0
print(f"[done] pages={pages} bytes={size} file={pdf_path}")
assert pdf_path.exists() and size > 0, f"PDF not created or empty: {pdf_path}"
```

```python
# ============================================================
# GRAD_MAP "publication" sampler + beta-tempered tail fitting
# (single-cell, end-to-end; load-or-generate; deterministic)
#
# PURPOSE (as requested):
#   Build the SAME style report, but for the *stochastic gradient itself*:
#       x := |g(w; z)|    (not noise difference, not rolling delta)
#
# What this cell does:
#   1) Loads precomputed grad_map NPZ if present (unless OVERWRITE=True).
#   2) Otherwise:
#        - loads nanoGPT-style GPT checkpoint from RUN_DIR/checkpoints/ckpt_iterXXXXXXX.pt
#        - samples FP32 gradients (NO AMP, TF32 off by default)
#        - collects |g| on random coordinates per parameter-group
#        - stores an UNBIASED reservoir sample per group on CPU (float32)
#        - saves grad_map to NPZ + metadata JSON
#   3) Fits robust "beta-tempered power law" tail:
#        CCDF(x) ≈ C * x^{-alpha} * exp(-(x/lambda)^beta), with power-law fallback
#   4) Produces a single multi-page PDF report with 3 plots per group:
#        - CCDF + fit
#        - beta-check linearity
#        - B_eff curve + model
#
# Notes:
#   - Deterministic (no Python hash); stable seeds.
#   - Unbiased reservoir sampling (Vitter Algorithm R).
#   - True FP32 math (TF32 disabled by default).
# ============================================================

import os, sys, json, time
from pathlib import Path
from contextlib import contextmanager, nullcontext

import numpy as np
import torch
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages

try:
    from tqdm import tqdm
except Exception:
    tqdm = None

# -----------------------------
# CONFIG (edit these)
# -----------------------------
RUN_DIR = "out/E4_clean_2k_adamw"
ITER = 800

# how many random mini-batches to sample (more -> better tail resolution)
K_BATCHES = 1024

# reproducibility
SEED = 123

# coordinate sampling
COORD_BUDGET_PER_GROUP = 600_000
N_COORD_SEEDS = 3

# reservoir size (CPU float32) per group
SAMPLES_PER_GROUP = 5_000_000

# output
FIG_DIRNAME = "figures_tempered"
OVERWRITE = False
COMPRESS = True

# performance / correctness flags
ENABLE_TF32 = False
USE_CUDNN_BENCHMARK = True

# progress display
UPDATE_EVERY = 8
PRINT_EVERY  = 64

# plotting/report
SHOW_PLOTS = True
# -----------------------------

# -----------------------------
# Sanity checks
# -----------------------------
run_dir = Path(RUN_DIR).resolve()
assert run_dir.exists(), f"RUN_DIR not found: {run_dir}"

fig_dir = run_dir / FIG_DIRNAME
fig_dir.mkdir(parents=True, exist_ok=True)

npz_path  = fig_dir / f"grad_map_iter{ITER:07d}.npz"
meta_path = fig_dir / f"grad_map_iter{ITER:07d}.meta.json"
pdf_path  = fig_dir / f"beta_tempered_grad_report_iter{ITER:07d}__ALL.pdf"

print("[info] RUN_DIR =", run_dir)
print("[info] ITER    =", ITER)
print("[info] saving  =", npz_path)
print("[info] report  =", pdf_path)

torch.manual_seed(SEED)
np.random.seed(SEED)
rng = np.random.default_rng(SEED)

PHASE_SEED = {"early": 111, "mid": 222, "late": 333}

# -----------------------------
# Part 1: Load or build grad_map (|g|)
# -----------------------------
if npz_path.exists() and not OVERWRITE:
    print("[ok] NPZ exists and OVERWRITE=False -> loading")
    z = np.load(npz_path, allow_pickle=False)
    grad_map = {k: np.asarray(z[k]) for k in z.files}
    print("[ok] loaded groups:", len(grad_map))
else:
    if not torch.cuda.is_available():
        raise RuntimeError("CUDA не доступна. Нужен GPU для этой ячейки.")

    DEVICE = "cuda"
    DTYPE  = torch.float32

    torch.backends.cuda.matmul.allow_tf32 = bool(ENABLE_TF32)
    torch.backends.cudnn.allow_tf32 = bool(ENABLE_TF32)
    torch.set_float32_matmul_precision("highest" if not ENABLE_TF32 else "high")
    torch.backends.cudnn.benchmark = bool(USE_CUDNN_BENCHMARK)

    sys.path.insert(0, str(Path.cwd()))
    from model import GPT, GPTConfig  # nanoGPT-compatible model.py

    cfg = json.load(open(run_dir / "config_resolved.json", "r", encoding="utf-8"))
    n_layer = int(cfg.get("n_layer", 12))
    n_head  = int(cfg.get("n_head", 12))
    n_embd  = int(cfg.get("n_embd", 768))
    block_size = int(cfg.get("block_size", 1024))
    bias = bool(cfg.get("bias", False))
    dropout = float(cfg.get("dropout", 0.0))
    batch_size = int(cfg.get("batch_size", 12))
    vocab_size = int(cfg.get("vocab_size", 50304))

    data_dir = Path(cfg.get("data_dir", "data/openwebtext"))
    data_dir = data_dir if data_dir.is_absolute() else (Path.cwd() / data_dir).resolve()

    model = GPT(GPTConfig(
        block_size=block_size,
        vocab_size=vocab_size,
        n_layer=n_layer,
        n_head=n_head,
        n_embd=n_embd,
        dropout=dropout,
        bias=bias,
    )).to(DEVICE)

    ckpt_path = run_dir / "checkpoints" / f"ckpt_iter{ITER:07d}.pt"
    assert ckpt_path.exists(), f"Checkpoint not found: {ckpt_path}"
    ckpt = torch.load(str(ckpt_path), map_location=DEVICE)
    model.load_state_dict(ckpt["model"], strict=True)

    model = model.to(dtype=torch.float32)
    model.train()

    name2param = dict(model.named_parameters())
    print(f"[ok] model loaded. n_layer={n_layer} batch={batch_size} block={block_size} bias={bias} dropout={dropout}")

    # dataset memmap
    train_bin = data_dir / "train.bin"
    assert train_bin.exists(), f"train.bin not found: {train_bin}"
    train_data = np.memmap(train_bin, dtype=np.uint16, mode="r")
    train_len = int(train_data.shape[0])
    assert train_len > block_size + 2

    def get_batch_fast(bs: int, T: int, device: str):
        ix = torch.randint(train_len - T - 1, (bs,), device="cpu")
        x = torch.stack([torch.from_numpy(train_data[i:i+T].astype(np.int64, copy=False)) for i in ix])
        y = torch.stack([torch.from_numpy(train_data[i+1:i+1+T].astype(np.int64, copy=False)) for i in ix])
        x = x.pin_memory().to(device, non_blocking=True)
        y = y.pin_memory().to(device, non_blocking=True)
        return x, y

    # phases (early/mid/late)
    b0 = n_layer // 3
    b1 = 2 * n_layer // 3
    PHASE_LAYERS = {
        "early": list(range(0, b0)),
        "mid":   list(range(b0, b1)),
        "late":  list(range(b1, n_layer)),
    }
    print("[info] phase layers:", PHASE_LAYERS)

    # sampler builders
    def sample_indices(numel: int, m: int, rng_local: np.random.Generator) -> np.ndarray:
        m = min(m, numel)
        return rng_local.choice(numel, size=m, replace=False).astype(np.int64)

    def to_idx_t(idx_np: np.ndarray) -> torch.Tensor:
        return torch.from_numpy(idx_np).to(device=DEVICE, dtype=torch.long)

    def build_attn_qkv_samplers(layers, block: str, total_samples: int, rng_local):
        assert block in ("q","k","v")
        samplers = []
        per_layer = max(1, total_samples // max(1, len(layers)))
        for li in layers:
            pname = f"transformer.h.{li}.attn.c_attn.weight"
            p = name2param[pname]
            numel = int(p.numel())
            block_numel = numel // 3
            offset = {"q":0,"k":1,"v":2}[block] * block_numel
            idx = sample_indices(block_numel, per_layer, rng_local) + offset
            samplers.append((pname, to_idx_t(idx)))
        return samplers

    def build_attn_proj_samplers(layers, total_samples: int, rng_local):
        samplers = []
        per_layer = max(1, total_samples // max(1, len(layers)))
        for li in layers:
            pname = f"transformer.h.{li}.attn.c_proj.weight"
            p = name2param[pname]
            idx = sample_indices(int(p.numel()), per_layer, rng_local)
            samplers.append((pname, to_idx_t(idx)))
        return samplers

    def build_mlp_fc_samplers(layers, total_samples: int, rng_local):
        samplers = []
        per_layer = max(1, total_samples // max(1, len(layers)))
        for li in layers:
            pname = f"transformer.h.{li}.mlp.c_fc.weight"
            p = name2param[pname]
            idx = sample_indices(int(p.numel()), per_layer, rng_local)
            samplers.append((pname, to_idx_t(idx)))
        return samplers

    def build_mlp_proj_samplers(layers, total_samples: int, rng_local):
        samplers = []
        per_layer = max(1, total_samples // max(1, len(layers)))
        for li in layers:
            pname = f"transformer.h.{li}.mlp.c_proj.weight"
            p = name2param[pname]
            idx = sample_indices(int(p.numel()), per_layer, rng_local)
            samplers.append((pname, to_idx_t(idx)))
        return samplers

    def ordered_groups():
        out = []
        for phase in ["early","mid","late"]:
            for comp in ["q","k","v","proj"]:
                out.append(f"attn_{phase}_{comp}")
            for comp in ["fc","proj"]:
                out.append(f"mlp_{phase}_{comp}")
        return out

    GROUPS = ordered_groups()
    print("[info] groups:", GROUPS)
    print("[info] N_COORD_SEEDS:", N_COORD_SEEDS)

    group_samplers = {}
    for phase, layers in PHASE_LAYERS.items():
        for comp in ["q","k","v","proj"]:
            group_samplers[f"attn_{phase}_{comp}"] = []
        for comp in ["fc","proj"]:
            group_samplers[f"mlp_{phase}_{comp}"] = []

        for s in range(N_COORD_SEEDS):
            seed_s = SEED + 10_000 * s + PHASE_SEED[phase]
            rng_s = np.random.default_rng(seed_s)

            group_samplers[f"attn_{phase}_q"].append(build_attn_qkv_samplers(layers, "q", COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"attn_{phase}_k"].append(build_attn_qkv_samplers(layers, "k", COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"attn_{phase}_v"].append(build_attn_qkv_samplers(layers, "v", COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"attn_{phase}_proj"].append(build_attn_proj_samplers(layers, COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"mlp_{phase}_fc"].append(build_mlp_fc_samplers(layers, COORD_BUDGET_PER_GROUP, rng_s))
            group_samplers[f"mlp_{phase}_proj"].append(build_mlp_proj_samplers(layers, COORD_BUDGET_PER_GROUP, rng_s))

    used_param_names = sorted({
        pname
        for seedlist in group_samplers.values()
        for samps in seedlist
        for (pname, _) in samps
    })

    # Reservoir
    class Reservoir:
        def __init__(self, size: int, rng: np.random.Generator):
            self.size = int(size)
            self.rng = rng
            self.buf = np.empty((self.size,), dtype=np.float32)
            self.filled = 0
            self.seen = 0
            self.replaced = 0

        def update(self, x: np.ndarray):
            x = np.asarray(x, dtype=np.float32).reshape(-1)
            m = int(x.size)
            if m <= 0:
                return

            if self.filled < self.size:
                take = min(self.size - self.filled, m)
                self.buf[self.filled:self.filled+take] = x[:take]
                self.filled += take
                self.seen += take
                x = x[take:]
                m = int(x.size)
                if m <= 0:
                    return

            base = int(self.seen)
            idx = base + np.arange(1, m + 1, dtype=np.int64)
            prob = self.size / idx.astype(np.float64)
            u = self.rng.random(m)
            keep = (u < prob)
            nk = int(np.sum(keep))
            if nk > 0:
                repl_idx = self.rng.integers(0, self.size, size=nk, endpoint=False)
                self.buf[repl_idx] = x[keep]
                self.replaced += nk
            self.seen += m

        def to_array(self) -> np.ndarray:
            return self.buf.copy() if self.filled >= self.size else self.buf[:self.filled].copy()

    reservoirs = {g: Reservoir(SAMPLES_PER_GROUP, rng=np.random.default_rng(SEED + 999 + i))
                  for i, g in enumerate(GROUPS)}

    amp_ctx = lambda: nullcontext()

    def _fill_stats():
        fills = np.array([reservoirs[g].filled / SAMPLES_PER_GROUP for g in GROUPS], dtype=np.float64)
        filled_cnt = int(np.sum(fills >= 0.999))
        return fills, filled_cnt

    if tqdm is None:
        it = range(K_BATCHES)
    else:
        it = tqdm(range(K_BATCHES), total=K_BATCHES, desc="batches", mininterval=0.5, dynamic_ncols=True)

    t0 = time.time()
    for k in it:
        model.zero_grad(set_to_none=True)
        X, Y = get_batch_fast(batch_size, block_size, DEVICE)
        with amp_ctx():
            _, loss = model(X, Y)
        loss.backward()

        # flatten grads once (GPU)
        flat_cache = {}
        for pname in used_param_names:
            gg = name2param[pname].grad
            flat_cache[pname] = None if gg is None else gg.detach().flatten()

        # stream |g| on sampled coordinates
        for gname in GROUPS:
            for sidx, samplers in enumerate(group_samplers[gname]):
                chunks = []
                for pname, idx_t in samplers:
                    flat = flat_cache[pname]
                    if flat is None:
                        chunks.append(torch.zeros((idx_t.numel(),), device=DEVICE, dtype=torch.float32))
                    else:
                        chunks.append(flat.index_select(0, idx_t))
                vec = torch.cat(chunks, dim=0)
                x_cpu = vec.abs_().detach().cpu().numpy().astype(np.float32, copy=False)
                reservoirs[gname].update(x_cpu)

        if tqdm is not None and (k % UPDATE_EVERY == 0 or k == K_BATCHES - 1):
            fills, filled_cnt = _fill_stats()
            avg_it = (time.time() - t0) / max(1, (k + 1))
            avg_repl = float(np.mean([reservoirs[g].replaced for g in GROUPS]))
            it.set_postfix(
                filled=f"{filled_cnt}/{len(fills)}",
                min=f"{fills.min()*100:4.1f}%",
                avg=f"{fills.mean()*100:4.1f}%",
                max=f"{fills.max()*100:4.1f}%",
                sec=f"{avg_it:4.2f}",
                repl=f"{avg_repl/1e6:4.2f}M"
            )

        if (k + 1) % PRINT_EVERY == 0:
            fills, filled_cnt = _fill_stats()
            avg_repl = float(np.mean([reservoirs[g].replaced for g in GROUPS]))
            avg_seen = float(np.mean([reservoirs[g].seen for g in GROUPS]))
            print("\n[detail]")
            print(f"  batch {k+1}/{K_BATCHES}")
            print(f"  groups filled: {filled_cnt}/{len(fills)}")
            print(f"  fill % (min/avg/max): {fills.min()*100:.1f} / {fills.mean()*100:.1f} / {fills.max()*100:.1f}")
            print(f"  avg reservoir replacements per group: {avg_repl/1e6:.2f}M")
            print(f"  avg samples seen per group: {avg_seen/1e6:.2f}M")

    grad_map = {g: reservoirs[g].to_array() for g in GROUPS}

    print("\n[done] collected grad_map.")
    for g in GROUPS[:8]:
        r = reservoirs[g]
        print(f"  {g:16s} N={grad_map[g].size:,} replaced={r.replaced:,} seen={r.seen:,}")

    print("\n[saving] writing NPZ...")
    tS = time.time()
    if COMPRESS:
        np.savez_compressed(str(npz_path), **grad_map)
    else:
        np.savez(str(npz_path), **grad_map)
    print(f"[saving] NPZ done in {time.time()-tS:.2f}s -> {npz_path}")

    meta = dict(
        run_dir=str(run_dir),
        iter=int(ITER),
        device=str(DEVICE),
        dtype=str(DTYPE),
        enable_tf32=bool(ENABLE_TF32),
        k_batches=int(K_BATCHES),
        seed=int(SEED),
        n_coord_seeds=int(N_COORD_SEEDS),
        coord_budget_per_group=int(COORD_BUDGET_PER_GROUP),
        samples_per_group=int(SAMPLES_PER_GROUP),
        groups=GROUPS,
        reservoir={g: dict(seen=int(reservoirs[g].seen), replaced=int(reservoirs[g].replaced), filled=int(reservoirs[g].filled))
                   for g in GROUPS},
        npz=str(npz_path),
        note="stochastic gradient tails: store |g(w;z)| on sampled coordinates; FP32; no AMP; TF32 optional"
    )
    with open(meta_path, "w", encoding="utf-8") as f:
        json.dump(meta, f, ensure_ascii=False, indent=2)

    print("[saved] npz :", npz_path)
    print("[saved] meta:", meta_path)

# ============================================================
# Part 2: beta-tempered fit + plots (copied from your last cell)
# ============================================================

Q_LO, Q_HI = 0.001, 0.99998
NBINS_CCDF = 800
NBINS_FIT  = 180
MIN_FIT_BINS = 60
BEFF_WINDOW = 21

CCDF_MAX_CANDIDATES = (0.2, 0.1, 0.05, 0.02, 0.01, 0.005, 0.002, 0.001)
MIN_TAIL_COUNT_CANDIDATES = (300, 500, 1000, 2000, 5000, 10000, 20000)

BETA_GRID = np.unique(np.concatenate([np.linspace(0.2, 1.0, 17), np.array([0.15, 0.12, 0.10])]))
CUTOFF_EPS_EFFECT = 0.02

WEIGHT_TC_POWER   = 0.65
WEIGHT_CC_POWER   = 0.30
WEIGHT_CLIP_MAX   = 200_000.0

LEFT_K,  LEFT_BOOST  = 6,  2.0
RIGHT_K, RIGHT_BOOST = 18, 80.0

ENDPOINT_PENALTY = 0.12
FARRIGHT_PENALTY = 0.18
FARRIGHT_K       = 8
EPS = 1e-300

def ccdf_logbins(x, q_lo=Q_LO, q_hi=Q_HI, nbins=NBINS_CCDF):
    x = np.asarray(x, dtype=np.float64)
    x = x[np.isfinite(x)]
    x = x[x > 0]
    x.sort()
    n = x.size
    if n < 10:
        return np.array([]), np.array([]), np.array([])

    xmin = float(np.quantile(x, q_lo))
    xmax = float(np.quantile(x, q_hi))
    if not np.isfinite(xmin) or not np.isfinite(xmax) or xmax <= xmin:
        return np.array([]), np.array([]), np.array([])

    edges = np.logspace(np.log10(xmin), np.log10(xmax), nbins + 1)
    g = edges[1:]
    idx = np.searchsorted(x, g, side="right")
    cc = (n - idx) / n
    cc = np.clip(cc, 1.0/(n+1.0), 1.0)
    tc = n * cc

    keep = np.r_[True, np.diff(idx) != 0]
    return g[keep], cc[keep], tc[keep]

def aggregate_by_logx_bins(x, y, tc, cc, nbins=NBINS_FIT):
    x = np.asarray(x, dtype=np.float64)
    y = np.asarray(y, dtype=np.float64)
    tc = np.asarray(tc, dtype=np.float64)
    cc = np.asarray(cc, dtype=np.float64)

    lx = np.log(x)
    lo, hi = float(lx.min()), float(lx.max())
    if not np.isfinite(lo) or not np.isfinite(hi) or hi <= lo:
        return x, y, np.ones_like(y)

    edges = np.linspace(lo, hi, nbins + 1)
    b = np.searchsorted(edges, lx, side="right") - 1
    b = np.clip(b, 0, nbins - 1)

    xs, ys, ws = [], [], []
    for bi in range(nbins):
        m = (b == bi)
        if not np.any(m):
            continue
        x_m = x[m]
        y_m = y[m]
        tc_m = tc[m]
        cc_m = cc[m]

        w = (np.clip(tc_m, 1.0, WEIGHT_CLIP_MAX) ** WEIGHT_TC_POWER) * (np.clip(cc_m, EPS, 1.0) ** (-WEIGHT_CC_POWER))
        wsum = float(np.sum(w))
        if wsum <= 0:
            continue

        x_bin = float(np.exp(np.sum(w * np.log(x_m)) / wsum))
        y_bin = float(np.sum(w * y_m) / wsum)

        xs.append(x_bin); ys.append(y_bin); ws.append(wsum)

    xs = np.asarray(xs); ys = np.asarray(ys); ws = np.asarray(ws)
    if xs.size:
        ws = ws.copy()
        ws[:min(LEFT_K, xs.size)] *= LEFT_BOOST
        ws[max(0, xs.size - RIGHT_K):] *= RIGHT_BOOST
        med = np.median(ws)
        if np.isfinite(med) and med > 0:
            ws /= med
        ws = np.clip(ws, 1e-12, 1e12)

    return xs, ys, ws

def weighted_lstsq(A, y, w):
    w = np.asarray(w, dtype=np.float64)
    w = np.clip(w, 1e-12, None)
    sw = np.sqrt(w)
    beta, *_ = np.linalg.lstsq(A * sw[:, None], y * sw, rcond=None)
    return beta

def weighted_r2(y, yhat, w):
    w = np.asarray(w, dtype=np.float64)
    w = np.clip(w, 1e-12, None)
    y = np.asarray(y, dtype=np.float64)
    yhat = np.asarray(yhat, dtype=np.float64)
    ybar = np.sum(w * y) / np.sum(w)
    ss_res = np.sum(w * (y - yhat) ** 2)
    ss_tot = np.sum(w * (y - ybar) ** 2)
    return 1 - ss_res / ss_tot if ss_tot > 0 else np.nan

def beff_from_ccdf(g, cc, window=BEFF_WINDOW):
    logx = np.log(g)
    logy = np.log(np.clip(cc, EPS, 1.0))
    w = window // 2
    slopes = np.full_like(logx, np.nan, dtype=np.float64)
    for i in range(len(logx)):
        j0 = max(0, i - w)
        j1 = min(len(logx), i + w + 1)
        Xw = logx[j0:j1]; Yw = logy[j0:j1]
        if len(Xw) < 2:
            continue
        Xc = Xw - Xw.mean()
        denom = np.sum(Xc * Xc)
        if denom <= 0:
            continue
        slopes[i] = np.sum(Xc * (Yw - Yw.mean())) / denom
    return -slopes

def model_ccdf(x, alpha, lam, beta, c0):
    x = np.asarray(x, dtype=np.float64)
    x = np.clip(x, 1e-300, np.inf)
    logx = np.log(x)
    if np.isfinite(lam) and lam > 0:
        logy = c0 - alpha * logx - (x / lam) ** beta
    else:
        logy = c0 - alpha * logx
    logy = np.minimum(logy, 0.0)
    logy = np.maximum(logy, -745.0)
    return np.exp(logy)

def beta_check_fit(x, y_check, w, lam, beta):
    if not (np.isfinite(lam) and lam > 0):
        A = np.stack([np.ones_like(x)], axis=1)
        coef = weighted_lstsq(A, y_check, w)
        yhat = A @ coef
        r2 = weighted_r2(y_check, yhat, w)
        return float(r2), float(np.nan), yhat

    u = (x / lam) ** beta
    A = np.stack([np.ones_like(u), u], axis=1)
    coef = weighted_lstsq(A, y_check, w)
    yhat = A @ coef
    r2 = weighted_r2(y_check, yhat, w)
    slope = float(coef[1])
    return float(r2), slope, yhat

def endpoint_logerr(g_fit, cc_fit, alpha, lam, beta, c0):
    iL = int(np.argmin(g_fit))
    iR = int(np.argmax(g_fit))
    xL, yL = float(g_fit[iL]), float(cc_fit[iL])
    xR, yR = float(g_fit[iR]), float(cc_fit[iR])
    mL = float(model_ccdf(np.array([xL]), alpha, lam, beta, c0)[0])
    mR = float(model_ccdf(np.array([xR]), alpha, lam, beta, c0)[0])
    yL = max(yL, EPS); yR = max(yR, EPS); mL = max(mL, EPS); mR = max(mR, EPS)
    eL = abs(np.log(mL / yL))
    eR = abs(np.log(mR / yR))
    return float(eL), float(eR)

def far_right_err(g, cc, alpha, lam, beta, c0, k=FARRIGHT_K):
    k = int(min(k, len(g)))
    if k <= 2:
        return 0.0
    xR = g[-k:]
    yR = np.clip(cc[-k:], EPS, 1.0)
    mR = np.clip(model_ccdf(xR, alpha, lam, beta, c0), EPS, 1.0)
    return float(np.mean(np.abs(np.log(mR) - np.log(yR))))

def fit_beta_tempered_auto_anchored(g, cc, tc):
    best = None
    for ccdf_max in CCDF_MAX_CANDIDATES:
        for min_tail in MIN_TAIL_COUNT_CANDIDATES:
            fitmask = (cc <= ccdf_max) & (tc >= min_tail)
            if int(fitmask.sum()) < max(80, NBINS_FIT // 2):
                continue

            g_fit = g[fitmask]
            cc_fit = cc[fitmask]
            tc_fit = tc[fitmask]

            x0 = float(g_fit.min())
            cc0 = float(cc_fit[np.argmin(g_fit)])
            if not (np.isfinite(x0) and np.isfinite(cc0) and x0 > 0 and cc0 > 0):
                continue

            Y = np.log(np.clip(cc_fit, EPS, 1.0))
            xA, YA, wA = aggregate_by_logx_bins(g_fit, Y, tc_fit, cc_fit, nbins=NBINS_FIT)
            if xA.size < MIN_FIT_BINS:
                continue

            X1 = np.log(xA)

            # power-law fallback
            Apl = np.stack([np.ones_like(X1), X1], axis=1)
            c0_hat, b1 = weighted_lstsq(Apl, YA, wA)
            alpha_pl = float(-b1)
            if np.isfinite(alpha_pl) and alpha_pl > 0:
                lam_pl = np.inf
                beta_pl = 1.0
                c0_pl = float(np.log(cc0) + alpha_pl * np.log(x0))
                Yhat_pl = c0_pl - alpha_pl * np.log(xA)
                r2_fit_pl = float(weighted_r2(YA, Yhat_pl, wA))
                y_check_pl = YA + alpha_pl * np.log(xA)
                r2_check_pl, slope_pl, _ = beta_check_fit(xA, y_check_pl, wA, lam_pl, beta_pl)

                eL, eR = endpoint_logerr(g_fit, cc_fit, alpha_pl, lam_pl, beta_pl, c0_pl)
                fr = far_right_err(g, cc, alpha_pl, lam_pl, beta_pl, c0_pl)

                score_pl = float(r2_check_pl) + 0.12 * float(r2_fit_pl) \
                           - ENDPOINT_PENALTY * (eL + eR) \
                           - FARRIGHT_PENALTY * fr \
                           - 0.02 * float(np.log10(ccdf_max / 0.01)) \
                           + 0.01 * float(np.log10(min_tail / 1000))

                cand = (score_pl, alpha_pl, lam_pl, beta_pl, c0_pl, r2_fit_pl, r2_check_pl, float(slope_pl),
                        fitmask, ccdf_max, min_tail, (xA, YA, wA), (eL, eR, fr))
                if best is None or cand[0] > best[0]:
                    best = cand

            # beta-tempered candidates
            for beta in BETA_GRID:
                beta = float(beta)
                x_ref = float(np.exp(np.mean(np.log(xA))))
                u = (xA / x_ref) ** beta

                A = np.stack([np.ones_like(X1), X1, u], axis=1)
                c0_hat, b1, b2 = weighted_lstsq(A, YA, wA)

                if not np.isfinite(b2) or b2 >= 0:
                    continue
                alpha = float(-b1)
                if not (np.isfinite(alpha) and alpha > 0):
                    continue

                Acoef = float(-b2)
                lam = float(x_ref / (Acoef ** (1.0 / beta)))
                if not (np.isfinite(lam) and lam > 0):
                    continue

                tmax = float((xA.max() / lam) ** beta)
                if tmax < CUTOFF_EPS_EFFECT:
                    continue

                c0 = float(np.log(cc0) + alpha * np.log(x0) + (x0 / lam) ** beta)

                Yhat = c0 - alpha * np.log(xA) - (xA / lam) ** beta
                r2_fit = float(weighted_r2(YA, Yhat, wA))

                y_check = YA + alpha * np.log(xA)
                r2_check, slope, _ = beta_check_fit(xA, y_check, wA, lam, beta)

                slope_pen = 0.12 * float(abs(slope + 1.0)) if np.isfinite(slope) else 0.25
                eL, eR = endpoint_logerr(g_fit, cc_fit, alpha, lam, beta, c0)
                fr = far_right_err(g, cc, alpha, lam, beta, c0)

                score = float(r2_check) + 0.12 * float(r2_fit) \
                        - slope_pen \
                        - ENDPOINT_PENALTY * (eL + eR) \
                        - FARRIGHT_PENALTY * fr \
                        - 0.02 * float(np.log10(ccdf_max / 0.01)) \
                        + 0.01 * float(np.log10(min_tail / 1000))

                cand = (score, alpha, lam, beta, c0, r2_fit, r2_check, float(slope),
                        fitmask, ccdf_max, min_tail, (xA, YA, wA), (eL, eR, fr))
                if best is None or cand[0] > best[0]:
                    best = cand

    if best is None:
        fitmask = (cc <= 0.02) & (tc >= 5000)
        return np.nan, np.nan, np.nan, np.nan, np.nan, fitmask, 0.02, 5000, np.nan, np.nan, (np.array([]), np.array([]), np.array([])), (np.nan, np.nan, np.nan)

    (_, alpha, lam, beta, c0, r2_fit, r2_check, slope, fitmask, ccdf_max, min_tail, agg, ends) = best
    return alpha, lam, beta, c0, r2_fit, fitmask, ccdf_max, min_tail, r2_check, slope, agg, ends

def plot_tempered_suite(x_abs, title):
    x_abs = np.asarray(x_abs, dtype=np.float64)
    x_abs = x_abs[np.isfinite(x_abs)]
    x_abs = x_abs[x_abs > 0]
    if x_abs.size < 200_000:
        print("[skip small]", title, "N=", x_abs.size)
        return

    g, cc, tc = ccdf_logbins(x_abs)
    if g.size < 200:
        print("[skip ccdf too small]", title, "N=", x_abs.size)
        return

    alpha, lam, beta, c0, r2_fit, fitmask, ccdf_max_used, min_tail_used, r2_check, slope_check, (xA, YA, wA), (eL, eR, fr) = \
        fit_beta_tempered_auto_anchored(g, cc, tc)

    x_min_fit = float(g[fitmask].min()) if np.any(fitmask) else float(g.min())
    cc_model = model_ccdf(g, alpha, lam, beta, c0)

    left = g < x_min_fit
    right = ~left

    plt.figure(figsize=(8.2, 5.4))
    plt.plot(g, cc, lw=2, label="CCDF (data)")
    plt.plot(g[left],  cc_model[left],  lw=2, alpha=0.25, label="model (left of fit window)")
    lam_str = "inf" if (not np.isfinite(lam) or lam <= 0) else f"{lam:.2g}"
    plt.plot(g[right], cc_model[right], lw=2, alpha=1.00,
             label=f"fit: alpha={alpha:.2f}, beta={beta:.2f}, lambda={lam_str}")
    plt.xscale("log"); plt.yscale("log")
    plt.xlabel("|g| (log)"); plt.ylabel("CCDF (log)")
    slope_str = "n/a" if not np.isfinite(slope_check) else f"{slope_check:.3f}"
    plt.title(
        f"{title}\n"
        f"fit: CCDF<= {ccdf_max_used}, tail_count>= {min_tail_used} | "
        f"R2_check={r2_check:.4f} | slope={slope_str} | end=({eL:.3f},{eR:.3f}) fr={fr:.3f}"
    )
    plt.grid(True, which="both", ls="--", alpha=0.5)
    plt.legend()
    if SHOW_PLOTS:
        plt.show()

    if xA.size >= 20 and np.isfinite(alpha):
        y_check = YA + alpha * np.log(xA)
        plt.figure(figsize=(8.2, 5.0))
        if np.isfinite(lam) and lam > 0:
            u = (xA / lam) ** beta
            o = np.argsort(u)
            u = u[o]; y_check_s = y_check[o]; w_s = wA[o]
            A = np.stack([np.ones_like(u), u], axis=1)
            coef = weighted_lstsq(A, y_check_s, w_s)
            yhat = A @ coef
            r2_lin = weighted_r2(y_check_s, yhat, w_s)
            plt.plot(u, y_check_s, lw=2, label="log(CCDF)+alpha log x (aggregated)")
            plt.plot(u, yhat,  lw=2, label=f"linear-in-u (weighted R2={r2_lin:.4f}, slope={coef[1]:.3f})")
            plt.xlabel(r"u = (|g|/lambda)^beta")
        else:
            yhat = np.full_like(y_check, np.average(y_check, weights=wA))
            r2_lin = weighted_r2(y_check, yhat, wA)
            plt.plot(xA, y_check, lw=2, label="log(CCDF)+alpha log x (aggregated)")
            plt.plot(xA, yhat,  lw=2, label=f"constant (weighted R2={r2_lin:.4f})")
            plt.xscale("log")
            plt.xlabel("|g| (log)")
        plt.ylabel("value")
        plt.title("beta-tempered check (aggregated fit points)")
        plt.grid(True, ls="--", alpha=0.5)
        plt.legend()
        if SHOW_PLOTS:
            plt.show()

    beff = beff_from_ccdf(g, cc, window=BEFF_WINDOW)
    bm = np.isfinite(beff) & fitmask
    plt.figure(figsize=(8.2, 5.0))
    plt.plot(g[bm], beff[bm], lw=2, label="B_eff(|g|) on fit window")
    if np.isfinite(lam) and lam > 0:
        plt.plot(g[bm], alpha + beta * (g[bm] / lam) ** beta, lw=2, label="model: alpha + beta*(x/lambda)^beta")
    else:
        plt.plot(g[bm], np.full_like(g[bm], alpha), lw=2, label="model: alpha (power law)")
    plt.xscale("log")
    plt.xlabel("|g| (log)")
    plt.ylabel("B_eff")
    plt.title(f"B_eff (fit window only), window={BEFF_WINDOW}")
    plt.grid(True, which="both", ls="--", alpha=0.5)
    plt.legend()
    if SHOW_PLOTS:
        plt.show()

# ============================================================
# Part 3: Save ALL figures into a single PDF
# ============================================================

@contextmanager
def _no_show():
    old = plt.show
    plt.show = lambda *a, **k: None
    try:
        yield
    finally:
        plt.show = old

def ordered_groups():
    out = []
    for phase in ["early", "mid", "late"]:
        for comp in ["q","k","v","proj"]:
            out.append(f"attn_{phase}_{comp}")
        for comp in ["fc","proj"]:
            out.append(f"mlp_{phase}_{comp}")
    return out

print("[info] writing PDF report:", pdf_path)
pages = 0
with PdfPages(str(pdf_path)) as pdf, _no_show():
    plt.close("all")
    groups = ordered_groups()

    for i, gname in enumerate(groups, 1):
        if gname not in grad_map:
            print(f"[{i:02d}] [skip missing] {gname}")
            continue

        title = f"{gname}: |g| (iter={ITER})"
        before = set(plt.get_fignums())
        plot_tempered_suite(grad_map[gname], title)
        after = set(plt.get_fignums())
        new_figs = sorted(list(after - before))

        for fn in new_figs:
            pdf.savefig(plt.figure(fn), bbox_inches="tight")
            pages += 1
        for fn in new_figs:
            plt.close(plt.figure(fn))

        print(f"[{i:02d}/{len(groups):02d}] saved {len(new_figs)} figs for {gname}")

    fig = plt.figure(figsize=(10.5, 2.6))
    plt.axis("off")
    plt.text(0.02, 0.65,
             f"beta-tempered GRAD report\niter={ITER}\nfigures={pages}\nPDF={pdf_path.name}\nNPZ={npz_path.name}",
             fontsize=12, family="monospace")
    pdf.savefig(fig, bbox_inches="tight")
    plt.close(fig)
    pages += 1

size = pdf_path.stat().st_size if pdf_path.exists() else 0
print(f"[done] pages={pages} bytes={size} file={pdf_path}")
assert pdf_path.exists() and size > 0, f"PDF not created or empty: {pdf_path}"
```

```python
# ============================================================
# FULL WORKING single-cell: model-selection fit (PL vs LN vs beta-tempered)
# + holdout validation + τ-report + SAME 3-figure layout + optional PDF export
#
# Fixes your crash:
#   - np.erfc does NOT exist -> use math.erfc vectorized
#   - (No SciPy needed)
#
# Assumes you ALREADY have:
#   - grad_map (dict) or noise_map (dict)
#   - ITER (int) (optional, used in titles)
# If you want PDF export, set PDF_PATH and EXPORT_PDF=True.
# ============================================================

import math
import numpy as np
import matplotlib.pyplot as plt
from contextlib import contextmanager
from matplotlib.backends.backend_pdf import PdfPages

# -----------------------------
# CONFIG (fit/plots)
# -----------------------------
Q_LO, Q_HI = 0.001, 0.99998
NBINS_CCDF = 800
NBINS_FIT  = 180
MIN_FIT_BINS = 60
BEFF_WINDOW = 21

CCDF_MAX_CANDIDATES = (0.2, 0.1, 0.05, 0.02, 0.01, 0.005, 0.002, 0.001)
MIN_TAIL_COUNT_CANDIDATES = (300, 500, 1000, 2000, 5000, 10000, 20000)

BETA_GRID = np.unique(np.concatenate([np.linspace(0.2, 1.0, 17), np.array([0.15, 0.12, 0.10])]))
CUTOFF_EPS_EFFECT = 0.02

WEIGHT_TC_POWER   = 0.65
WEIGHT_CC_POWER   = 0.30
WEIGHT_CLIP_MAX   = 200_000.0

LEFT_K,  LEFT_BOOST  = 6,  2.0
RIGHT_K, RIGHT_BOOST = 18, 40.0

EPS = 1e-300

# holdout/model selection
HOLDOUT_FRAC = 0.30
HOLDOUT_SEED = 777
USE_BIC_FOR_SELECTION = True
MIN_SIGMA_LN = 1e-3
PRINT_SUMMARY = True

# plotting + export
SHOW_PLOTS = True
EXPORT_PDF = True
PDF_PATH = "beta_tempered_model_selection_report__ALL.pdf"

# -----------------------------
# helpers
# -----------------------------
def ordered_groups():
    out = []
    for phase in ["early", "mid", "late"]:
        for comp in ["q","k","v","proj"]:
            out.append(f"attn_{phase}_{comp}")
        for comp in ["fc","proj"]:
            out.append(f"mlp_{phase}_{comp}")
    return out

@contextmanager
def _no_show():
    old = plt.show
    plt.show = lambda *a, **k: None
    try:
        yield
    finally:
        plt.show = old

def _ccdf_logbins(x, q_lo=Q_LO, q_hi=Q_HI, nbins=NBINS_CCDF):
    x = np.asarray(x, dtype=np.float64)
    x = x[np.isfinite(x)]
    x = x[x > 0]
    x.sort()
    n = x.size
    if n < 10:
        return np.array([]), np.array([]), np.array([])

    xmin = float(np.quantile(x, q_lo))
    xmax = float(np.quantile(x, q_hi))
    if not np.isfinite(xmin) or not np.isfinite(xmax) or xmax <= xmin:
        return np.array([]), np.array([]), np.array([])

    edges = np.logspace(np.log10(xmin), np.log10(xmax), nbins + 1)
    g = edges[1:]
    idx = np.searchsorted(x, g, side="right")
    cc = (n - idx) / n
    cc = np.clip(cc, 1.0/(n+1.0), 1.0)
    tc = n * cc

    keep = np.r_[True, np.diff(idx) != 0]
    return g[keep], cc[keep], tc[keep]

def _aggregate_by_logx_bins(x, ylog, tc, cc, nbins=NBINS_FIT):
    x = np.asarray(x, dtype=np.float64)
    ylog = np.asarray(ylog, dtype=np.float64)
    tc = np.asarray(tc, dtype=np.float64)
    cc = np.asarray(cc, dtype=np.float64)

    lx = np.log(x)
    lo, hi = float(lx.min()), float(lx.max())
    if not np.isfinite(lo) or not np.isfinite(hi) or hi <= lo:
        return x, ylog, np.ones_like(ylog)

    edges = np.linspace(lo, hi, nbins + 1)
    b = np.searchsorted(edges, lx, side="right") - 1
    b = np.clip(b, 0, nbins - 1)

    xs, ys, ws = [], [], []
    for bi in range(nbins):
        m = (b == bi)
        if not np.any(m):
            continue
        x_m = x[m]
        y_m = ylog[m]
        tc_m = tc[m]
        cc_m = cc[m]

        w = (np.clip(tc_m, 1.0, WEIGHT_CLIP_MAX) ** WEIGHT_TC_POWER) * (np.clip(cc_m, EPS, 1.0) ** (-WEIGHT_CC_POWER))
        wsum = float(np.sum(w))
        if wsum <= 0:
            continue

        x_bin = float(np.exp(np.sum(w * np.log(x_m)) / wsum))
        y_bin = float(np.sum(w * y_m) / wsum)

        xs.append(x_bin); ys.append(y_bin); ws.append(wsum)

    xs = np.asarray(xs); ys = np.asarray(ys); ws = np.asarray(ws)
    if xs.size:
        ws = ws.copy()
        ws[:min(LEFT_K, xs.size)] *= LEFT_BOOST
        ws[max(0, xs.size - RIGHT_K):] *= RIGHT_BOOST
        med = np.median(ws)
        if np.isfinite(med) and med > 0:
            ws /= med
        ws = np.clip(ws, 1e-12, 1e12)

    return xs, ys, ws

def _weighted_lstsq(A, y, w):
    w = np.asarray(w, dtype=np.float64)
    w = np.clip(w, 1e-12, None)
    sw = np.sqrt(w)
    beta, *_ = np.linalg.lstsq(A * sw[:, None], y * sw, rcond=None)
    return beta

def _aic_bic_from_sse(sse, n, k):
    sse = float(max(sse, 1e-300))
    n = int(n); k = int(k)
    aic = n * np.log(sse / n) + 2 * k
    bic = n * np.log(sse / n) + k * np.log(n)
    return float(aic), float(bic)

def _split_train_valid(n, seed=HOLDOUT_SEED, frac=HOLDOUT_FRAC):
    rng = np.random.default_rng(seed)
    idx = np.arange(n)
    rng.shuffle(idx)
    nv = max(1, int(np.floor(frac * n)))
    vidx = idx[:nv]
    tidx = idx[nv:]
    if tidx.size < 10:
        tidx = idx[nv//2:]
        vidx = idx[:nv//2]
    return tidx, vidx

def _beff_from_ccdf(g, cc, window=BEFF_WINDOW):
    logx = np.log(g)
    logy = np.log(np.clip(cc, EPS, 1.0))
    w = window // 2
    slopes = np.full_like(logx, np.nan, dtype=np.float64)
    for i in range(len(logx)):
        j0 = max(0, i - w)
        j1 = min(len(logx), i + w + 1)
        Xw = logx[j0:j1]; Yw = logy[j0:j1]
        if len(Xw) < 2:
            continue
        Xc = Xw - Xw.mean()
        denom = np.sum(Xc * Xc)
        if denom <= 0:
            continue
        slopes[i] = np.sum(Xc * (Yw - Yw.mean())) / denom
    return -slopes

# -----------------------------
# model definitions (log domain)
# -----------------------------
def _log_ccdf_powerlaw(x, alpha, c0):
    x = np.asarray(x, dtype=np.float64)
    return c0 - alpha * np.log(np.clip(x, 1e-300, np.inf))

def _log_ccdf_tempered(x, alpha, lam, beta, c0):
    x = np.asarray(x, dtype=np.float64)
    x = np.clip(x, 1e-300, np.inf)
    if not (np.isfinite(lam) and lam > 0):
        return _log_ccdf_powerlaw(x, alpha, c0)
    return c0 - alpha * np.log(x) - (x / lam) ** beta

# FIX: numpy has no erfc -> math.erfc vectorized
_erfc = np.vectorize(math.erfc, otypes=[np.float64])

def _ccdf_lognormal(x, mu, sigma):
    x = np.asarray(x, dtype=np.float64)
    x = np.clip(x, 1e-300, np.inf)
    sigma = float(max(sigma, MIN_SIGMA_LN))
    z = (np.log(x) - mu) / (sigma * np.sqrt(2.0))
    return 0.5 * _erfc(z)

def _log_ccdf_lognormal(x, mu, sigma):
    return np.log(np.clip(_ccdf_lognormal(x, mu, sigma), EPS, 1.0))

def _model_params_count(model_name):
    return {"powerlaw": 2, "lognormal": 2, "tempered": 4}[model_name]

def _eval_sse(x, ylog, w, fit):
    m = fit["model"]
    if m == "powerlaw":
        yhat = _log_ccdf_powerlaw(x, fit["alpha"], fit["c0"])
    elif m == "tempered":
        yhat = _log_ccdf_tempered(x, fit["alpha"], fit["lam"], fit["beta"], fit["c0"])
    elif m == "lognormal":
        yhat = _log_ccdf_lognormal(x, fit["mu"], fit["sigma"])
    else:
        raise ValueError("unknown model")
    sse = float(np.sum(w * (ylog - yhat) ** 2))
    return sse, yhat

# -----------------------------
# fit routines (weighted SSE in log CCDF domain)
# -----------------------------
def _fit_powerlaw(x, ylog, w):
    X1 = np.log(x)
    A = np.stack([np.ones_like(X1), X1], axis=1)
    c0_hat, b1 = _weighted_lstsq(A, ylog, w)
    return dict(model="powerlaw", alpha=float(-b1), c0=float(c0_hat))

def _fit_tempered_grid(x, ylog, w):
    best = None
    X1 = np.log(x)
    x_ref = float(np.exp(np.mean(X1)))
    for beta in BETA_GRID:
        beta = float(beta)
        u = (x / x_ref) ** beta
        A = np.stack([np.ones_like(X1), X1, u], axis=1)
        c0_hat, b1, b2 = _weighted_lstsq(A, ylog, w)
        if not np.isfinite(b2) or b2 >= 0:
            continue
        alpha = float(-b1)
        if not (np.isfinite(alpha) and alpha > 0):
            continue
        Acoef = float(-b2)
        lam = float(x_ref / (Acoef ** (1.0 / beta)))
        if not (np.isfinite(lam) and lam > 0):
            continue
        tmax = float((x.max() / lam) ** beta)
        if tmax < CUTOFF_EPS_EFFECT:
            continue

        c0 = float(c0_hat)  # no anchoring -> better identifiability & honest holdout
        yhat = _log_ccdf_tempered(x, alpha, lam, beta, c0)
        sse = float(np.sum(w * (ylog - yhat) ** 2))
        if best is None or sse < best["sse_train"]:
            best = dict(model="tempered", alpha=alpha, beta=beta, lam=lam, c0=c0, sse_train=sse)
    return best

def _fit_lognormal(x, ylog, w):
    lx = np.log(x)
    sig_grid = np.linspace(0.2, 3.0, 60)
    best = None
    mu0 = float(np.sum(w * lx) / np.sum(w))
    for sigma in sig_grid:
        sigma = float(sigma)
        mu_grid = np.linspace(mu0 - 3*sigma, mu0 + 3*sigma, 61)
        for mu in mu_grid:
            yhat = _log_ccdf_lognormal(x, float(mu), sigma)
            sse = float(np.sum(w * (ylog - yhat) ** 2))
            if best is None or sse < best["sse_train"]:
                best = dict(model="lognormal", mu=float(mu), sigma=float(sigma), sse_train=sse)
    return best

# -----------------------------
# select window + fit models + choose by AIC/BIC + holdout
# -----------------------------
def fit_models_auto(g, cc, tc):
    best_global = None

    for ccdf_max in CCDF_MAX_CANDIDATES:
        for min_tail in MIN_TAIL_COUNT_CANDIDATES:
            fitmask = (cc <= ccdf_max) & (tc >= min_tail)
            if int(fitmask.sum()) < max(80, NBINS_FIT // 2):
                continue

            g_fit = g[fitmask]
            cc_fit = cc[fitmask]
            tc_fit = tc[fitmask]
            ylog = np.log(np.clip(cc_fit, EPS, 1.0))

            xA, yA, wA = _aggregate_by_logx_bins(g_fit, ylog, tc_fit, cc_fit, nbins=NBINS_FIT)
            if xA.size < MIN_FIT_BINS:
                continue

            tidx, vidx = _split_train_valid(xA.size)
            xT, yT, wT = xA[tidx], yA[tidx], wA[tidx]
            xV, yV, wV = xA[vidx], yA[vidx], wA[vidx]

            f_pl = _fit_powerlaw(xT, yT, wT)
            f_ln = _fit_lognormal(xT, yT, wT)
            f_tp = _fit_tempered_grid(xT, yT, wT)

            fits = [f_pl, f_ln] + ([f_tp] if f_tp is not None else [])

            scored = []
            for f in fits:
                sseT, _ = _eval_sse(xT, yT, wT, f)
                sseV, _ = _eval_sse(xV, yV, wV, f)
                k = _model_params_count(f["model"])
                aic, bic = _aic_bic_from_sse(sseT, n=len(xT), k=k)
                score = bic if USE_BIC_FOR_SELECTION else aic
                scored.append(dict(fit=f, sse_train=sseT, sse_valid=sseV, aic=aic, bic=bic, score=score))

            scored.sort(key=lambda d: d["score"])
            chosen = scored[0]

            # τ visibility if tempered
            tau = np.nan
            ident = ""
            if chosen["fit"]["model"] == "tempered":
                lam = chosen["fit"]["lam"]
                beta = chosen["fit"]["beta"]
                xmax = float(xA.max())
                if np.isfinite(lam) and lam > 0:
                    tau = float((xmax / lam) ** beta)
                if (beta < 0.2) or (not np.isfinite(tau)) or (tau < 0.2) or (tau > 50):
                    ident = " (λ weakly identifiable)"
            chosen.update(dict(
                fitmask=fitmask,
                ccdf_max=float(ccdf_max),
                min_tail=int(min_tail),
                agg=(xA, yA, wA),
                all_scored=scored,
                tau=tau,
                ident_flag=ident
            ))

            key = (chosen["score"], chosen["sse_valid"])
            if best_global is None or key < best_global["key"]:
                best_global = dict(key=key, chosen=chosen)

    return None if best_global is None else best_global["chosen"]

# -----------------------------
# plotting
# -----------------------------
def plot_suite_with_model_selection(x_abs, title):
    x_abs = np.asarray(x_abs, dtype=np.float64)
    x_abs = x_abs[np.isfinite(x_abs)]
    x_abs = x_abs[x_abs > 0]
    if x_abs.size < 200_000:
        print("[skip small]", title, "N=", x_abs.size)
        return []

    g, cc, tc = _ccdf_logbins(x_abs)
    if g.size < 200:
        print("[skip ccdf too small]", title, "N=", x_abs.size)
        return []

    sel = fit_models_auto(g, cc, tc)
    if sel is None:
        print("[fit failed]", title)
        return []

    chosen = sel["fit"]
    scored = sel["all_scored"]
    fitmask = sel["fitmask"]

    # model curve on full grid
    if chosen["model"] == "powerlaw":
        cc_model = np.exp(np.clip(_log_ccdf_powerlaw(g, chosen["alpha"], chosen["c0"]), -745, 0.0))
    elif chosen["model"] == "lognormal":
        cc_model = np.exp(np.clip(_log_ccdf_lognormal(g, chosen["mu"], chosen["sigma"]), -745, 0.0))
    else:
        cc_model = np.exp(np.clip(_log_ccdf_tempered(g, chosen["alpha"], chosen["lam"], chosen["beta"], chosen["c0"]), -745, 0.0))

    # tail zoom bounds
    x_min_fit = float(g[fitmask].min()) if np.any(fitmask) else float(g.min())
    x_max_fit = float(g[fitmask].max()) if np.any(fitmask) else float(g.max())
    y_top = min(1.0, sel["ccdf_max"] * 1.2)
    y_bot = max(1e-8, sel["ccdf_max"] / 50.0)
    x_left = x_min_fit / 3.0
    x_right = x_max_fit * 1.2

    def _fmt_fit(f):
        if f["model"] == "powerlaw":
            return f"PL(α={f['alpha']:.2f})"
        if f["model"] == "lognormal":
            return f"LN(σ={f['sigma']:.2f})"
        tau = sel["tau"]
        tau_str = "nan" if not np.isfinite(tau) else f"{tau:.2g}"
        return f"TP(α={f['alpha']:.2f}, β={f['beta']:.2f}, τ={tau_str})"

    # ---- (1) CCDF + chosen model ----
    fig1 = plt.figure(figsize=(8.2, 5.4))
    plt.plot(g, cc, lw=2, label="CCDF (data)")
    plt.plot(g, cc_model, lw=2, alpha=0.95, label=f"chosen: {_fmt_fit(chosen)}{sel['ident_flag']}")
    plt.xscale("log"); plt.yscale("log")
    plt.xlim(x_left, x_right); plt.ylim(y_bot, y_top)
    plt.xlabel("x (log)"); plt.ylabel("CCDF (log)")
    top2 = ", ".join([_fmt_fit(d["fit"]) + f" BIC={d['bic']:.1f}" for d in scored[:2]])
    plt.title(f"{title}\nwindow: CCDF<= {sel['ccdf_max']}, tail_count>= {sel['min_tail']} | {top2}")
    plt.grid(True, which="both", ls="--", alpha=0.25)
    plt.legend()
    if SHOW_PLOTS:
        plt.show()

    # ---- (2) holdout diagnostic ----
    xA, yA, wA = sel["agg"]
    tidx, vidx = _split_train_valid(xA.size)
    xT, yT, wT = xA[tidx], yA[tidx], wA[tidx]
    xV, yV, wV = xA[vidx], yA[vidx], wA[vidx]

    fig2 = plt.figure(figsize=(8.2, 5.0))
    if chosen["model"] == "tempered":
        alpha = chosen["alpha"]; lam = chosen["lam"]; beta = chosen["beta"]; c0 = chosen["c0"]
        ycheckT = yT + alpha * np.log(xT)
        ycheckV = yV + alpha * np.log(xV)
        uT = (xT / lam) ** beta
        uV = (xV / lam) ** beta
        A = np.stack([np.ones_like(uT), uT], axis=1)
        a_hat, b_hat = _weighted_lstsq(A, ycheckT, wT)
        yhatT = a_hat + b_hat * uT
        yhatV = a_hat + b_hat * uV
        sseT = float(np.sum(wT * (ycheckT - yhatT) ** 2))
        sseV = float(np.sum(wV * (ycheckV - yhatV) ** 2))
        plt.plot(uT, ycheckT, ".", alpha=0.6, label=f"train (SSE={sseT:.3g})")
        plt.plot(uV, ycheckV, ".", alpha=0.6, label=f"valid (SSE={sseV:.3g})")
        uu = np.linspace(min(uT.min(), uV.min()), max(uT.max(), uV.max()), 200)
        plt.plot(uu, a_hat + b_hat*uu, lw=2, label=f"train fit slope={b_hat:.3f} (ideal -1)")
        plt.xlabel(r"u=(x/λ)^β"); plt.ylabel(r"log CCDF + α log x")
        plt.title("Tempered check with holdout")
    else:
        # predicted vs observed on holdout
        sseT, yhatT = _eval_sse(xT, yT, wT, chosen)
        sseV, yhatV = _eval_sse(xV, yV, wV, chosen)
        plt.plot(yhatT, yT, ".", alpha=0.6, label=f"train (SSE={sseT:.3g})")
        plt.plot(yhatV, yV, ".", alpha=0.6, label=f"valid (SSE={sseV:.3g})")
        lim0 = float(min(yT.min(), yV.min(), yhatT.min(), yhatV.min()))
        lim1 = float(max(yT.max(), yV.max(), yhatT.max(), yhatV.max()))
        plt.plot([lim0, lim1], [lim0, lim1], lw=2, label="y=x")
        plt.xlabel("pred log CCDF"); plt.ylabel("obs log CCDF")
        plt.title(f"{chosen['model'].capitalize()} holdout: pred vs obs")
    plt.grid(True, ls="--", alpha=0.25)
    plt.legend()
    if SHOW_PLOTS:
        plt.show()

    # ---- (3) B_eff + model ----
    fig3 = plt.figure(figsize=(8.2, 5.0))
    beff = _beff_from_ccdf(g, cc, window=BEFF_WINDOW)
    bm = np.isfinite(beff) & fitmask
    plt.plot(g[bm], beff[bm], lw=2, label="B_eff(x) on fit window")
    if chosen["model"] == "powerlaw":
        plt.plot(g[bm], np.full_like(g[bm], chosen["alpha"]), lw=2, label="model: α (const)")
    elif chosen["model"] == "tempered":
        alpha = chosen["alpha"]; lam = chosen["lam"]; beta = chosen["beta"]
        plt.plot(g[bm], alpha + beta * (g[bm] / lam) ** beta, lw=2, label="model: α + β*(x/λ)^β")
    else:
        lg = np.log(g[bm])
        ly = _log_ccdf_lognormal(g[bm], chosen["mu"], chosen["sigma"])
        d = np.gradient(ly, lg)
        plt.plot(g[bm], -d, lw=2, label="model: B_eff (num, LN)")
    plt.xscale("log")
    plt.xlabel("x (log)"); plt.ylabel("B_eff(x)")
    plt.title(f"B_eff (fit window), chosen={chosen['model']}")
    plt.grid(True, which="both", ls="--", alpha=0.25)
    plt.legend()
    if SHOW_PLOTS:
        plt.show()

    if PRINT_SUMMARY:
        print(f"\n[{title}] window: CCDF<= {sel['ccdf_max']}, tail_count>= {sel['min_tail']}, bins={xA.size}")
        for j, d in enumerate(scored[:3], 1):
            f = d["fit"]
            tau = "—"
            if f["model"] == "tempered":
                tau = f"{(xA.max()/f['lam'])**f['beta']:.3g}" if (np.isfinite(f['lam']) and f['lam']>0) else "nan"
            print(f"  #{j}: {f['model']:9s}  valid_SSE={d['sse_valid']:.3g}  "
                  f"AIC={d['aic']:.1f}  BIC={d['bic']:.1f}  τ={tau}")

    return [fig1, fig2, fig3]

# -----------------------------
# RUN (choose one map below)
# -----------------------------
# Pick which dataset you want to plot:
#   - grad_map (|g|)
#   - noise_map (|g_a - g_b|)
#   - delta_map (|g_t - g_{t-1}|) if you still have it
DATA_MAP = None
if "grad_map" in globals() and isinstance(grad_map, dict) and len(grad_map) > 0:
    DATA_MAP = grad_map
    _label = "|g|"
elif "noise_map" in globals() and isinstance(noise_map, dict) and len(noise_map) > 0:
    DATA_MAP = noise_map
    _label = "|g_a-g_b|"
elif "delta_map" in globals() and isinstance(delta_map, dict) and len(delta_map) > 0:
    DATA_MAP = delta_map
    _label = "|g_t-g_{t-1}|"
else:
    raise RuntimeError("No grad_map/noise_map/delta_map found in globals(). Load or compute one of them first.")

print(f"[info] plotting map='{_label}'  groups={len(DATA_MAP)}  EXPORT_PDF={EXPORT_PDF}")

if EXPORT_PDF:
    pages = 0
    with PdfPages(PDF_PATH) as pdf, _no_show():
        plt.close("all")
        for i, gname in enumerate(ordered_groups(), 1):
            if gname not in DATA_MAP:
                print(f"[{i:02d}] skip missing {gname}")
                continue
            figs = plot_suite_with_model_selection(DATA_MAP[gname], f"{gname}: {_label} (iter={globals().get('ITER','?')})")
            for fig in figs:
                pdf.savefig(fig, bbox_inches="tight")
                pages += 1
            for fig in figs:
                plt.close(fig)
            print(f"[{i:02d}/{len(ordered_groups()):02d}] saved {len(figs)} figs for {gname}")
    print(f"[done] PDF pages={pages} -> {PDF_PATH}")
else:
    for gname in ordered_groups():
        if gname in DATA_MAP:
            plot_suite_with_model_selection(DATA_MAP[gname], f"{gname}: {_label} (iter={globals().get('ITER','?')})")
```

```python
# ============================================================
# FINAL reviewer-proof plotting cell (MAIN+APP PDFs)
# ASCII-safe labels (no missing λ/β glyphs), no legend truncation,
# explicit definitions on every page.
#
# Outputs:
#   - paper_main_figures.pdf
#   - paper_appendix_tail_audit.pdf
#
# Requires in memory one of:
#   grad_map / noise_map / delta_map
# ============================================================

import math
import numpy as np
import matplotlib.pyplot as plt
from contextlib import contextmanager
from matplotlib.backends.backend_pdf import PdfPages

# -----------------------------
# USER SETTINGS
# -----------------------------
SHOW_PLOTS = False
EXPORT_PDFS = True

MAIN_PDF = "paper_main_figures.pdf"
APP_PDF  = "paper_appendix_tail_audit.pdf"

FORCE_MAP = None          # "grad_map" / "noise_map" / "delta_map" / None
MAIN_N_REP = 6            # 6 => 2x3 grid

# -----------------------------
# FIT SETTINGS
# -----------------------------
Q_LO, Q_HI = 0.001, 0.99998
NBINS_CCDF = 800
NBINS_FIT  = 180
MIN_FIT_BINS = 60
BEFF_WINDOW = 21

CCDF_MAX_CANDIDATES = (0.2, 0.1, 0.05, 0.02, 0.01, 0.005, 0.002, 0.001)
MIN_TAIL_COUNT_CANDIDATES = (300, 500, 1000, 2000, 5000, 10000, 20000)

BETA_GRID = np.unique(np.concatenate([np.linspace(0.2, 1.0, 17), np.array([0.15, 0.12, 0.10])]))
CUTOFF_EPS_EFFECT = 0.02

WEIGHT_TC_POWER   = 0.65
WEIGHT_CC_POWER   = 0.30
WEIGHT_CLIP_MAX   = 200_000.0

LEFT_K,  LEFT_BOOST  = 6,  2.0
RIGHT_K, RIGHT_BOOST = 18, 40.0

EPS = 1e-300

# model selection / holdout
HOLDOUT_FRAC = 0.30
HOLDOUT_SEED = 777
USE_BIC_FOR_SELECTION = True
MIN_SIGMA_LN = 1e-3

# -----------------------------
# UTILITIES
# -----------------------------
@contextmanager
def _no_show():
    old = plt.show
    plt.show = lambda *a, **k: None
    try:
        yield
    finally:
        plt.show = old

def ordered_groups():
    out = []
    for phase in ["early", "mid", "late"]:
        for comp in ["q","k","v","proj"]:
            out.append(f"attn_{phase}_{comp}")
        for comp in ["fc","proj"]:
            out.append(f"mlp_{phase}_{comp}")
    return out

def _ccdf_logbins(x, q_lo=Q_LO, q_hi=Q_HI, nbins=NBINS_CCDF):
    x = np.asarray(x, dtype=np.float64)
    x = x[np.isfinite(x)]
    x = x[x > 0]
    x.sort()
    n = x.size
    if n < 10:
        return np.array([]), np.array([]), np.array([])

    xmin = float(np.quantile(x, q_lo))
    xmax = float(np.quantile(x, q_hi))
    if not np.isfinite(xmin) or not np.isfinite(xmax) or xmax <= xmin:
        return np.array([]), np.array([]), np.array([])

    edges = np.logspace(np.log10(xmin), np.log10(xmax), nbins + 1)
    g = edges[1:]
    idx = np.searchsorted(x, g, side="right")
    cc = (n - idx) / n
    cc = np.clip(cc, 1.0/(n+1.0), 1.0)
    tc = n * cc

    keep = np.r_[True, np.diff(idx) != 0]
    return g[keep], cc[keep], tc[keep]

def _aggregate_by_logx_bins(x, ylog, tc, cc, nbins=NBINS_FIT):
    x = np.asarray(x, dtype=np.float64)
    ylog = np.asarray(ylog, dtype=np.float64)
    tc = np.asarray(tc, dtype=np.float64)
    cc = np.asarray(cc, dtype=np.float64)

    lx = np.log(x)
    lo, hi = float(lx.min()), float(lx.max())
    if not np.isfinite(lo) or not np.isfinite(hi) or hi <= lo:
        return x, ylog, np.ones_like(ylog)

    edges = np.linspace(lo, hi, nbins + 1)
    b = np.searchsorted(edges, lx, side="right") - 1
    b = np.clip(b, 0, nbins - 1)

    xs, ys, ws = [], [], []
    for bi in range(nbins):
        m = (b == bi)
        if not np.any(m):
            continue
        x_m = x[m]
        y_m = ylog[m]
        tc_m = tc[m]
        cc_m = cc[m]

        w = (np.clip(tc_m, 1.0, WEIGHT_CLIP_MAX) ** WEIGHT_TC_POWER) * (np.clip(cc_m, EPS, 1.0) ** (-WEIGHT_CC_POWER))
        wsum = float(np.sum(w))
        if wsum <= 0:
            continue

        x_bin = float(np.exp(np.sum(w * np.log(x_m)) / wsum))
        y_bin = float(np.sum(w * y_m) / wsum)

        xs.append(x_bin); ys.append(y_bin); ws.append(wsum)

    xs = np.asarray(xs); ys = np.asarray(ys); ws = np.asarray(ws)
    if xs.size:
        ws = ws.copy()
        ws[:min(LEFT_K, xs.size)] *= LEFT_BOOST
        ws[max(0, xs.size - RIGHT_K):] *= RIGHT_BOOST
        med = np.median(ws)
        if np.isfinite(med) and med > 0:
            ws /= med
        ws = np.clip(ws, 1e-12, 1e12)

    return xs, ys, ws

def _weighted_lstsq(A, y, w):
    w = np.asarray(w, dtype=np.float64)
    w = np.clip(w, 1e-12, None)
    sw = np.sqrt(w)
    beta, *_ = np.linalg.lstsq(A * sw[:, None], y * sw, rcond=None)
    return beta

def _aic_bic_from_sse(sse, n, k):
    sse = float(max(sse, 1e-300))
    n = int(n); k = int(k)
    aic = n * np.log(sse / n) + 2 * k
    bic = n * np.log(sse / n) + k * np.log(n)
    return float(aic), float(bic)

def _split_train_valid(n, seed=HOLDOUT_SEED, frac=HOLDOUT_FRAC):
    rng = np.random.default_rng(seed)
    idx = np.arange(n)
    rng.shuffle(idx)
    nv = max(1, int(np.floor(frac * n)))
    vidx = idx[:nv]
    tidx = idx[nv:]
    if tidx.size < 10:
        tidx = idx[nv//2:]
        vidx = idx[:nv//2]
    return tidx, vidx

def _beff_from_ccdf(g, cc, window=BEFF_WINDOW):
    logx = np.log(g)
    logy = np.log(np.clip(cc, EPS, 1.0))
    w = window // 2
    slopes = np.full_like(logx, np.nan, dtype=np.float64)
    for i in range(len(logx)):
        j0 = max(0, i - w)
        j1 = min(len(logx), i + w + 1)
        Xw = logx[j0:j1]; Yw = logy[j0:j1]
        if len(Xw) < 2:
            continue
        Xc = Xw - Xw.mean()
        denom = np.sum(Xc * Xc)
        if denom <= 0:
            continue
        slopes[i] = np.sum(Xc * (Yw - Yw.mean())) / denom
    return -slopes

# -----------------------------
# MODELS
# -----------------------------
def _log_ccdf_powerlaw(x, alpha, c0):
    x = np.asarray(x, dtype=np.float64)
    return c0 - alpha * np.log(np.clip(x, 1e-300, np.inf))

def _log_ccdf_tempered(x, alpha, lam, beta, c0):
    x = np.asarray(x, dtype=np.float64)
    x = np.clip(x, 1e-300, np.inf)
    if not (np.isfinite(lam) and lam > 0):
        return _log_ccdf_powerlaw(x, alpha, c0)
    return c0 - alpha * np.log(x) - (x / lam) ** beta

_erfc = np.vectorize(math.erfc, otypes=[np.float64])

def _ccdf_lognormal(x, mu, sigma):
    x = np.asarray(x, dtype=np.float64)
    x = np.clip(x, 1e-300, np.inf)
    sigma = float(max(sigma, MIN_SIGMA_LN))
    z = (np.log(x) - mu) / (sigma * np.sqrt(2.0))
    return 0.5 * _erfc(z)

def _log_ccdf_lognormal(x, mu, sigma):
    return np.log(np.clip(_ccdf_lognormal(x, mu, sigma), EPS, 1.0))

def _k_params(model):
    return {"powerlaw": 2, "lognormal": 2, "tempered": 4}[model]

def _eval_sse(x, ylog, w, fit):
    m = fit["model"]
    if m == "powerlaw":
        yhat = _log_ccdf_powerlaw(x, fit["alpha"], fit["c0"])
    elif m == "tempered":
        yhat = _log_ccdf_tempered(x, fit["alpha"], fit["lam"], fit["beta"], fit["c0"])
    else:
        yhat = _log_ccdf_lognormal(x, fit["mu"], fit["sigma"])
    sse = float(np.sum(w * (ylog - yhat) ** 2))
    return sse, yhat

# -----------------------------
# FITTERS
# -----------------------------
def _fit_powerlaw(x, ylog, w):
    X1 = np.log(x)
    A = np.stack([np.ones_like(X1), X1], axis=1)
    c0_hat, b1 = _weighted_lstsq(A, ylog, w)
    return dict(model="powerlaw", alpha=float(-b1), c0=float(c0_hat))

def _fit_tempered_grid(x, ylog, w):
    best = None
    X1 = np.log(x)
    x_ref = float(np.exp(np.mean(X1)))
    for beta in BETA_GRID:
        beta = float(beta)
        u = (x / x_ref) ** beta
        A = np.stack([np.ones_like(X1), X1, u], axis=1)
        c0_hat, b1, b2 = _weighted_lstsq(A, ylog, w)
        if not np.isfinite(b2) or b2 >= 0:
            continue
        alpha = float(-b1)
        if not (np.isfinite(alpha) and alpha > 0):
            continue
        Acoef = float(-b2)
        lam = float(x_ref / (Acoef ** (1.0 / beta)))
        if not (np.isfinite(lam) and lam > 0):
            continue
        tmax = float((x.max() / lam) ** beta)
        if tmax < CUTOFF_EPS_EFFECT:
            continue
        c0 = float(c0_hat)
        yhat = _log_ccdf_tempered(x, alpha, lam, beta, c0)
        sse = float(np.sum(w * (ylog - yhat) ** 2))
        if best is None or sse < best["sse_train"]:
            best = dict(model="tempered", alpha=alpha, beta=beta, lam=lam, c0=c0, sse_train=sse)
    return best

def _fit_lognormal(x, ylog, w):
    lx = np.log(x)
    sig_grid = np.linspace(0.2, 3.0, 60)
    best = None
    mu0 = float(np.sum(w * lx) / np.sum(w))
    for sigma in sig_grid:
        sigma = float(sigma)
        mu_grid = np.linspace(mu0 - 3*sigma, mu0 + 3*sigma, 61)
        for mu in mu_grid:
            yhat = _log_ccdf_lognormal(x, float(mu), sigma)
            sse = float(np.sum(w * (ylog - yhat) ** 2))
            if best is None or sse < best["sse_train"]:
                best = dict(model="lognormal", mu=float(mu), sigma=float(sigma), sse_train=sse)
    return best

# -----------------------------
# SELECTION
# -----------------------------
def fit_models_auto(g, cc, tc):
    best_global = None
    for ccdf_max in CCDF_MAX_CANDIDATES:
        for min_tail in MIN_TAIL_COUNT_CANDIDATES:
            fitmask = (cc <= ccdf_max) & (tc >= min_tail)
            if int(fitmask.sum()) < max(80, NBINS_FIT // 2):
                continue

            g_fit = g[fitmask]
            cc_fit = cc[fitmask]
            tc_fit = tc[fitmask]
            ylog = np.log(np.clip(cc_fit, EPS, 1.0))

            xA, yA, wA = _aggregate_by_logx_bins(g_fit, ylog, tc_fit, cc_fit, nbins=NBINS_FIT)
            if xA.size < MIN_FIT_BINS:
                continue

            tidx, vidx = _split_train_valid(xA.size)
            xT, yT, wT = xA[tidx], yA[tidx], wA[tidx]
            xV, yV, wV = xA[vidx], yA[vidx], wA[vidx]

            f_pl = _fit_powerlaw(xT, yT, wT)
            f_ln = _fit_lognormal(xT, yT, wT)
            f_tp = _fit_tempered_grid(xT, yT, wT)

            fits = [f_pl, f_ln] + ([f_tp] if f_tp is not None else [])

            scored = []
            for f in fits:
                sseT, _ = _eval_sse(xT, yT, wT, f)
                sseV, _ = _eval_sse(xV, yV, wV, f)
                aic, bic = _aic_bic_from_sse(sseT, n=len(xT), k=_k_params(f["model"]))
                scored.append(dict(fit=f, sse_train=sseT, sse_valid=sseV, aic=aic, bic=bic))
            scored.sort(key=lambda d: d["bic"] if USE_BIC_FOR_SELECTION else d["aic"])

            winner = scored[0]
            runner = scored[1] if len(scored) > 1 else None
            dBIC = (runner["bic"] - winner["bic"]) if runner is not None else np.nan

            tau = np.nan
            ident = ""
            if winner["fit"]["model"] == "tempered":
                lam = winner["fit"]["lam"]; beta = winner["fit"]["beta"]
                xmax = float(xA.max())
                if np.isfinite(lam) and lam > 0:
                    tau = float((xmax / lam) ** beta)  # tau = (x_max/lambda)^beta
                if (beta < 0.2) or (not np.isfinite(tau)) or (tau < 0.2) or (tau > 50):
                    ident = "lambda weakly identifiable"
            hold_ratio = float(winner["sse_valid"] / max(winner["sse_train"], 1e-300))

            key = (winner["bic"], winner["sse_valid"])
            pack = dict(
                fitmask=fitmask, ccdf_max=float(ccdf_max), min_tail=int(min_tail),
                agg=(xA, yA, wA),
                scored=scored, winner=winner, runner=runner,
                dBIC=float(dBIC) if np.isfinite(dBIC) else np.nan,
                tau=tau, ident=ident, hold_ratio=hold_ratio
            )
            if best_global is None or key < best_global["key"]:
                best_global = dict(key=key, sel=pack)

    return None if best_global is None else best_global["sel"]

# -----------------------------
# Plot helpers (ASCII-safe)
# -----------------------------
def _ccdf_model_on_grid(g, fit):
    if fit["model"] == "powerlaw":
        return np.exp(np.clip(_log_ccdf_powerlaw(g, fit["alpha"], fit["c0"]), -745, 0.0))
    if fit["model"] == "lognormal":
        return np.exp(np.clip(_log_ccdf_lognormal(g, fit["mu"], fit["sigma"]), -745, 0.0))
    return np.exp(np.clip(_log_ccdf_tempered(g, fit["alpha"], fit["lam"], fit["beta"], fit["c0"]), -745, 0.0))

def _tail_zoom_limits(g, fitmask, ccdf_max):
    x_min_fit = float(g[fitmask].min()) if np.any(fitmask) else float(g.min())
    x_max_fit = float(g[fitmask].max()) if np.any(fitmask) else float(g.max())
    y_top = min(1.0, ccdf_max * 1.2)
    y_bot = max(1e-9, ccdf_max / 50.0)
    x_left = x_min_fit / 3.0
    x_right = x_max_fit * 1.25
    return x_left, x_right, y_bot, y_top

def _legend_outside(ax):
    ax.legend(loc="center left", bbox_to_anchor=(1.02, 0.5), borderaxespad=0.0, fontsize=8, frameon=True)

def _fmt_fit_short(fit, tau=np.nan):
    m = fit["model"]
    if m == "powerlaw":
        return f"PL alpha={fit['alpha']:.2f}"
    if m == "lognormal":
        return f"LN sigma={fit['sigma']:.2f}"
    tau_str = "nan" if not np.isfinite(tau) else f"{tau:.2g}"
    return f"TP alpha={fit['alpha']:.2f}, beta={fit['beta']:.2f}, tau={tau_str}"

def _diag_box(sel):
    w = sel["winner"]["fit"]
    r = sel["runner"]["fit"] if sel["runner"] is not None else None
    lines = [
        f"winner: {w['model']}   runner: {(r['model'] if r else '—')}",
        f"BIC(w)={sel['winner']['bic']:.1f}  BIC(r)={(sel['runner']['bic'] if sel['runner'] else np.nan):.1f}  dBIC={sel['dBIC']:.1f}",
        f"holdout ratio (valid/train SSE)={sel['hold_ratio']:.2f}",
        f"window: CCDF<={sel['ccdf_max']}  tail_count>={sel['min_tail']}",
    ]
    if w["model"] == "tempered":
        tau_str = "nan" if not np.isfinite(sel["tau"]) else f"{sel['tau']:.3g}"
        lines.append(f"tau=(x_max/lambda)^beta={tau_str}  {sel['ident']}")
    return "\n".join(lines)

def fig_model_map(rows18, title, x_label, iter_tag):
    enc = {"powerlaw":0, "lognormal":1, "tempered":2}
    M = np.array([enc[r["model"]] for r in rows18], dtype=np.int64).reshape(3, 6)
    D = np.array([r["dBIC"] for r in rows18], dtype=np.float64).reshape(3, 6)

    fig = plt.figure(figsize=(9.6, 4.2))
    ax = plt.gca()
    ax.imshow(M, aspect="auto")

    for i in range(3):
        for j in range(6):
            mname = ["PL","LN","TP"][M[i, j]]
            dv = D[i, j]
            s = "—" if not np.isfinite(dv) else f"{dv:.0f}"
            ax.text(j, i, f"{mname}\nd{s}", ha="center", va="center", fontsize=8)

    ax.set_yticks([0,1,2]); ax.set_yticklabels(["early","mid","late"])
    ax.set_xticks(list(range(6)))
    ax.set_xticklabels(["attn-q","attn-k","attn-v","attn-proj","mlp-fc","mlp-proj"])
    ax.set_title(f"{title} (x={x_label}, iter={iter_tag})\n"
                 f"Each cell: winner model; d = dBIC = BIC(runner-up) - BIC(winner); larger d => more decisive.",
                 fontsize=11)

    from matplotlib.patches import Patch
    cmap = plt.cm.viridis
    leg = [Patch(label="PL: power law CCDF ~ x^{-alpha}", facecolor=cmap(0.15)),
           Patch(label="LN: lognormal CCDF", facecolor=cmap(0.50)),
           Patch(label="TP: tempered CCDF ~ x^{-alpha} * exp(-(x/lambda)^beta)", facecolor=cmap(0.85))]
    ax.legend(handles=leg, loc="upper right", fontsize=8, frameon=True)
    plt.tight_layout()
    return fig

def fig_beta_hist(rows, title, x_label, iter_tag):
    betas = [r["beta"] for r in rows if r["model"]=="tempered" and np.isfinite(r["beta"])]
    taus  = [r["tau"]  for r in rows if r["model"]=="tempered" and np.isfinite(r["tau"])]
    fig = plt.figure(figsize=(9.0, 3.6))
    ax = plt.gca()
    if len(betas) == 0:
        ax.text(0.1, 0.5, "No TP winners.", fontsize=12)
        ax.axis("off")
        return fig
    ax.hist(betas, bins=12)
    ax.set_xlabel("beta for TP winners")
    ax.set_ylabel("count")
    ax.grid(alpha=0.25)
    ax.set_title(
        f"{title} (x={x_label}, iter={iter_tag})\n"
        f"tau=(x_max/lambda)^beta measures cutoff visibility on the selected tail window. "
        f"N_TP={len(betas)}, median beta={np.median(betas):.2f}, median tau={np.median(taus):.2g}",
        fontsize=11
    )
    plt.tight_layout()
    return fig

def fig_ccdf(arr, gname, sel, x_label, iter_tag):
    arr = np.asarray(arr, dtype=np.float64)
    arr = arr[np.isfinite(arr)]
    arr = arr[arr > 0]
    g, cc, tc = _ccdf_logbins(arr)
    fitmask = sel["fitmask"]
    fit = sel["winner"]["fit"]
    cc_model = _ccdf_model_on_grid(g, fit)

    x_left, x_right, y_bot, y_top = _tail_zoom_limits(g, fitmask, sel["ccdf_max"])

    fig = plt.figure(figsize=(7.8, 4.8))
    ax = plt.gca()
    ax.plot(g, cc, lw=2, label="data CCDF")
    ax.plot(g, cc_model, lw=2, label=f"model: {_fmt_fit_short(fit, sel['tau'])}")
    ax.set_xscale("log"); ax.set_yscale("log")
    ax.set_xlim(x_left, x_right); ax.set_ylim(y_bot, y_top)
    ax.grid(True, which="both", ls="--", alpha=0.25)
    ax.set_xlabel(f"x = {x_label}")
    ax.set_ylabel("CCDF(x)")
    ax.set_title(f"{gname} (iter={iter_tag}) — tail zoom", fontsize=11)

    strip = "Selection: BIC on train bins + holdout SSE. dBIC=BIC(runner)-BIC(winner). tau=(x_max/lambda)^beta (TP only)."
    ax.text(0.01, -0.28, strip, transform=ax.transAxes, fontsize=8, va="top")

    _legend_outside(ax)
    plt.tight_layout()
    return fig

def fig_beff(arr, gname, sel, x_label, iter_tag):
    arr = np.asarray(arr, dtype=np.float64)
    arr = arr[np.isfinite(arr)]
    arr = arr[arr > 0]
    g, cc, tc = _ccdf_logbins(arr)
    fitmask = sel["fitmask"]
    fit = sel["winner"]["fit"]
    beff = _beff_from_ccdf(g, cc, window=BEFF_WINDOW)
    bm = np.isfinite(beff) & fitmask

    fig = plt.figure(figsize=(7.8, 4.6))
    ax = plt.gca()
    ax.plot(g[bm], beff[bm], lw=2, label="B_eff(data)")

    if fit["model"] == "powerlaw":
        ax.plot(g[bm], np.full_like(g[bm], fit["alpha"]), lw=2, label="PL: alpha (const)")
    elif fit["model"] == "tempered":
        ax.plot(g[bm], fit["alpha"] + fit["beta"] * (g[bm] / fit["lam"]) ** fit["beta"],
                lw=2, label="TP: alpha + beta*(x/lambda)^beta")
    else:
        lg = np.log(g[bm])
        ly = _log_ccdf_lognormal(g[bm], fit["mu"], fit["sigma"])
        d = np.gradient(ly, lg)
        ax.plot(g[bm], -d, lw=2, label="LN: numerical B_eff")

    ax.set_xscale("log")
    ax.grid(True, which="both", ls="--", alpha=0.25)
    ax.set_xlabel(f"x = {x_label}")
    ax.set_ylabel("B_eff(x) = - d log(CCDF) / d log(x)")
    ax.set_title(f"{gname} (iter={iter_tag}) — slope diagnostic", fontsize=11)

    box = _diag_box(sel)
    ax.text(1.02, 0.98, box, transform=ax.transAxes, fontsize=8, va="top",
            bbox=dict(boxstyle="round,pad=0.35", facecolor="white", alpha=0.95, edgecolor="0.6"))

    _legend_outside(ax)
    plt.tight_layout()
    return fig

# -----------------------------
# 0) Pick data map and label
# -----------------------------
DATA_MAP = None
LABEL = None

if FORCE_MAP is not None:
    if FORCE_MAP in globals():
        DATA_MAP = globals()[FORCE_MAP]
        LABEL = FORCE_MAP
    else:
        raise RuntimeError(f"FORCE_MAP='{FORCE_MAP}' not found in globals().")
else:
    for name in ["grad_map", "noise_map", "delta_map"]:
        if name in globals() and isinstance(globals()[name], dict) and len(globals()[name]) > 0:
            DATA_MAP = globals()[name]
            LABEL = name
            break

if DATA_MAP is None:
    raise RuntimeError("No grad_map/noise_map/delta_map found. Compute or load one first.")

if LABEL == "grad_map":
    X_LABEL = "|g| (stochastic gradient)"
elif LABEL == "noise_map":
    X_LABEL = "|g_a - g_b| (noise-only proxy)"
else:
    X_LABEL = "|g_t - g_{t-1}| (rolling increments proxy)"

ITER_TAG = globals().get("ITER", "?")
print(f"[info] Using {LABEL} with x={X_LABEL}, iter={ITER_TAG}")

# -----------------------------
# 1) Fit all groups
# -----------------------------
sel_by_group = {}
summary = []
for gname in ordered_groups():
    if gname not in DATA_MAP:
        continue
    arr = np.asarray(DATA_MAP[gname], dtype=np.float64)
    arr = arr[np.isfinite(arr)]
    arr = arr[arr > 0]
    if arr.size < 200_000:
        continue
    g, cc, tc = _ccdf_logbins(arr)
    sel = fit_models_auto(g, cc, tc)
    if sel is None:
        continue
    winner_model = sel["winner"]["fit"]["model"]
    row = dict(group=gname, model=winner_model, dBIC=sel["dBIC"], alpha=np.nan, beta=np.nan, sigma=np.nan, tau=np.nan)
    if winner_model == "powerlaw":
        row["alpha"] = sel["winner"]["fit"]["alpha"]
    elif winner_model == "lognormal":
        row["sigma"] = sel["winner"]["fit"]["sigma"]
    else:
        row["alpha"] = sel["winner"]["fit"]["alpha"]
        row["beta"]  = sel["winner"]["fit"]["beta"]
        row["tau"]   = sel["tau"]
    summary.append(row)
    sel_by_group[gname] = sel

assert len(summary) >= 10, "Too few groups fitted; check your map arrays."

# reorder to 18 for model map
row_by = {r["group"]: r for r in summary}
rows18 = [row_by[g] for g in ordered_groups() if g in row_by]
if len(rows18) < 18:
    rows18 = rows18 + [rows18[-1]] * (18 - len(rows18))

# reps: top dBIC per model, fill up
def _pick_reps(rows, n_total=MAIN_N_REP):
    reps = []
    for m in ["tempered", "lognormal", "powerlaw"]:
        cand = [r for r in rows if r["model"] == m and np.isfinite(r["dBIC"])]
        cand.sort(key=lambda r: -r["dBIC"])
        reps.extend(cand[: max(1, n_total//3)])
    if len(reps) < n_total:
        allc = [r for r in rows if np.isfinite(r["dBIC"])]
        allc.sort(key=lambda r: -r["dBIC"])
        seen = set([r["group"] for r in reps])
        for r in allc:
            if r["group"] not in seen:
                reps.append(r); seen.add(r["group"])
            if len(reps) >= n_total:
                break
    return reps[:n_total]

rep_groups = [r["group"] for r in _pick_reps(summary, MAIN_N_REP)]
print("[main] representative groups:", rep_groups)

# -----------------------------
# 2) MAIN figures
# -----------------------------
main_figs = []
main_figs.append(fig_model_map(rows18, "Tail model selection across blocks", X_LABEL, ITER_TAG))
main_figs.append(fig_beta_hist(rows18, "Beta distribution among TP winners", X_LABEL, ITER_TAG))

# rep CCDF grid
fig = plt.figure(figsize=(12.8, 7.2))
rows, cols = 2, 3
for i, gname in enumerate(rep_groups, 1):
    ax = plt.subplot(rows, cols, i)
    arr = np.asarray(DATA_MAP[gname], dtype=np.float64)
    arr = arr[np.isfinite(arr)]
    arr = arr[arr > 0]
    g, cc, tc = _ccdf_logbins(arr)
    sel = sel_by_group[gname]
    fit = sel["winner"]["fit"]
    cc_model = _ccdf_model_on_grid(g, fit)
    x_left, x_right, y_bot, y_top = _tail_zoom_limits(g, sel["fitmask"], sel["ccdf_max"])
    ax.plot(g, cc, lw=1.6)
    ax.plot(g, cc_model, lw=1.6)
    ax.set_xscale("log"); ax.set_yscale("log")
    ax.set_xlim(x_left, x_right); ax.set_ylim(y_bot, y_top)
    ax.grid(True, which="both", ls="--", alpha=0.2)
    ttl = f"{gname}\n{_fmt_fit_short(fit, sel['tau'])}"
    if sel["ident"]:
        ttl += f"  [{sel['ident']}]"
    ax.set_title(ttl, fontsize=9)
    if i > 3: ax.set_xlabel("x")
    if i % 3 == 1: ax.set_ylabel("CCDF")
plt.suptitle(
    f"Representative tail fits (data vs chosen model)\n"
    f"Selection: BIC on train bins + holdout SSE; dBIC=BIC(runner)-BIC(winner) | x={X_LABEL}, iter={ITER_TAG}",
    y=0.98, fontsize=12
)
plt.tight_layout(rect=[0,0,1,0.92])
main_figs.append(fig)

# B_eff example: most decisive tempered if exists else most decisive overall
cand = [r for r in summary if np.isfinite(r["dBIC"])]
cand.sort(key=lambda r: -r["dBIC"])
pick = next((r["group"] for r in cand if r["model"]=="tempered"), cand[0]["group"])
main_figs.append(fig_beff(DATA_MAP[pick], pick, sel_by_group[pick], X_LABEL, ITER_TAG))

# -----------------------------
# 3) APPENDIX figures
# -----------------------------
appendix_figs = []
appendix_figs.extend([main_figs[0], main_figs[1]])
for gname in ordered_groups():
    if gname not in sel_by_group:
        continue
    sel = sel_by_group[gname]
    appendix_figs.append(fig_ccdf(DATA_MAP[gname], gname, sel, X_LABEL, ITER_TAG))
    appendix_figs.append(fig_beff(DATA_MAP[gname], gname, sel, X_LABEL, ITER_TAG))

# -----------------------------
# 4) Export PDFs
# -----------------------------
if EXPORT_PDFS:
    with PdfPages(MAIN_PDF) as pdf, _no_show():
        for f in main_figs:
            pdf.savefig(f, bbox_inches="tight")
    with PdfPages(APP_PDF) as pdf, _no_show():
        for f in appendix_figs:
            pdf.savefig(f, bbox_inches="tight")
    print("[saved]", MAIN_PDF)
    print("[saved]", APP_PDF)

if not SHOW_PLOTS:
    for f in main_figs:
        plt.close(f)
    for f in appendix_figs:
        try: plt.close(f)
        except Exception: pass

print("[done] MAIN pages:", len(main_figs), "| APP pages:", len(appendix_figs))
```
